# <a id="tc1" name="tc1"></a>Text cell 1: Scalar inference as predictive coding

**Goal.** Build a predictive-coding / free-energy model of scalar inference with:

- a latent **world state** $s$: the proportion of a set that satisfies the predicate, as in the fraction of a jar of marbles that sank

- an observed **utterance** $y \in \{\textit{no},\textit{some},\textit{all}\}$

- a base prior over $s$ whose sharpness is under the modeller's control, and is swept in Text cell 6

Text cell 3 fixes each of these and states the notation the rest of the notebook keeps to: $s$ lies
in the open interval $(0,1)$ and is carried in logit coordinates on a quadrature grid; $y$ clamps a
lexical field through the exclusion set of its entry; the base prior is sampled on the same grid. The
letters here already follow that convention, so the observation is written $y$, since Text cell 3's
commitment 5 reserves $u$ for the utility level.

**Predictive-coding motivation.** Bogacz (2017) defines a log-joint quantity $F$ and performs gradient ascent on it to infer latent states (his Eq. (6), **p. 200**):

$$
F(\phi;y) = \ln p(\phi) + \ln p(y\mid \phi),
\qquad
\dot\phi \propto \frac{\partial F}{\partial \phi}.
$$

## <a id="tc2" name="tc2"></a>Text cell 2: High-level predictive-coding pseudocode (reference implementation plan)

This cell is **pseudocode**. It sketches the classical predictive-coding pattern:

1. Maintain **state units** (latent causes) and **error units** (prediction errors).

2. Run **inner-loop inference dynamics** until errors settle.

3. Run **outer-loop learning** (Hebbian-style) on synapses generating predictions.

**Connection to Bogacz (2017).**
- In the hierarchical predictive-coding network, state and error dynamics take the form (Bogacz Eqs. (53)–(54), **p. 206**):

$$
\dot\phi_i = -\varepsilon_i + h'(\phi_i)\,\Theta_{i-1}^\top\varepsilon_{i-1}
$$

$$
\dot\varepsilon_i = \phi_i - \Theta_i h(\phi_{i+1}) - \Sigma_i\varepsilon_i.
$$

- Learning the generative weights is Hebbian in the sense that updates are proportional to a product of an error term and an activity term (e.g., Bogacz Eq. (50), **p. 206**):

$$
\Delta\Theta \propto \varepsilon\,h(\phi)^\top.
$$

- For learning **noise/uncertainty** parameters with local plasticity, Bogacz introduces an auxiliary inhibitory interneuron $e_i$ and dynamics (Eqs. (59)–(60), **p. 207**) so that error nodes converge to precision-weighted errors (Bogacz Eq. (61), **p. 207**):

$$
\dot\varepsilon_i = \phi_i - g_i(\phi_{i+1}) - e_i,\qquad
\dot e_i = \Sigma_i\varepsilon_i - e_i.
$$

We will later implement a discrete analogue of uncertainty learning, where $w$ plays the role of a scalar precision controlling the prior over $s$.

<a id="pseudocode" name="pseudocode"></a>

```python
# PSEUDOCODE
# shapes: batch B, latent dim D2, sensory dim D1
#batch size is the number of independent scenarios/participants we are simlutating at time t
#dimension is the number of features in a given layer

phi2 = torch.zeros(B, D2)          # state units (layer 2)
eps1 = torch.zeros(B, D1)          # error at sensory level
eps2 = torch.zeros(B, D2)          # error at latent level (optional, if adding a prior)
e1   = torch.zeros(B, D1)          # interneuron for local Sigma(variance) learning (optional, conditions how much we care about errors form different sources)

for t in range(T): #time iteration
    pred1 = (h(phi2) @ Theta1.T)   # generate(phi2) = Theta1 h(phi2)

    if local_sigma:               # cf. Bogacz Eq. 59–60
        eps1 = eps1 + dt * (u - pred1 - e1) # error = (input - prediction) - sigma_eps
        e1   = e1   + dt * (eps1 * Sigma1 - e1)
    else:                         # cf. Bogacz Eq. 54
        eps1 = eps1 + dt * (u - pred1 - eps1 * Sigma1)

    # phi2 update (Eq. 53): bottom-up drive from eps1
    bu = (eps1 @ Theta1)          # take the sensory error eps1 and multiply by the weight matrix to feed the higher layer
    phi2 = phi2 + dt * (hprime(phi2) * bu - eps2)  # take derivative (dh/dphi) and multiply it by bottom-up error, subtract with top-down pressure (eps2)

# After inference settles: Hebbian weight update (Bogacz Eq. 56 / 50)
Theta1 += lr_theta * (eps1.T @ h(phi2)) / B

# If learning Sigma locally: Bogacz Eq. 71 (diagonal case -> elementwise)
if local_sigma:
    Sigma1 += lr_sigma * ((eps1 * e1).mean(dim=0) - 1.0)
    Sigma1 = Sigma1.clamp_min(1e-3)   # keep positive
```

## <a id="tc3" name="tc3"></a>Text Cell 3: three-level predictive-coding network over a continuous world state

The model makes several commitments that the reader should note while reading:

1. The chain of prediction runs utterance ($y$) $\leftarrow$ lexical ($L$) $\leftarrow$ situation ($S$) $\leftarrow$ utility ($u$).

2. The world state is continuous, $s\in(0,1)$, tracked in logit coordinates on a fixed quadrature grid. The grid is a numerical substrate and not a state space.

3. At this phase, the model does not implement any precision inference. Hence, every variance ($\sigma$) is fixed at $1$. The value is the multiplicative identity because $\sigma$ weights a synaptic connection, so fixing every $\sigma$ at $1$ assigns equal weight to every connection.

4. **Sign and language convention.** $\mathcal F$ below is equivalent to the $F$ defined by Bogacz (2017), henceforth "Bogacz". $F$ is *negative* free energy, with dynamics given by gradient *ascent*. Every gradient sign in this document follows Bogacz. The word "construction" refers to the general constraints that the free-energy principle imposes on a
predictive-coding network, as explicated in Bogacz.

5. **Subscript convention.** Four letters index four levels, capitalized exactly when the level is field-valued: $L$ and $S$ carry $\mathbb R^K$ fields, $y$ and $u$ do not. Each letter names one thing only. $y$ denotes the observed utterance ($E_y$, $\chi_y$, and the read-out $q_y$ of Text cell 4). $L$ denotes the lexical level ($\varphi_L$, $\varepsilon_L$, $\sigma_L$, $g_L$). $S$ denotes the situation level ($\varphi_S$, $\varepsilon_S$, $\sigma_S$, $g_S$); $S$ is capitalized because the level is field-valued, and the lowercase $s$ is reserved for the world state itself, the proportion of Eq. (1). $u$ denotes the utility level ($\theta_u$, $\varphi_u$, $\varepsilon_u$, $\sigma_u$, $\mu_u$) and does not denote utterance at any point.

### <a id="tc3-1" name="tc3-1"></a>1. State space and numerical substrate

The world state is a proportion $s\in(0,1)$, carried in logit coordinates

$$\zeta=\operatorname{logit}(s)=\log\frac{s}{1-s}\in\mathbb R,
\qquad
s=\operatorname{logit}^{-1}(\zeta)=\varsigma(\zeta)=\frac{1}{1+e^{-\zeta}},
\tag{1}$$

where $\varsigma$ is the standard logistic function. It is written $\varsigma$ rather than the usual
$\sigma$ throughout, because $\sigma$ is reserved here for the variances
$\sigma_y,\sigma_L,\sigma_S,\sigma_u$ of §3. Since $\operatorname{logit}$ is a bijection
$(0,1)\to\mathbb R$, Eq. (1) is a change of coordinates and not a modelling assumption; all Gaussian
machinery henceforth lives in $\zeta$.

Integrals over $\zeta$ are evaluated on a **fixed** grid ($\zeta_1,\dots,\zeta_K$) with fixed
positive quadrature weights ($w_1,\dots,w_K$), where $w_k,\zeta_k\in\mathbb R$. For $f,h\in\mathbb R^K$ we write

$$\langle f,h\rangle=\sum_{k=1}^K w_k\,f_k h_k,
\qquad
\|f\|^2=\langle f,f\rangle,
\qquad
w_k,\,f_k,\,h_k\in\mathbb R,\quad w_k>0 .
\tag{2}$$

We call such $\langle\cdot,\cdot\rangle$ the **quadrature inner product**: the inner product induced on
$\mathbb R^K$ by the quadrature rule. Equivalently, $\langle\cdot,\cdot\rangle$ is the inner product
of the weighted space $\ell^2(w)$. With $w_k>0$, Eq. (2) is a genuine inner product and $\|\cdot\|$ a norm. §8.1's proof will use this fact.

Where matrix notation is more convenient we write $W=\operatorname{diag}(w_1,\dots,w_K)$ for the
same measure, so that

$$\langle f,h\rangle=f^{\mathsf T}Wh,\qquad
B^{\mathsf T}Wf=\big(\langle b_1,f\rangle,\dots,\langle b_m,f\rangle\big)^{\mathsf T},\qquad
B^{\mathsf T}WB=\big(\langle b_i,b_j\rangle\big)_{ij}.
\tag{3}$$

The two notations are the same object throughout and are used interchangeably: $\langle\cdot,\cdot\rangle$
where a single pairing is meant, $W$ where several projections are taken at once. $W$ is diagonal and
positive, and never a free parameter: it is the grid's own measure, fixed with the grid.

The scalar $\langle f,h\rangle$ is a discretization of the $L^2$ inner product $\int f(\zeta)h(\zeta)\,d\zeta$, which it approaches as $K\to\infty$. What the grid discretizes is therefore the integral, and $K$ is accordingly an accuracy parameter: no quantity below is defined in terms of a state indexed by $k$.

**Where $\varepsilon_y$ lives.** The $y$ level is lowercase under commitment 5 for the same reason $u$
is: it carries one unit per word form, not a field over the grid. $g_y$ *contracts* the $K$-node
lexical field onto those $\lvert\mathcal Y\rvert$ units, by $\hat\varphi_y=\theta_L AW\varphi_L$ with
$A\in\mathbb R^{\lvert\mathcal Y\rvert\times K}$ (Appendix A, Eq. A2), whereas $g_S$ expands $\mathbb R^m$ into a
field. Hence

$$\varepsilon_y=\sigma_y^{-1}\big(\varphi_y-\theta_L AW\varphi_L\big)\in\mathbb R^{\lvert\mathcal Y\rvert},
\tag{4}$$

the mirror of $\varepsilon_u\in\mathbb R^m$ at the other end of the chain, and the space that makes
Eq. (A4) a scalar as $\theta_L$ requires. Eq. (8) pins $\varepsilon_y$ at $0$ because $\varphi_L$ is
clamped, not because the space is trivial: at $\lvert\mathcal Y\rvert=3$ the three clamped fields
$\Lambda\chi_y$ have Gram $X^{\mathsf T}WX$ of rank $3$ (condition number $21.0$), so the $A$ of Eq. (A2) exists, with $\theta_L AWX=I_3$ to
$2.4\times10^{-15}$, and unclamping reopens a genuinely three-dimensional error channel.

§§4–7 are written with $m=1$, writing $b$ for the single column of $B$, purely for legibility;
§8.4 gives the general form, and the implementation uses it.

### <a id="tc3-3" name="tc3-3"></a>3. Fixed quantities and their algebraic dependencies

1. **Variances $\sigma$** \
   $\sigma_y=\sigma_L=\sigma_S=\sigma_u=1$. This is the minimum admitted by Friston (2005)
   and adopted by Bogacz to keep the error-unit dynamics from diverging; it is also the whole of our
   commitment on precision, since this model performs no precision inference. Where $\sigma$ appears below
   it is retained symbolically so that the equations remain readable if a later phase unfixes it.

   Note that fixing $\sigma$ removes precision-learning (Bogacz Eqs. 20–21, 48–49, 55), but the absence of
   precision learning does not remove Hebbian learning from this implementation. In particular, the outer Hebbian
   loop performs $\Theta$-learning (see §8, $\theta_u$ remains plastic on the slow timescale (Eq. 20)). Precision learning is what
   requires the interneuron construction (Bogacz §5), which this model therefore does not carry. For further notes see Appendix B; for an auxiliary node introduced on quite different grounds, Appendix E (`appendix_E.ipynb`).

2. **Emission $E_y$** \
   Each utterance $y$ carries an exclusion set $E_y\subseteq[0,1]$, the set of states
   its lexical entry rules out. On the scale $\{\textit{no},\textit{some},\textit{all}\}$,

   $$E_{\textit{no}}=(0,1],\qquad E_{\textit{some}}=\{0\},\qquad E_{\textit{all}}=[0,1).
\tag{5}$$

   $E$ stands for emission and its denotation is not related to the error quantity $\varepsilon$. Note the latter would not be
   true if $E_y$ defines the inclusion of a word. Note that, at this phase, $E_y$ is fixed by the utterance entry
   alone and makes no reference to which other utterances might exist in some alternatives space.

   $E_y$'s indicator sampled on the grid is $\chi_{y,k}=\mathbb 1[\varsigma(\zeta_k)\in E_y]$, and the
   lexical field is $\chi_{y}$ scaled by $\Lambda\in\mathbb R_{>0}$, which denotes lexical strength
   (see entry "Lexical strength"):

   $$\varphi_L=\Lambda\,\chi_y.
\tag{6}$$

   For further notes, see Appendix A on boundary resolution, and Appendix D on why the entry is fixed by exclusion
   rather than by the truth set.

3. **Utility basis $B$** \
   The $u$ level carries $\varphi_u\in\mathbb R^m$ and the $S$ level carries a
   field, so the $u$ level can address the $S$ level only through a fixed profile matrix
   $B\in\mathbb R^{K\times m}$, as in Eq. (10). Every result in §8 is stated for arbitrary fixed $B$ and
   arbitrary $m$, and no proof below in this cell constrains the choice. Mind that, even though this cell is
   indifferent to the column size of $B$, the model's expressive power is not (see Text cell 5, Part A).
   The implementation takes $m=2$. Note that $B$ is the spatial profile of a single projection, which is what makes Eq. (20) an instance of
   the local plasticity rule. For further notes see Appendix B.

4. **Base world prior $\ell_0$**\
   $\ell_{0,k}=\log p_0(\zeta_k)$, where $p_0$ can be
   supplied by the elicited prior. This samples a base prior $p_0$ over the world state at the grid nodes.
   It enters through $g_L$ (Eq. 9) as the reference point of the lexical complement, and is then renormalized
   so that $\sum_k w_k e^{\ell_{0,k}}=1$, since an analytically normalized density is not grid-normalized once the grid is truncated.

5. **Utility prior mean $\mu_u$**\
   $\mu_u\in\mathbb R^m$ is the vector of weights on the terminating tonically
   active node, whose activity is $1$ by construction (Bogacz, Fig. 3). $\mu_u\neq0$ is a standing requirement by construction.
   The terminating contribution to $g_S$ is $\theta_u B\mu_u$ (see §4 and §8.5), which at $\mu_u=0$ vanishes for every $\theta_u$; the
   chain then stops rather than terminates, leaving $\theta_u$ unidentifiable, cf. the corresponding failure of Bogacz Eq. 25
   at $v_p=0$. However, different from Bogacz Eq. 25, $\mu_u\neq0$ is necessary but not by itself sufficient; the exact condition
   is stated in Appendix B. The implementation takes $\mu_u=\mathbf 1\in\mathbb R^m$.

6. **Lexical strength $\Lambda$**\
   $\Lambda>0$ mediates between the lexicon and the world belief. $\Lambda\to\infty$ recovers a hard truth-conditional constraint;
   finite $\Lambda$ makes it soft and differentiable. The implementation takes $\Lambda=8$ if not said otherwise. See also Text cell 6 for a
   sweep on $\Lambda$.

### <a id="tc3-4" name="tc3-4"></a>4. Generative model

The chain runs $E_y \leftarrow \varphi_L \leftarrow \varphi_S \leftarrow \varphi_u \leftarrow 1$, each
level predicting the one below:

$$\hat\varphi_y=g_y(\varphi_L),\qquad
\hat\varphi_L=g_L(\varphi_S),\qquad
\hat\varphi_S=g_S(\varphi_u),\qquad
\hat\varphi_u=\mu_u\cdot 1 .
\tag{7}$$

1. **$g_y$: lexical content to word form.** \
   This map is out of scope. The map from a lexical field to an utterance
   form is the province of an auditory/word-recognition model, and this network is written so that such a
   model can be stacked beneath it. In the present phase $\varphi_L$ is clamped by $y$ through the entry
   (Eqs. 5–6), so that

   $$\varepsilon_y\equiv 0
\tag{8}$$

   identically. Eq. (8) is therefore a *consequence* of the clamp on $\varphi_L$. Because $\varphi_L$ is clamped, the $y$ level takes no part in the variational dynamics at this
   phase. It is retained rather than deleted so that unclamping $\varphi_L$ later reopens the channel
   with no change to anything below $\varphi_L$; the definition of $g_y$, and what follows from it,
   are left to Appendix A.

2. **$g_L$: situation belief to lexical exclusion.** \
   $\varphi_S$ is an unnormalized log-density; a state is
   excluded to the degree its log-weight is low. Hence

   $$g_L(\varphi_S)=\ell_0-\varphi_S.
\tag{9}$$

   This function is a transported set-complement map. On indicator vectors, the complement is $x\mapsto 1-x$,
   which is the order-reversing involution of $[0,1]$; on log-weights, the corresponding order-reversing involution of
   $\mathbb R$ is $x\mapsto \ell_0-x$, reflection through the base world prior. To paraphrase: by placing the base-world-prior in $g_L$,
   we take $\ell_0$ as the reference instead of zero; that is, the complement of lexical exclusion remains when the
   base-world-prior is subtracted. As both are involutions and are affine, Eq. (9) therefore preserves the complement structure.

   Note $g_L'=-I$, that is, the lexical error ($\varepsilon_L$) reaches the situation belief ($\varphi_S$) *inhibitorily*.
   Appendix D derives this sign, and $\ell_0$'s place here, from the exclusion convention of §3.

3. **$g_S$: utility to situation belief.**

   $$g_S(\varphi_u)=\theta_u\varphi_u\,b.
\tag{10}$$

   This function is an instance of $\Theta h(\varphi)$ (cf. Bogacz Eq. 42) with $\Theta=\theta_u b$ and
   $h(\varphi)=\mathrm{id}(\varphi)$. Eq. (10) is written at $m=1$, as are §§8.1–8.3; the general form is
   $g_S(\varphi_u)=\theta_u B\varphi_u$, with $b$ the single column of $B$ at $m=1$, and §8.4 gives the
   lift. The scalar $\theta_u$ is the whole of the parameter here: $B$ is fixed (§3).

   Note that $g_S$ is **homogeneous**, that is, it depends on $\theta_u$ and $\varphi_u$ only through their product.
   Since $\varphi_u$ is anchored only by $\mu_u$ (Eq. 11), the map $(\theta_u,\mu_u)\mapsto(-\theta_u,-\mu_u)$ leaves
   every belief unchanged: the sign of $\theta_u$ carries no content on its own, and is fixed only relative to $\mu_u$.
   What $\mu_u$ does and does not move between utterances is covered in Text cell 5 Part B.

4. **Residuals.** \
   With every $\sigma=1$ the error units and the raw residuals coincide at equilibrium,
   so write

   $$\varepsilon_L=\sigma_L^{-1}(\varphi_L-g_L(\varphi_S))=1(\varphi_L-\ell_0+\varphi_S)=r_L,\qquad
   \varepsilon_S=\sigma_S^{-1}(\varphi_S-g_S(\varphi_u))=1(\varphi_S-\theta_u\varphi_u b)=r_S,\qquad
   \varepsilon_u=\sigma_u^{-1}(\varphi_u-\mu_u)=r_u.
\tag{11}$$

5. **Read-out.** \
   The reportable belief is the normalized density

   $$q(\zeta_k)=\frac{e^{\varphi_{S,k}}}{\sum_j w_j e^{\varphi_{S,j}}},\qquad
   \sum_k w_k\,q(\zeta_k)=1.
\tag{12}$$

   The statistics reported from Eq. (12) in the evaluation, namely $\mathbb E[s]$, $\mathrm{sd}(s)$, the region
   $q$-masses and the leak, are summaries of the read-out $q$. They are not model quantities, and are defined in
   Text cell 4 at Eqs. (25)–(27).

### <a id="tc3-5" name="tc3-5"></a>5. Free energy

Following Bogacz Eq. (7) generalized to a hierarchy (his Eqs. 51–54), with the quadrature inner
product of Eq. (2) supplying the measure on the two field levels:

$$\boxed{\;
\mathcal F=-\frac{\|r_L\|^2}{2\sigma_L}-\frac{\|r_S\|^2}{2\sigma_S}-\frac{\|r_u\|^2}{2\sigma_u}
\;-\;\tfrac12\big(K\log\sigma_L+K\log\sigma_S+m\log\sigma_u\big)+C\;}
\tag{13}$$

Each level contributes one $\log\sigma$ per component, hence $K$ at the two field levels and $m$ at
the utility level; $\|r_u\|^2$ is the ordinary squared norm on $\mathbb R^m$, the two field norms
being those of Eq. (2).

The $\log\sigma$ terms vanish at $\sigma=1$ and are constant in every variational quantity; they are
carried only to mark where a later precision-bearing version would reintroduce them. $r_y\equiv0$
contributes nothing. $\mathcal F$ is to be **maximized**.

### <a id="tc3-6" name="tc3-6"></a>6. Targets of the variational quantities

The target of each quantity is its value at the unique stationary point of $\mathcal F$ (uniqueness is
established in §8).

**Error units.** Each error unit's target is its residual,

$$\varepsilon_L^\ast=\frac{r_L}{\sigma_L},\qquad
\varepsilon_S^\ast=\frac{r_S}{\sigma_S},\qquad
\varepsilon_u^\ast=\frac{r_u}{\sigma_u},
\tag{14}$$

which is Bogacz Eq. (57).

**World belief.** Setting $\partial\mathcal F/\partial\varphi_S=0$ gives
$\varepsilon_S=-\varepsilon_L$, i.e. $r_S/\sigma_S=-r_L/\sigma_L$, and the residual form
$r_S=-r_L$ holds only at $\sigma_L=\sigma_S$; hence

$$\varphi_S^\ast=\frac{\sigma_L\,\theta_u\varphi_u b+\sigma_S\big(\ell_0-\varphi_L\big)}
{\sigma_L+\sigma_S}.
\tag{15}$$

Two remarks follow, both stated at $\sigma=1$ for concreteness. First, the excluded region receives
$-\Lambda/2$, so its $q$-mass
vanishes as $\Lambda\to\infty$. Second, the factor $\tfrac12$ is a genuine temperature, which is defined
as the precision-weighted combination of two unit-variance channels, and, because the read-out (12) is exponential,
it flattens the reported belief $q$. That the factor equals $\tfrac12$ is a consequence of $\sigma_L=\sigma_S=1$
and is the natural place for a later phase to intervene.

**Utility state.** Setting $\partial\mathcal F/\partial\varphi_u=0$ gives
$\varepsilon_u=\theta_u\langle\varepsilon_S,b\rangle$; substituting Eq. (15),

$$\varphi_u^\ast=\frac{\mu_u+\dfrac{\sigma_u\theta_u}{S}\,\langle\ell_0-\varphi_L,\;b\rangle}
{1+\dfrac{\sigma_u\theta_u^{2}}{S}\,\|b\|^2},
\qquad S=\sigma_L+\sigma_S.
\tag{16}$$

**Two fixed-point identities**:

$$\varepsilon_S=-\varepsilon_L,\qquad\qquad \varepsilon_u=\theta_u\langle\varepsilon_S,b\rangle.
\tag{17}$$

Eqs. (15)–(17) are closed-form and are the correctness test for the implementation: the network's
numerical fixed point must reproduce them.

### <a id="tc3-7" name="tc3-7"></a>7. Dynamics

Error units converge toward their residuals (Bogacz Eqs. 53–54 at $\Sigma=1$):

$$\tau_\varepsilon\dot\varepsilon_L=\frac{r_L}{\sigma_L}-\varepsilon_L,\qquad
\tau_\varepsilon\dot\varepsilon_S=\frac{r_S}{\sigma_S}-\varepsilon_S,\qquad
\tau_\varepsilon\dot\varepsilon_u=\frac{r_u}{\sigma_u}-\varepsilon_u.
\tag{18}$$

State units change in proportion of $\mathcal F$ and converge at the maximum of $\mathcal F$. Differentiating Eq. (13),
and using $g_L'=-I$ and $\partial g_S/\partial\varphi_u=\theta_u b$:

$$\tau_\varphi\dot\varphi_S=\frac{\partial\mathcal F}{\partial\varphi_S}=-\varepsilon_S-\varepsilon_L,
\qquad
\tau_\varphi\dot\varphi_u=\frac{\partial\mathcal F}{\partial\varphi_u}
=-\varepsilon_u+\theta_u\langle\varepsilon_S,b\rangle.
\tag{19}$$

Both lines are instances of Bogacz Eq. (53), $\dot\varphi_i=-\varepsilon_i+h'(\varphi_i).\!*\,
\Theta_{i-1}^{\mathsf T}\varepsilon_{i-1}$: the first has $\Theta_{L}=-I$, the second $\Theta_S=b$
contracted under Eq. (2).

The slow parameter follows Bogacz's own gradient. His Eq. (25) gives
$\partial F/\partial\theta=\varepsilon_u\phi$, the error *below* times the state *above*, and his
Eq. (29) gives the same for a nonlinear $h$. Both are stated at the inference stage, before plasticity is
introduced, and both are exactly our quantity:

$$\tau_\theta\,\dot\theta_u=\frac{\partial\mathcal F}{\partial\theta_u}
=\varphi_u\langle\varepsilon_S,b\rangle,
\qquad
\tau_\varepsilon\ll\tau_\varphi\ll\tau_\theta.
\tag{20}$$

*There is no learning rate here, and none is required.* Bogacz supplies the gradient and says that
parameters move proportionally to it; the constant of proportionality is a **time constant**, of
exactly the kind Eqs. (18) and (19) already carry for the error and state units. So $\theta_u$ takes a
third, $\tau_\theta$, and the model carries one ordering of timescales. Its slow end, $\tau_\varphi\ll\tau_\theta$, is what §8.5 needs, exactly as its
fast end $\tau_\varepsilon\ll\tau_\varphi$ is what §8.3 needs. $\tau_\theta$ sets the **rate only**:
the fixed point of Eq. (20) is the stationary point of $\tilde{\mathcal F}$ and does not depend on it.

Note: Eqs. (19) and (20) are written at $m=1$, and §8.4 lifts the convergence proof to $m>1$. **However, at $m>1$, the
locality of Eq. (20) no longer holds.** At $m>1$ this update requires a quantity that no single
synapse holds, so the rule is Hebbian as written here and not in the case the implementation runs.
See Appendix B for this violation stated in exact terms; see Appendix E for the resolution of this locality violation.

### <a id="tc3-8" name="tc3-8"></a>8. Convergence

Throughout §§8.1–8.3, $\theta_u$ is held fixed; the two-timescale separation that licenses this is
§8.5. Write $x=(\varphi_S,\varphi_u)\in\mathbb R^{K}\times\mathbb R$.

#### <a id="tc3-8-1" name="tc3-8-1"></a>8.1 $\mathcal F$ is strictly concave in $x$

Each of $r_L,r_S,r_u$ is **affine** in $x$. This is where the linearity of $g_L$ (Eq. 9) and of $g_S$
in $\varphi_u$ (Eq. 10) is used, and it is the only place either is needed. Hence $\mathcal F$ is a
negative sum of squares of affine functions, so concave. For strictness, take a direction
$(v,t)\in\mathbb R^K\times\mathbb R$, $(v,t)\neq 0$, and evaluate the quadratic form of the Hessian by
reading off the directional derivative of each residual: $\partial_{(v,t)}r_L=v$, $\partial_{(v,t)}r_S
=v-t\theta_u b$, $\partial_{(v,t)}r_u=t$. Therefore

$$(v,t)^{\mathsf T}\nabla^2\mathcal F\,(v,t)
=-\frac{\|v\|^2}{\sigma_L}-\frac{\|v-t\theta_u b\|^2}{\sigma_S}-\frac{t^2}{\sigma_u}.
\tag{21}$$

Every term is $\le 0$; the first vanishes only if $v=0$ and the third only if $t=0$. So the form is
strictly negative for $(v,t)\neq0$ and $\nabla^2\mathcal F$ is negative definite.

Three consequences are worth recording. **(i)** $\mathcal F$ has a unique stationary point, which is its
global maximum; Eqs. (15)–(16) are therefore *the* solution, not *a* solution. **(ii)** Eq. (21) holds
for every $\theta_u\in\mathbb R$ and every $\Lambda>0$, so neither the lexical strength nor the
generative weight can destabilize the fast subsystem. **(iii)** $\varphi_L$ does not appear in Eq. (21)
at all: it enters $r_L$ additively and so shifts the maximizer without touching the Hessian. **The
result is independent of the shape of the lexical field.** §9.2 and §9.3 rest on this.

#### <a id="tc3-8-2" name="tc3-8-2"></a>8.2 The fast subsystem converges

Along the flow (19), $\dot{\mathcal F}=\langle\nabla_x\mathcal F,\dot x\rangle=
\tau_\varphi^{-1}\|\nabla_x\mathcal F\|^2\ge0$, with equality only at the stationary point. $\mathcal F$
is bounded above (it is a negative sum of squares) and strictly concave, so gradient ascent converges
globally to the unique maximizer of §8.1, from any initialization.

*Standard result invoked without proof, following the omission convention of Bogacz's paper:*
convergence of gradient flow on a strictly concave, bounded-above, coercive objective. Coercivity is
immediate: $\mathcal F\to-\infty$ as $\|x\|\to\infty$, since $\|r_L\|^2+\|r_u\|^2$ already dominates
$\|x\|^2$.

#### <a id="tc3-8-3" name="tc3-8-3"></a>8.3 The error units do not disturb this

Eqs. (18)–(19) are jointly linear, so their combined fixed point is the one computed above and its
stability is a matter of the eigenvalues of a constant matrix. Bogacz gives exactly this analysis for
the corresponding subsystem (his §5.1, following Eqs. 59–61), establishing that $\varepsilon$ converges
to the residual when $\varphi$ evolves on a slower timescale. We adopt that result and impose the
corresponding condition $\tau_\varepsilon\ll\tau_\varphi$, with the further numerical requirement
$\mathrm{d}t\le\tau_\varepsilon$ for the Euler integration.

#### <a id="tc3-8-4" name="tc3-8-4"></a>8.4 The proof does not depend on dimension or basis

§§8.1–8.3 are written at $m=1$; the implementation takes $m=2$, so the lift is stated here. Let
$\varphi_u\in\mathbb R^m$ and $B\in\mathbb R^{K\times m}$, with $\theta_u$ the scalar of Eq. (10), so
$g_S=\theta_u B\varphi_u$. For a direction $(v,t)\in\mathbb R^K\times\mathbb R^m$ the residual
derivatives are $\partial_{(v,t)}r_L=v$, $\partial_{(v,t)}r_S=v-\theta_u Bt$ and
$\partial_{(v,t)}r_u=t$, which is Eq. (21) with $t\theta_u b$ replaced by $\theta_u Bt$, so

$$(v,t)^{\mathsf T}\nabla^2\mathcal F\,(v,t)
=-\frac{\|v\|^2}{\sigma_L}-\frac{\|v-\theta_u Bt\|^2}{\sigma_S}-\frac{\|t\|^2}{\sigma_u},
\tag{22}$$

the field norms being those of Eq. (2) and $\|t\|^2$ the ordinary norm on $\mathbb R^m$.

**Why $B$ cannot matter.** $B$ appears only in the middle term, and that term does no work: it is
$\le0$ throughout, while strictness is carried by the outer two: the third excludes $t\neq0$, and
the first and second together exclude $t=0,\,v\neq0$. Eq. (22) is therefore negative definite for
**every** $B\in\mathbb R^{K\times m}$ (rank-deficient, non-orthonormal, $m>K$, any profile shape in
$\zeta$, and $B=0$ included) and for **every** $\theta_u\in\mathbb R$, zero and negative alike. Nothing
in §§8.1–8.2 used $m=1$, the shape of $b$ as a function of $\zeta$, or the dimensionality of $\zeta$.
(The linearity that *is* used is linearity of $g_S$ in $\varphi_u$, which makes $r_S$ affine in $x$;
that is a different property, named in §8.1.)

**The matrix case.** Eq. (22) is negative definite with any $M\in\mathbb R^{K\times m}$ in place of $\theta_u B$, so
concavity would also survive promoting $\theta_u$ to a matrix. Mind that is a statement about the reach of the
proof, not a licence to adopt it: with $\theta_u$ a matrix the $u\to S$ weights become $K\times m$
independently modifiable synapses, which is a different circuit from the one specified. For further notes
see Appendix B. This implementation still takes $\theta_u\in\mathbb R$.

#### <a id="tc3-8-5" name="tc3-8-5"></a>8.5 The slow parameter

$\mathcal F$ is concave in $\theta_u$ for fixed $x$, but **not jointly concave** in
$(\theta_u,\varphi_u)$: the term $\theta_u\varphi_u b$ is bilinear. Under
$\tau_\varphi\ll\tau_\theta$ the fast subsystem equilibrates between $\theta_u$ updates, so Eq. (20)
is ascent on $\tilde{\mathcal F}(\theta_u)=\max_x\mathcal F(x,\theta_u)$, whose gradient coincides
with $\partial\mathcal F/\partial\theta_u$ at the equilibrium $x^\ast(\theta_u)$ (Danskin's theorem,
invoked without proof; its hypotheses hold because $x^\ast$ is unique by §8.1). $\tilde{\mathcal F}$
is bounded above by $0$, being a maximum of negative sums of squares, so the ascent converges.

Nevertheless, that $\tilde{\mathcal F}$ converges does not by itself supply a maximizer to converge to.
Appendix B treats the premise this rests on, the sign and location of $\theta_u^\ast$, the locality of
Eq. (20), and what a learned $\theta_u$ implies about the alternative space of an observed utterance.

#### <a id="tc3-8-6" name="tc3-8-6"></a>8.6 A note on what the convergence in this section does and does not show

With $\varphi_L$ clamped and both $g_L,g_S$ linear, the model is linear-Gaussian at this phase and
Eqs. (15)–(16) solve it in closed form. The recurrent dynamics are therefore a claim about *neural
implementation*, the same status they have in Bogacz's own linear examples, and the closed forms make
§6 an exact test of that implementation.
Nonlinearity would be present if $h$ is made nonlinear in $\varphi_u$ (Eq. 10, currently the identity).
However, the shape of $b$ as a function of $\zeta$ is not such a route: $b$ is a fixed vector on the grid,
so $g_S$ stays linear in $\varphi_u$ whatever profile it carries (§8.4).

To conclude, the proofs in this section depend on three conditions: that $g_L$ and $g_S$ are affine
in $x$ (§8.1); that $\varphi_L$ is clamped, so that $\varphi_L$ is not a variable (§8.6); and that
$\sigma_u$ is finite with at least one of $\sigma_L,\sigma_S$ finite as well. Finite $\sigma_u$ carries
strictness in $t$, and the finite member of $\sigma_L,\sigma_S$ carries it in $v$. Should a later phase
unfix $\sigma$ (§3), that last condition is the one Eq. (22) needs.

### <a id="tc3-9" name="tc3-9"></a>9. Foreseen extensions

#### <a id="tc3-9-1" name="tc3-9-1"></a>9.1 The flat direction: resolved, with a standing condition

**Claim: compatible; the concern does not arise.** An unnormalized log-density is ordinarily defined
only up to an additive constant, so one expects $\mathcal F$ to be flat along $\mathbf 1$ and
$\varphi_S$ to drift without converging. Here it does not: §8.1 shows $\nabla^2\mathcal F$ is negative
definite, so there is no flat direction in any direction, $\mathbf 1$ included. The reason is structural: $\varphi_S$ is a *state* with its own Gaussian errors at both the $L$ and $S$ levels, and
those pin it in absolute terms. It is not merely a parameterization of the distribution (12).

**Caveat.** Because the absolute scale of $\varphi_S$ is pinned,
that scale does work: it is the temperature noted under Eq. (15), and the read-out (12) is sensitive to
it. The consequence is that the factor $\tfrac12$ in Eq. (15) is a real flattening of the reported
belief, so any future change to $\sigma_L$ or $\sigma_S$ changes the reported distribution's sharpness
even when it leaves the support untouched.

#### <a id="tc3-9-2" name="tc3-9-2"></a>9.2 One-dimensional dense-scale predicates

**Claim: compatible.** By §8.1(iii) the
convergence result is independent of the shape of $\varphi_L$: the lexical field enters only additively,
and the Hessian does not see it. Moving from a sharp quantifier to a dense-scale predicate replaces
$\chi_y\in\{0,1\}^K$ with a smooth profile $\chi_y\in[0,1]^K$ and changes nothing else: not the level structure, not the messages, not $\theta_u$,
and not any proof in §8. Eqs. (15)–(17) hold verbatim.

The state space is already continuous (Eq. 1), so no discretization is being relaxed; $K$ is a
quadrature parameter throughout.

**Caveats.**
1. *Quadrature accuracy runs opposite to intuition.* A smooth $\chi_y$ is **easier** to integrate on a
   fixed grid than a step. The sharp quantifier boundaries of the present phase are the numerically
   demanding case, incurring $O(1/K)$ error near the discontinuity; dense-scale predicates relax this.
   The grid is thus the representation that serves both phases.

2. *$\Lambda$ is bounded in practice.* Large $\Lambda$ makes $\varphi_S$ badly scaled, and
   $\Lambda\to\infty$ is a limit, not a setting. The read-out (10) should be computed in log-space.

3. *A parametric posterior is closed off.* If a later phase wants $q$ to be a two-parameter family
   (logit-normal, say) rather than a grid-borne field, sharp exclusions
   become unrepresentable: an exponential family with fixed support cannot move its support. This is a restriction on *parametric* $q$, not on the
   present formulation, and it is the reason the grid is retained.

4. *Unclamping $\varphi_L$ is a separate step.* A predicate whose lexical content is itself inferred
   requires $\varphi_L$ to become variational, which reopens $\varepsilon_y$ and therefore requires
   $g_y$, which §4 declares out of scope. The architecture accommodates this without modification; the
   *content* of $g_y$ is not supplied here.

#### <a id="tc3-9-3" name="tc3-9-3"></a>9.3 Multidimensional predicates

**Claim: theoretically compatible.** §8.4 establishes negative definiteness for arbitrary $M$ and
arbitrary $m$, and nothing in §§4–7 assumed $\zeta$ to be scalar. A predicate over $d$ jointly inferred
degree scales requires only that $\zeta\in\mathbb R^d$, that $b$ become a matrix, and that $\varphi_u$
become a vector. Every equation and every proof above carries over unchanged.

**Caveats, and this is where a real ceiling sits.**
1. *The grid is exponential in $d$.* A tensor-product quadrature costs $K^d$ nodes. At $d=1$, $K\approx101$
   against two variational states is negligible; at $d=3$ it is already $10^6$ nodes per iteration. The
   limit is **dimensional, not dense**: refining a one-dimensional scale is cheap, adding a scale is not.

2. *The escape routes conflict with §9.2's caveat 3.* Factored, parametric, or sampled representations
   are the standard remedies for grid blow-up, but a parametric $q$ cannot represent sharp exclusions.
   **Multidimensional semantics with sharp lexical boundaries is therefore the one corner this
   formulation does not reach**, and it is declared as a non-compatibility rather than a caveat. A
   factored grid over dimensions that are semantically independent is the most promising route, but it
   assumes an independence that scalar predicates may not license, and no claim is made here.

3. *Identifiability is untested.* With $m>1$ utility parameters addressing a $d$-dimensional field,
   $\theta_u$ acquires rotational degeneracies that §8.5's local convergence claim does not address.
   Nothing above rules this out; nothing above establishes it either.

````python
# === code cell 1: three-level predictive-coding network over a continuous
#     world state. ARCHITECTURE ===
#
# Implements the specification in the preceding markdown cell. Equation numbers below
# refer to that cell throughout.
#
# SIGN CONVENTION: F is Bogacz's negative free energy and is MAXIMIZED; every state
# equation is gradient ASCENT (spec preamble item 4, Eq. 13).

import math
import torch

DTYPE = torch.float64

UTTERANCES = ["no", "some", "all"]


# 1. NUMERICAL SUBSTRATE (spec Sec. 1)

def build_grid(num_nodes=101, half_width=6.0, dtype=DTYPE, device=None):
    """Fixed quadrature grid over zeta, with trapezoidal weights (Eqs. 1-2).

    The grid discretizes the integral, not the state: `num_nodes` is an accuracy
    parameter, and no quantity in the model is defined in terms of a state indexed
    by a node.
    """
    if num_nodes < 3:
        raise ValueError("num_nodes must be at least 3")
    if half_width <= 0:
        raise ValueError("half_width must be positive")

    zeta = torch.linspace(-half_width, half_width, num_nodes, dtype=dtype, device=device)
    spacing = (zeta[1] - zeta[0]).item()

    weights = torch.full_like(zeta, spacing)
    weights[0] *= 0.5
    weights[-1] *= 0.5
    return zeta, weights


def logistic(x):
    """s = sigma(zeta), recovering the proportion from its logit (Eq. 1)."""
    return torch.sigmoid(x)


# 2. LEXICON (spec Sec. 3, Eqs. 5-6)

def exclusion_indicator(utterance, zeta, theta_L, sharpness=None):
    """chi_y on the grid: 1 where the entry for `utterance` excludes the state (Eq. A1).

        E_no    ->  excluded where zeta >  -theta_L   (outside the cell of s=0)
        E_some  ->  excluded where zeta <= -theta_L   (inside  the cell of s=0)
        E_all   ->  excluded where zeta <   theta_L   (outside the cell of s=1)

    `theta_L` is semantic, not a parameter of the discretization: Appendix A gives its
    denotation (Eqs. A5-A6), the Voronoi reading of Eq. (A1), and the grid cap. The
    step is also the demanding case of Sec. 9.2 caveat 1.

    `sharpness=None` gives the step of the current phase (chi in {0, 1}); a positive
    float gives a logistic profile (chi in [0, 1]) for dense-scale predicates. Spec
    Sec. 8.1(iii) guarantees every convergence result is independent of this choice.
    """
    if utterance not in UTTERANCES:
        raise ValueError(f"utterance must be one of {UTTERANCES}")
    if theta_L <= 0:
        raise ValueError("theta_L must be positive")
    if sharpness is not None and sharpness <= 0:
        raise ValueError("sharpness must be positive when given")

    # signed distance into the excluded region, positive where excluded
    if utterance == "no":
        margin = zeta + theta_L
    elif utterance == "some":
        margin = -(zeta + theta_L)
    else:  # "all"
        margin = theta_L - zeta

    if sharpness is None:
        return (margin > 0).to(zeta.dtype)
    return torch.sigmoid(margin / sharpness)


# 3. BASE WORLD PRIORS (spec Sec. 3, the fixed quantity ell_0)

# ell_0 is a log-density in zeta. Because the state is carried in logit coordinates,
# a prior specified over the proportion s must be PUSHED FORWARD through the change of
# variables, picking up the Jacobian ds/dzeta = s(1-s). The distinction is not cosmetic:
# a distribution that is flat in s is NOT flat in zeta.

def gaussian_world_prior(mean=0.0, precision=1.0):
    """Gaussian in zeta: the elicited base prior ell_0 of Sec. 3."""
    if precision <= 0:
        raise ValueError("precision must be positive")

    def log_density(zeta):
        return (
            0.5 * math.log(precision / (2.0 * math.pi))
            - 0.5 * precision * (zeta - mean).square()
        )

    log_density.description = f"Gaussian in zeta (mean {mean}, precision {precision})"
    return log_density


def beta_world_prior(alpha, beta):
    """Beta(alpha, beta) in s, pushed forward to zeta.

    With s = sigma(zeta) and ds/dzeta = s(1 - s), the Jacobian absorbs one factor from
    each side of the Beta density and the log-density collapses to

        log p(zeta) = alpha * log s + beta * log (1 - s) - log B(alpha, beta),

    evaluated through softplus for numerical stability. Note the exponents are alpha
    and beta rather than the usual alpha - 1 and beta - 1: that shift IS the Jacobian.

    Beta(1, 1) is the uniform prior over the proportion; alpha != beta is skewed.
    """
    if alpha <= 0 or beta <= 0:
        raise ValueError("alpha and beta must be positive")

    log_normalizer = (
        math.lgamma(alpha) + math.lgamma(beta) - math.lgamma(alpha + beta)
    )

    def log_density(zeta):
        log_s = -torch.nn.functional.softplus(-zeta)          # log sigma(zeta)
        log_one_minus_s = -torch.nn.functional.softplus(zeta)  # log (1 - sigma(zeta))
        return alpha * log_s + beta * log_one_minus_s - log_normalizer

    log_density.description = f"Beta({alpha}, {beta}) in s"
    return log_density


def uniform_world_prior():
    """Literal Uniform(0, 1) over the proportion.

    Implemented as its own case, NOT as the low-precision limit of the Gaussian:
    sending precision -> 0 inflates the variance in
    zeta and the sigmoid squashes the result into bimodal prior mass at the endpoints, which
    is the opposite of uniform. Uniform in s is Beta(1, 1), whose push-forward to zeta
    is the logistic density, unimodal about zeta = 0 and not flat there.
    """
    density = beta_world_prior(1.0, 1.0)
    density.description = "Uniform(0, 1) in s"
    return density


BASE_WORLD_PRIORS = {
    "gaussian": gaussian_world_prior(0.0, 1.0),
    "flat": uniform_world_prior(),
    "skewed low": beta_world_prior(1.0, 3.0),
    "skewed high": beta_world_prior(3.0, 1.0),
}


# 4. THE NETWORK

class LexicalPredictiveCodingNetwork:
    """y <- phi_L <- phi_S <- phi_u <- 1, with explicit error units.

    Clamped:      y, phi_L                       (Eqs. 6, 8)
    Variational:  phi_S, phi_u, and the error units    (fast, Eqs. 18-19)
                  theta_u                              (slow, Eq. 20)
    Fixed:        every sigma, Lambda, b, ell_0, mu_u, and the grid (Sec. 3)
    """

    def __init__(
        self,
        *,
        num_nodes=101,
        grid_half_width=6.0,
        base_prior=None,
        prior_mean=0.0,
        prior_precision=1.0,
        theta_L=None,
        num_atoms=10,
        lexical_strength=8.0,
        mu_u=1.0,
        theta_u=1.0,
        basis_degree=2,
        sigma_lexical=1.0,
        sigma_state=1.0,
        sigma_utility=1.0,
        exclusion_sharpness=None,
        device=None,
    ):
        if prior_precision <= 0:
            raise ValueError("prior_precision must be positive")
        if lexical_strength <= 0:
            raise ValueError("lexical_strength (Lambda) must be positive")
        if theta_L is None and (int(num_atoms) != num_atoms or num_atoms < 2):
            raise ValueError(
                "num_atoms (n) must be an integer >= 2; below 2 the entries for "
                "\"some\" and \"all\" coincide"
            )
        if min(sigma_lexical, sigma_state, sigma_utility) <= 0:
            raise ValueError("variances must be positive")
        if min(sigma_lexical, sigma_state, sigma_utility) < 1.0:
            raise ValueError(
                "variances below 1 are outside the floor adopted in Sec. 3 "
                "(Friston 2005; Bogacz footnote 3)"
            )

        self.dtype = DTYPE
        self.device = device

        self.zeta, self.weights = build_grid(
            num_nodes=num_nodes,
            half_width=grid_half_width,
            dtype=self.dtype,
            device=device,
        )
        self.num_nodes = num_nodes

        # Fixed quantities (Sec. 3). Every sigma is carried symbolically although the
        # present phase fixes all three at 1, so that a later phase can unfix them
        # without the equations below changing shape.
        self.sigma_lexical = float(sigma_lexical)
        self.sigma_state = float(sigma_state)
        self.sigma_utility = float(sigma_utility)

        self.lexical_strength = float(lexical_strength)

        # LEXICAL THRESHOLD theta_L.
        #
        # logit is asymptotic at s = 0 and s = 1, so it is a bijection only on the OPEN
        # interval and the endpoints where Eq. (5) anchors "no" and "all" have no zeta
        # coordinate. Realizing E_y on the grid therefore requires a threshold
        # theta_L > 0, with E_y mapping to {zeta > -theta_L}, {zeta <= -theta_L} and
        # {zeta < theta_L} for "no", "some" and "all".
        #
        # theta_L is a generative weight of g_y, the lexical-content-to-word-form map:
        # recovering y from phi_L requires the threshold that produced it. Its gradient
        # is dF/dtheta_L = eps_y * dg_y/dtheta_L, which vanishes identically while
        # eps_y == 0 (Eq. 8), so theta_L is fixed BECAUSE phi_L is clamped rather than
        # by stipulation; unclamping phi_L makes it learnable.
        #
        # n is theta_L's denotation: sigma(-theta_L) = 1/(2n) (Eq. A5). A count where
        # the predicate has atoms, else 1/delta for a JND delta (Eq. A6, Appendix A).
        if theta_L is None:
            self.theta_L = math.log(2 * int(num_atoms) - 1)
        else:
            self.theta_L = float(theta_L)
        if self.theta_L <= 0:
            raise ValueError("theta_L must be positive")
        # derived from theta_L (Eq. A5), not an independent quantity
        self.num_atoms = (math.exp(self.theta_L) + 1.0) / 2.0
        if self.theta_L >= grid_half_width:
            raise ValueError(
                f"theta_L = {self.theta_L:.4f} exceeds the grid half-width "
                f"{grid_half_width}: E_no leaves no grid node unexcluded. Widen the "
                f"grid, or lower n (n = {self.num_atoms:.4g} requires half_width > "
                f"{self.theta_L:.4f})."
            )
        self.exclusion_sharpness = exclusion_sharpness
        self.theta_u = float(theta_u)

        # UTILITY BASIS B (K x m). The default m = 2 takes zeta and zeta^2, orthonormalized under the
        # quadrature inner product against the constant function and each other. The
        # linear component separates "no" from {"some", "all"} and the quadratic
        # separates "some" from {"no", "all"}, so all three utterances receive
        # distinct projections.
        if basis_degree < 1:
            raise ValueError("basis_degree must be at least 1")
        self.basis_degree = int(basis_degree)

        raw = [torch.ones_like(self.zeta)] + [
            self.zeta ** d for d in range(1, self.basis_degree + 1)
        ]
        orthonormal = []
        for vector in raw:
            for previous in orthonormal:
                vector = vector - self.inner(vector, previous) * previous
            norm = torch.sqrt(self.inner(vector, vector))
            if float(norm) <= 1e-12:
                raise ValueError("basis degenerated during orthonormalization")
            orthonormal.append(vector / norm)
        # discard the constant, keep the m read-out-relevant directions
        self.basis = torch.stack(orthonormal[1:], dim=1)          # (K, m)
        self.num_basis = self.basis.shape[1]

        # mu_u: the weight on the terminating tonic node, one per basis direction.
        # Scalars broadcast.
        #
        # mu_u MUST BE NONZERO. The terminating contribution to g_S is theta_u * B * mu_u,
        # so at mu_u = 0 it is identically zero for every theta_u: the chain stops rather
        # than terminates, phi_u loses its anchor, and theta_u becomes unidentifiable --
        # it diverges by shifting magnitude out of the priced factor phi_u into the
        # unpriced theta_u along the exact degeneracy (theta_u, phi_u) -> (c theta_u,
        # phi_u / c), which leaves g_S invariant. This is not a departure from Bogacz:
        # his Eq. (25) model diverges identically at v_p = 0 (argmax theta = u / v_p).
        # v_p != 0 is an unstated premise of his construction, whose worked example
        # takes v_p = 3. The value 1.0 is a stipulation; a terminating prior is a fixed
        # boundary condition and has no target to be derived from.
        mu_u_tensor = torch.as_tensor(mu_u, dtype=self.dtype, device=device)
        if mu_u_tensor.ndim == 0:
            mu_u_tensor = mu_u_tensor.expand(self.num_basis).clone()
        if mu_u_tensor.shape != (self.num_basis,):
            raise ValueError(f"mu_u must be a scalar or have shape ({self.num_basis},)")
        self.mu_u = mu_u_tensor

        # Base world log-prior ell_0 (Sec. 3). `base_prior` is any callable
        # zeta -> log-density; when omitted, `gaussian_world_prior` at the given
        # mean and precision.
        #
        # The result is renormalized against the quadrature measure so that
        # sum_k w_k exp(ell_0) = 1 exactly on the truncated grid. An analytically
        # normalized density is not grid-normalized once the grid is cut off, and the
        # spec (Sec. 9.1) notes that phi_S is pinned in absolute terms, so the constant
        # is worth getting right even though it happens to be invisible here: B is
        # orthogonal to the constant function by construction, so <1, b_j> = 0 and an
        # additive shift in ell_0 reaches neither phi_u* (Eq. 16) nor the read-out.
        if base_prior is None:
            base_prior = gaussian_world_prior(prior_mean, prior_precision)
        if not callable(base_prior):
            raise TypeError("base_prior must be a callable zeta -> log-density")

        self.prior_mean = float(prior_mean)
        self.prior_precision = float(prior_precision)
        self.base_prior = base_prior
        self.base_prior_description = getattr(base_prior, "description", "custom")

        raw_log_prior = base_prior(self.zeta)
        if raw_log_prior.shape != self.zeta.shape:
            raise ValueError("base_prior must return one log-density per grid node")
        shift = raw_log_prior.max()
        prior_grid_mass = torch.sum(self.weights * torch.exp(raw_log_prior - shift))
        self.base_log_prior = raw_log_prior - shift - torch.log(prior_grid_mass)

        # retained so that probe networks can be respawned with the same configuration
        self._configuration = dict(
            num_nodes=num_nodes,
            grid_half_width=grid_half_width,
            base_prior=base_prior,
            prior_mean=prior_mean,
            prior_precision=prior_precision,
            theta_L=theta_L,
            num_atoms=num_atoms,
            lexical_strength=lexical_strength,
            mu_u=self.mu_u.clone(),
            theta_u=theta_u,
            basis_degree=basis_degree,
            sigma_lexical=sigma_lexical,
            sigma_state=sigma_state,
            sigma_utility=sigma_utility,
            exclusion_sharpness=exclusion_sharpness,
            device=device,
        )

    def respawn(self, **overrides):
        """A new network with this one's configuration, plus any overrides."""
        return LexicalPredictiveCodingNetwork(**{**self._configuration, **overrides})

    # === quadrature (Eq. 2) ===

    def inner(self, f, h):
        """<f, h> under the fixed quadrature measure (Eq. 2)."""
        return torch.sum(self.weights * f * h)

    def project(self, f):
        """B^T W f: contract a field onto the m basis directions (Eq. 3)."""
        return torch.mv(self.basis.T, self.weights * f)

    def squared_norm(self, f):
        return self.inner(f, f)

    @staticmethod
    def dot(a, b):
        """Ordinary inner product, for the utility vector (no quadrature measure)."""
        return torch.sum(a * b)

    # === clamped lexical field (Eqs. 5-6) ===

    def lexical_field(self, utterance):
        """phi_L = Lambda * chi_y, clamped by the observed utterance."""
        chi = exclusion_indicator(
            utterance,
            self.zeta,
            self.theta_L,
            sharpness=self.exclusion_sharpness,
        )
        return self.lexical_strength * chi

    # === generative pass (Sec. 4) ===

    def predict_lexical(self, phi_S):
        """g_L(phi_S) = ell_0 - phi_S (Eq. 9).

        The complement, transported to log-weights and reflected through the base world
        prior rather than through zero. Still an affine involution --
        g_L(g_L(x)) = x, so the structure of Eq. (9) is unchanged; ell_0 supplies the
        reference point the complement is taken about."""
        return self.base_log_prior - phi_S

    def predict_state(self, phi_u, theta_u=None):
        """g_S(phi_u) = theta_u * B phi_u (Eq. 10, with M = B theta_u of Sec. 8.4).

        ell_0 has moved to g_L, so the utility level's prediction of the situation is
        now purely its own contribution, with no tonic offset."""
        theta = self.theta_u if theta_u is None else float(theta_u)
        return theta * torch.mv(self.basis, phi_u)

    def residuals(self, phi_S, phi_u, phi_L, theta_u=None):
        """r_L, r_S, r_u of Eq. (11). r_y == 0 by Eq. (8) and is omitted; Appendix A
        gives its space (Eq. 4) and why omitting it is safe (Eq. A3)."""
        r_lexical = phi_L - self.predict_lexical(phi_S)
        r_state = phi_S - self.predict_state(phi_u, theta_u=theta_u)
        r_utility = phi_u - self.mu_u
        return r_lexical, r_state, r_utility

    def free_energy(self, phi_S, phi_u, phi_L, theta_u=None):
        """Bogacz's F, to be MAXIMIZED (Eq. 13). The log-sigma terms vanish at
        sigma = 1 and are constant in every variational quantity; they are carried
        only to mark where a precision-bearing version would reintroduce them."""
        r_lexical, r_state, r_utility = self.residuals(
            phi_S, phi_u, phi_L, theta_u=theta_u
        )
        quadratic = (
            -0.5 * self.squared_norm(r_lexical) / self.sigma_lexical
            - 0.5 * self.squared_norm(r_state) / self.sigma_state
            - 0.5 * self.dot(r_utility, r_utility) / self.sigma_utility
        )
        normalizer = -0.5 * (
            self.num_nodes * math.log(self.sigma_lexical)
            + self.num_nodes * math.log(self.sigma_state)
            + self.num_basis * math.log(self.sigma_utility)
        )
        return quadratic + normalizer

    # === read-out (Eq. 12) ===

    def read_out(self, phi_S):
        """Normalized belief q(zeta_k), computed in log-space (Sec. 9.2 caveat 2)."""
        shifted = phi_S - phi_S.max()
        unnormalized = torch.exp(shifted)
        return unnormalized / torch.sum(self.weights * unnormalized)

    # === closed-form targets (Sec. 6) ===

    def closed_form_fixed_point(self, utterance, theta_u=None):
        """Eqs. (15)-(16), generalized to arbitrary fixed sigma and to B of Sec. 8.4.

        Writing S = sigma_L + sigma_S, G = B^T W B and c = B^T W (ell_0 - phi_L),

            (I + sigma_u theta^2 G / S) phi_u* = mu_u + sigma_u theta c / S
            phi_S* = [sigma_L theta B phi_u* + sigma_S (ell_0 - phi_L)] / S

        The signs follow ell_0 having moved into g_L (Eq. 9): the utility level couples
        to the NET ell_0 - phi_L where it coupled to the sum ell_0 + phi_L before.

        At sigma = 1 and m = 1 these reduce to Eqs. (15)-(16) as written. G = I for the
        orthonormal basis built above, but the m x m system is solved in general so the
        result stays correct if a later phase supplies a non-orthonormal B.

        Because the model is linear-Gaussian while phi_L is clamped (Sec. 8.6), these
        are exact, and they are the correctness test for `infer`.
        """
        theta = self.theta_u if theta_u is None else float(theta_u)
        phi_L = self.lexical_field(utterance)
        total_sigma = self.sigma_lexical + self.sigma_state

        gram = self.basis.T @ (self.weights[:, None] * self.basis)
        # ell_0 now enters through g_L, so the utility level couples to the NET of prior
        # and lexicon, ell_0 - phi_L, where before it coupled to their sum.
        coupling = self.project(self.base_log_prior - phi_L)

        matrix = (
            torch.eye(self.num_basis, dtype=self.dtype, device=self.device)
            + self.sigma_utility * theta ** 2 * gram / total_sigma
        )
        rhs = self.mu_u + self.sigma_utility * theta * coupling / total_sigma
        phi_u_star = torch.linalg.solve(matrix, rhs)

        phi_S_star = (
            self.sigma_lexical * self.predict_state(phi_u_star, theta_u=theta)
            + self.sigma_state * (self.base_log_prior - phi_L)
        ) / total_sigma

        return phi_S_star, phi_u_star

    # === the literal listener (Part C, Part D) ===

    def literal_fixed_point(self, utterance):
        """phi_S of the LITERAL listener: the base world prior restricted by the entry.

        phi_S = ell_0 - phi_L, whose read-out is q_lit proportional to
        p_0 * exp(-Lambda chi_y): the prior times the soft semantic mask, and nothing
        else. This is the baseline the strengthening conditions of Text cell 4 Part C
        are measured against.

        It is a fixed point of THIS network, not an external construction. Eq. (15)
        reads phi_S* = [sigma_L theta_u B phi_u + sigma_S (ell_0 - phi_L)] / S with
        S = sigma_L + sigma_S, so sigma_S -> infinity sends it to ell_0 - phi_L: the
        situation level keeps no prior of its own and the lexical channel alone fixes
        the belief. That limit is what "literal" means here.

        NOT the same as theta_u = 0. At theta_u = 0 the r_S term still penalizes phi_S
        against a prediction of zero, which leaves phi_S* = (ell_0 - phi_L)/2 -- the
        literal posterior TEMPERED by one half, since the read-out (Eq. 12) is
        exponential. The half is a genuine temperature, so the theta_u = 0 belief is
        flatter than the literal one and holds systematically more mass at the ends of
        the scale. Both are reported below: this is the baseline, and theta_u = 0 is
        kept beside it as the temperature-matched control.
        """
        return self.base_log_prior - self.lexical_field(utterance)

    # === Hessian check (Eq. 21) ===

    def hessian_quadratic_form(self, v, t, theta_u=None):
        """(v, t)^T grad^2 F (v, t) of Eq. (21), in the sigma-weighted form

            -||v||^2/sigma_L - ||v - theta_u B t||^2/sigma_S - ||t||^2/sigma_u,

        obtained by reading off the directional derivative of each residual. Strictly
        negative for (v, t) != 0, for every theta_u and every Lambda (Sec. 8.1), and
        for arbitrary B and m (Sec. 8.4).
        """
        theta = self.theta_u if theta_u is None else float(theta_u)
        return (
            -self.squared_norm(v) / self.sigma_lexical
            - self.squared_norm(v - theta * torch.mv(self.basis, t)) / self.sigma_state
            - self.dot(t, t) / self.sigma_utility
        )

    # === recurrent inference (Sec. 7) ===

    @torch.no_grad()
    def infer(
        self,
        utterance,
        *,
        dt=0.05,
        tau_error=0.10,
        tau_state=1.00,
        max_steps=20000,
        derivative_tolerance=1e-10,
        patience=10,
        record_history=True,
        theta_u=None,
    ):
        """Integrate Eqs. (18)-(19) to the fixed point.

        tau_error << tau_state is required by Sec. 8.3, and dt <= tau_error by the
        Euler scheme. F is monitored but never differentiated: the messages are
        computed locally and F only records that they are ascending it.
        """
        if utterance not in UTTERANCES:
            raise ValueError(f"utterance must be one of {UTTERANCES}")
        if min(dt, tau_error, tau_state) <= 0:
            raise ValueError("dt and both time constants must be positive")
        if dt > tau_error:
            raise ValueError("require dt <= tau_error (Euler stability)")
        if tau_error >= tau_state:
            raise ValueError("require tau_error << tau_state (Sec. 8.3)")

        # Explicit Euler on this linear system is stable only while dt * rate < 2 for
        # every mode. The stiffest state mode is phi_u, whose rate is read off the
        # denominator of Eq. (16); phi_S relaxes at (1/sigma_L + 1/sigma_S).
        theta_probe = self.theta_u if theta_u is None else float(theta_u)
        gram = self.basis.T @ (self.weights[:, None] * self.basis)
        gram_max = float(torch.linalg.eigvalsh(gram).max())
        state_rate = max(
            1.0 / self.sigma_lexical + 1.0 / self.sigma_state,
            1.0 / self.sigma_utility + theta_probe ** 2 * gram_max / self.sigma_state,
        )
        if dt * state_rate / tau_state >= 2.0:
            raise ValueError(
                f"dt={dt} is too large for the stiffest state mode (rate "
                f"{state_rate:.3g}/tau_state): reduce dt below "
                f"{2.0 * tau_state / state_rate:.3g}, or raise tau_state"
            )

        theta = self.theta_u if theta_u is None else float(theta_u)
        phi_L = self.lexical_field(utterance)

        # initialization: the belief starts at the base prior, the utility state at its
        # own prior mean, and every error unit silent.
        phi_S = self.base_log_prior.clone()
        phi_u = self.mu_u.clone()

        eps_lexical = torch.zeros_like(phi_S)
        eps_state = torch.zeros_like(phi_S)
        eps_utility = torch.zeros_like(phi_u)

        history = []
        settled = 0
        converged = False

        for step in range(max_steps):
            r_lexical, r_state, r_utility = self.residuals(
                phi_S, phi_u, phi_L, theta_u=theta
            )

            # Eq. (18): tau_e * eps_dot = r - eps
            eps_lexical_dot = (r_lexical / self.sigma_lexical - eps_lexical) / tau_error
            eps_state_dot = (r_state / self.sigma_state - eps_state) / tau_error
            eps_utility_dot = (
                r_utility / self.sigma_utility - eps_utility
            ) / tau_error

            eps_lexical = eps_lexical + dt * eps_lexical_dot
            eps_state = eps_state + dt * eps_state_dot
            eps_utility = eps_utility + dt * eps_utility_dot

            # Eq. (19): state units ascend F. The lexical error arrives inhibitorily
            # because g_L' = -I.
            phi_s_dot = (-eps_state - eps_lexical) / tau_state
            phi_u_dot = (
                -eps_utility + theta * self.project(eps_state)
            ) / tau_state

            max_derivative = max(
                float(eps_lexical_dot.abs().max()),
                float(eps_state_dot.abs().max()),
                float(eps_utility_dot.abs().max()),
                float(phi_s_dot.abs().max()),
                float(phi_u_dot.abs().max()),
            )

            if record_history:
                history.append(
                    {
                        "step": step,
                        "free_energy": float(
                            self.free_energy(phi_S, phi_u, phi_L, theta_u=theta)
                        ),
                        "max_derivative": max_derivative,
                        "phi_u": phi_u.tolist(),
                    }
                )

            settled = settled + 1 if max_derivative < derivative_tolerance else 0
            if settled >= patience:
                converged = True
                break

            phi_S = phi_S + dt * phi_s_dot
            phi_u = phi_u + dt * phi_u_dot

            if not (torch.isfinite(phi_S).all() and torch.isfinite(phi_u).all()):
                raise FloatingPointError(
                    "non-finite activity: reduce dt or lower the lexical strength"
                )

        r_lexical, r_state, r_utility = self.residuals(
            phi_S, phi_u, phi_L, theta_u=theta
        )

        return {
            "utterance": utterance,
            "converged": converged,
            "steps": len(history),
            "phi_S": phi_S,
            "phi_u": phi_u,
            "phi_L": phi_L,
            "q": self.read_out(phi_S),
            "free_energy": self.free_energy(phi_S, phi_u, phi_L, theta_u=theta),
            "error_units": {
                "lexical": eps_lexical,
                "state": eps_state,
                "utility": eps_utility,
            },
            "residuals": {
                "lexical": r_lexical,
                "state": r_state,
                "utility": r_utility,
            },
            "theta_u": theta,
            "history": history,
        }

    # === the slow parameter (Sec. 7 Eq. 20, Sec. 8.5) ===

    def theta_u_gradient(self, result):
        """dF/dtheta_u = phi_u <eps_S, b> (Eq. 20).

        This is Bogacz Eq. (25), dF/dtheta = eps_u * phi, the error BELOW times the
        state ABOVE, with phi_S being a field rather than a scalar, which is the only
        reason the contraction against b appears. Eq. (29) is the same quantity for a
        nonlinear h; ours is h = identity."""
        return self.dot(
            result["phi_u"], self.project(result["error_units"]["state"])
        )

    def learn_theta_u(
        self,
        utterances,
        *,
        tau_theta=20.0,
        num_updates=50,
        tau_state=1.00,
        **infer_kwargs,
    ):
        """Integrate Eq. (20), tau_theta * d(theta_u)/dt = dF/d(theta_u).

        There is no learning rate here, and none is needed. Bogacz gives the GRADIENT
        (his Eqs. 25 and 29) and states that parameters move proportionally to it; the
        constant of proportionality is a time constant of exactly the kind Eqs. (18)
        and (19) already carry for the error and state units. So theta_u takes a third,
        tau_theta, and the architecture requires

            tau_error << tau_state << tau_theta,

        the slow end of the same ordering whose fast end Sec. 8.3 imposes. tau_theta
        sets the RATE only: the fixed point of the flow is the stationary point of
        F~(theta_u) = max_x F(x, theta_u) and does not depend on it.

        Requires <mu_u, sum_y c_y> != 0 (Appendix B). mu_u != 0 (Sec. 3) is
        necessary but not sufficient; otherwise F~ has no stationary point and this
        flow diverges.
        """
        if tau_theta <= tau_state:
            raise ValueError(
                "require tau_state << tau_theta: the fast subsystem must equilibrate "
                "between theta_u updates for Eq. (20) to be ascent on F~"
            )
        # necessary part only; the full condition is Appendix B's
        if float(self.mu_u.abs().max()) == 0.0:
            raise ValueError(
                "mu_u = 0 leaves theta_u unidentifiable (Sec. 3). This test is the "
                "necessary part only: Appendix B requires <mu_u, sum_y c_y> != 0, "
                "which a nonzero mu_u on a degenerate ray still violates "
                "(Text cell 5 Part B Sec. 6)."
            )

        # |theta_u| grows during the ascent (theta_u* < 0 here) and the fast subsystem stiffens
        # with theta_u^2 (rate 1/sigma_u + theta_u^2 lambda_max(G)/sigma_S, cf. `infer`),
        # so dt is chosen per update from that bound rather than fixed.
        gram = self.basis.T @ (self.weights[:, None] * self.basis)
        gram_max = float(torch.linalg.eigvalsh(gram).max())
        automatic_dt = "dt" not in infer_kwargs

        trace = []
        for update in range(num_updates):
            if automatic_dt:
                rate = max(
                    1.0 / self.sigma_lexical + 1.0 / self.sigma_state,
                    1.0 / self.sigma_utility
                    + self.theta_u ** 2 * gram_max / self.sigma_state,
                )
                infer_kwargs["dt"] = min(0.02, 0.5 * tau_state / rate)

            gradient = 0.0
            free_energy = 0.0
            for utterance in utterances:
                result = self.infer(
                    utterance, record_history=False, tau_state=tau_state, **infer_kwargs
                )
                if not result["converged"]:
                    raise RuntimeError(
                        f"fast subsystem did not equilibrate for {utterance!r}; "
                        "the two-timescale argument of Sec. 8.5 does not apply"
                    )
                gradient += float(self.theta_u_gradient(result))
                free_energy += float(result["free_energy"])

            gradient /= len(utterances)
            free_energy /= len(utterances)

            trace.append(
                {"update": update, "theta_u": self.theta_u,
                 "gradient": gradient, "free_energy": free_energy}
            )
            self.theta_u = self.theta_u + (tau_state / tau_theta) * gradient

        return trace
````

## <a id="tc4" name="tc4"></a>Text cell 4: evaluation

The evaluation has four parts.

### <a id="tc4-parta" name="tc4-parta"></a>Part A
**A** tests every testable claim of the preceding specification: the closed
forms of Eqs. (15)-(16), the fixed-point identities of Eq. (17), monotonicity of $\mathcal F$
(Sec. 8.2), negative definiteness of the Hessian (Eq. 21) across $\theta_u$, the independence
of the relaxation rate from $\varphi_L$ (Sec. 8.1(iii)), agreement of the analytic messages
with finite differences (Sec. 7), stability of the belief under grid refinement (Sec. 1), and
compatibility with a smooth lexical field (Sec. 9.2).

Two of them are tested in ways that need comment. **Grid refinement** is a claim about *convergence* in $K$,
so it is tested on the tail of the sequence, $\lvert\mathbb E[s]_{801}-\mathbb E[s]_{401}\rvert$,
rather than on the gap between two coarse grids. The distinction matters here because the rate is
set by something Sec. 1 does not mention: $\chi_y$ is a step, and trapezoidal quadrature on a
discontinuous integrand converges at $O(h)$ with the step's position jittering against the nodes as
$K$ changes, which is §9.2 caveat 1 arriving as a measurement. The tail is
$1.6\times10^{-3}$ under the hard $\chi_y$ and $1.8\times10^{-6}$ under the smooth one, against a
coarse $51$–$401$ spread of $1.0\times10^{-2}$; the cell reports all three, and the smooth mask is
the control that identifies which of them is the model and which is the quadrature.

**Everything else in this cell is reported at $\theta_u=1$**, which is neither the learned optimum
nor its sign, since $\theta_u^\ast=-28.4375$. `theta_u_dependence` therefore recomputes every headline
quantity at the control, at $\pm1$, and at $\theta_u^\ast$, so that no conclusion rests on the
pinned value without that being visible. Two things it shows are worth carrying: the *all*$-$*some*
$\varphi_S$ contrast *grows* with $\lvert\theta_u\rvert$ ($9.6995\to10.9524\to13.9124$), so unlike
$\mu_u$ and $\ell_0$ the utility level does move contrasts; and the difference between $\theta_u$
and $-\theta_u$ is largest near $\lvert\theta_u\rvert=1$ and decays toward $\theta_u^\ast$, because
only the $\mu_u$-carried linear term changes sign while the $\theta_u^2$ term does not. The pinned
value is close to the worst case for sign sensitivity.

### <a id="tc4-partb" name="tc4-partb"></a>Part B
**B** runs the network and sweeps $\Lambda$, confirming that a hard lexical constraint is
recovered as the $\Lambda\to\infty$ limit.

### <a id="tc4-partc" name="tc4-partc"></a>Part C
**C** asks whether the network produces a scalar implicature. The question needs an operational
form, and we adopt one rather than derive it. Three quantities have to be named first, because the
criterion is a relation between them.

**The pragmatic listener and the literal baseline.** Write $q_y$ for the settled read-out under
utterance $y$, the pragmatic listener, written $q_H$ where the contrast needs a name. The baseline
it is measured against, and the dashed line in every panel of Part D, is $q_{\mathrm{lit}}$, the
**untempered literal posterior**: the belief $\varphi_S=\ell_0-\varphi_L$ that the base world prior
and the lexical entry fix between them, whose read-out is $p_0e^{-\Lambda\chi_y}$ normalized, the
prior restricted by the entry and nothing else. It is a fixed point of this network and not a
construction from outside it, being what Eq. (15) returns as $\sigma_S\to\infty$, where the
situation level keeps no prior of its own and the lexical channel alone fixes the belief. Reading
implicature strength off the gap between a literal and a pragmatic listener follows the practice of
the probabilistic accounts (Frank & Goodman, 2012; Goodman & Frank, 2016).

**The $\theta_u$ control is a third quantity, not the baseline.** Eq. (15) at $\theta_u=0$ leaves
$\varphi_S^\ast=(\ell_0-\varphi_L)/2$, which is $q_{\mathrm{lit}}$ tempered by one half, the
halving surviving into the belief because the read-out of Eq. (12) is exponential. The two differ
by that temperature and by nothing else. $q_{\mathrm{lit}}$ is the baseline because it is the
literal listener the criterion names; the control is kept in view because it holds the temperature
fixed, so what separates it from $q_y$ is the utility level alone. The shift against
$q_{\mathrm{lit}}$ therefore carries both the tempering and the utility level, and Part D reports
the two masses side by side so that the parts can be read off.

**The all-region is the continuous form of the top state.** $\theta_L$ is fixed by
$\varsigma(-\theta_L)=1/2n$ at $n=10$ (Eq. A5), so the region $\zeta>\theta_L$ is $s>0.95$, which on
a ten-atom scale contains the atom $s=1$ and no other. $P(\text{all-region})$ of Eq. (27) is
accordingly what $q(s{=}10)$ names on a discrete scale.

**The criterion.** We take scalar strengthening to be the conjunction of two conditions on the
all-region mass under *some*: that
$q_H(\text{all}\mid\textit{some})<q_{\mathrm{lit}}(\text{all}\mid\textit{some})$, the hierarchy
lowering the mass the entry and prior already assign; and that
$q_H(\text{all}\mid\textit{some})<\tfrac12$, the strengthened reading being the majority outcome
and not merely the suppressed alternative. Neither is sufficient alone. The second can hold on
literal conditioning by itself, since the non-all states stand together against the single all
state; the first can hold while *all* remains the most probable single outcome. Writing
$\Delta_{\textit{some}}=P(\text{all-region};q_{\textit{some}})-P(\text{all-region};q_{\mathrm{lit},\textit{some}})$
for the **shift**, the first condition is $\Delta_{\textit{some}}<0$. All of this is a stipulation
about how such an effect would have to show up *in this model*, not a definition of scalar
implicature, and every verdict below is relative to it. Eq. (37) is this shift, written out where
Text cell 6 maps it.

**The criterion is not met under the elicited prior.** Under the Gaussian prior
$\Delta_{\textit{some}}=+0.0098$ at $\theta_u=1$: the settled belief holds more all-region mass
than the literal listener, so the first condition fails while the second holds comfortably. The
utility level's own contribution there is negative, $-0.0077$; what carries the shift positive is
the tempering, $+0.0175$. Ascending $\mathcal F$ in $\theta_u$ carries the shift toward the
criterion without reaching it, to $+0.0009$ at the settled $\theta_u$ of the learning probe and
$+0.0008$ at the stationary point $\theta_u^\ast=-28.4375$ of Appendix B. Under the delta-like
prior of Part D, already concentrated on the all-region, the first condition is met and by a wide
margin, and the second is the one that fails. **Across Part D's five priors the two conditions
separate every row between them, and the conjunction holds under none.** It is not out of the
model's reach: Text cell 6 sweeps the same prior family against $\Lambda$ and finds a band of the
plane where both hold at once.

*Nor is the first condition, where it is met, specific to* some. Under the Gaussian prior it is met
by *all* ($-0.3144$ at $\theta_u=1$, $-0.4557$ at $\theta_u^\ast$) and by neither *no* ($+0.0041$,
$+0.0003$) nor *some*. Under the delta-like prior *no* and *all* are both saturated, holding
$0.0000$ and $1.0000$ of the all-region at baseline, so their zero shifts are saturation and not
restraint; only *some* had room to move, and moved.

Two things have to be stated about the row where the first condition is met, because they are
commitments anyone reading that outcome as scalar implicature thereby takes on.

*Firstly, strengthening is prior-relative.* Part D's table carries the five shifts and the sign
follows the prior across them. And it is the prior doing that rather than the raised $\Lambda$ the
delta-like row has to carry: with the Gaussian prior restored at $\Lambda=512$,
$\Delta_{\textit{some}}=+0.0373$, further from the criterion than the $+0.0098$ at $\Lambda=8$
rather than nearer to it. So the effect is not a property of the entry for *some*; it is a property
of that entry against a particular world belief, and it appears exactly where the prior most
favours *all*. A reader who wants scalar strengthening to follow from the lexical scale alone
cannot have it here.

*Secondly, the mechanism is non-discriminative towards all utterances in the inventory.* Differencing the closed form
of Eq. (16) against its own $\theta_u=0$ case, which is the control defined above and not the
baseline, gives

$$\varphi_S^\ast(\theta_u)-\varphi_S^\ast(0)=\tfrac{\theta_u}{2}\,B\varphi_u^\ast(\theta_u),
\tag{23}$$

The identity is exact to $1.8\times10^{-15}$, and growing $\lvert\theta_u\rvert$ drives

$$\varphi_S^\ast\;\longrightarrow\;\tfrac12\big(I+BB^{\mathsf T}W\big)(\ell_0-\varphi_L),
\tag{24}$$

which is the $\theta_u=0$ field with its component in $\operatorname{span}B$ **doubled** and its
orthogonal complement left alone. The limit is not a loose bound: it predicts
$\Delta_{\textit{some}}=-0.5217$ under the delta-like prior and $\theta_u^\ast$ delivers $-0.5222$,
and $\Delta_{\textit{some}}<0$ holds there for *every* $\theta_u\neq0$ of either sign, at most
$-0.0664$ as $\theta_u\to0$ and saturating at the limit from both directions. So the whole of the effect is an amplification of the utterance's own projection onto a
two-dimensional utility basis. Under the delta-like prior those projections, $c_y=B^{\mathsf T}W(\ell_0-\varphi_L)$ in the notation
Appendix B carries, are $c_{\textit{some}}=[+961.66,-529.77]$ and $c_{\textit{all}}=[+961.66,+208.51]$.
The two columns of $B$ at $m=2$ (§3.3) are an odd and an even profile in $\zeta$, so the two entries
are identical in the odd (tilt) coordinate and differ only in the even (width) one, which is the only
coordinate in which they differ at all and the entire source of the *some*/*all* asymmetry above. Doubling a negative width
coordinate lowers both tails, the all-region among them.

*What the mechanism is not.* Three properties about the architecture hold regardless of the sign of $\theta_u$.
$\varphi_L$ is a function of the observed utterance alone (Eqs. 5–6); $g_S$ reaches the world belief
only through $\varphi_u$, which is driven by $\varepsilon_S$, which is driven by $\varphi_L$
(Eqs. 10–11, 19); and no term in Eq. (13) involves any utterance other than the observed one. Nothing
in the chain of prediction represents that *all* was available and not chosen, and no quantity is compared against
an alternative at any point in an inference pass. Where the first condition is met it is therefore met *without* a
within-trial alternatives computation, by amplification of a single entry's own projection. (These three facts settle the *absence* of an alternatives computation and nothing
more: they are indifferent to the sign of the shift, and they do not say what the shift is made of.
That it is amplification of a low-rank component, and not shrinkage toward $g_S$, is the separate
measurement above.)

*Note that the above does not suggest the model predicts that alternative competition does not contribute to scalar strengthening* A phenomenon that holds under
one name (i.e. scalar stenghtening) is not thereby the consequence of one mechanism. The result here does not show that within-trial reasoning over an
alternatives space is absent from human scalar strengthening, and nothing in this implementation
could show that: the model has no such space to begin with. What the implementation does establish
is narrower, and is about the first condition alone: **the movement of $q$-mass out of the
all-region need not be the consequence of within-trial alternative inference**, since an
architecture containing none produces it. It produces it under the one prior that gives the *all*
reading the most support, which is where standard RSA has the most to overcome, and produces it
nowhere else among Part D's five. The second condition is not met alongside it there, $q_H$ still
holding $0.6080$ of the all-region; it is met alongside it at $22$ of the $121$ cells Text cell 6
sweeps, where the same prior family is paired with a stronger lexicon, and the delta-like row is
that band's neighbour rather than a member of it. So the conjunction is reachable in this
architecture without an alternatives space, and Part D's priors are simply not where it is
reached. Regarding whether direct alternative competition also contributes, and in what proportion, the implementation at this phase makes no prediction about either way.

Appendix B's *Alternative Spaces* section takes up the further question of what would be the nature of an implicature reaching
the belief through the learned $\theta_u^\ast$, namely conventionalized by exposure rather than
computed. This nature is a separate commitment from the one this part discussed.

### <a id="tc4-partd" name="tc4-partd"></a>Part D
**D** repeats the demonstration and the implicature probe under five base world priors: a
Gaussian in $\zeta$ (equivalently, $s$ logit-normal), a literal $\mathrm{Uniform}(0,1)$ over the
proportion, two skewed
$\mathrm{Beta}$ priors, and a delta-like $\mathrm{Beta}(64,1)$ concentrated on the all-region. The
last carries its own $\Lambda$ and is the case Part C leans on; the reason it needs one is given
with it below. The two beliefs being compared, the $\theta_u$ control kept beside them, the
all-region and the two conditions are all as Part C defines them.

A prior specified over $s$ must be pushed forward to $\zeta$, multiplying by
$\mathrm ds/\mathrm d\zeta=s(1-s)$. For a $\mathrm{Beta}$ prior the Jacobian absorbs one factor
from each side and the log-density collapses to
$\alpha\log s+\beta\log(1-s)-\log B(\alpha,\beta)$, writing $\beta$ for the second shape parameter. The
exponents are $\alpha,\beta$ rather than the usual $\alpha-1,\beta-1$, and that shift *is* the Jacobian. Each prior is then renormalized
against the quadrature measure, since an analytically normalized density is not grid-normalized
once the grid is truncated. Because a prior flat in $s$ is not flat in $\zeta$, the flat condition
is $\mathrm{Beta}(1,1)$ and not a low-precision Gaussian: sending the Gaussian's precision to zero
inflates the variance in $\zeta$, which the sigmoid squashes into bimodal mass at the endpoints, which is the opposite of flat.

This push-forward runs in the **opposite direction** to the transform the figures apply. The prior
enters multiplied by $\mathrm ds/\mathrm d\zeta$; a belief is plotted as a density in $s$ by
dividing by it, per the remark at Eq. (25). Both figures draw the base prior as a grey line, and
that the flat panel plots as a horizontal line at $1.0$ exercises the whole chain of push-forward,
renormalization and plotting transform in a single check.

**The delta-like row, and why it needs its own $\Lambda$.** The other four priors are all diffuse
over the scale, so none of them puts the criterion under any pressure: hearing *some* under a belief
that is already spread out, there is little all-region mass to move away from. The sharp case is the
one worth testing. A literal $P_0(s{=}1)=1$ is not representable, since $s$ lies in the *open* interval,
$\zeta=\operatorname{logit}(1)$ is off the grid, and $\ell_0$ would be $-\infty$ almost everywhere, making
$r_S$ infinite. We therefore use the limit family $\mathrm{Beta}(\alpha,1)$ that Text cell 6 sweeps.
$\alpha=64$ puts the mode in $\zeta$ at $\log\alpha=4.16$, inside the all-region and clear of the
grid ceiling Text cell 6 describes; it carries $95.7\%$ of its prior mass in the all-region and only
$0.95\%$ on the top node, and the three utterances are still $0.0061$ apart in $\mathbb E[s]$ there,
whereas by $\alpha=1024$ they are $0.0002$ apart and the probe would be measuring the truncation. $\Lambda$ must rise with it, or the prior
simply overrides the entry and the probe measures the override instead of the criterion: Eq. (41)
puts the threshold at $\Lambda_{\mathrm{crit}}\approx\alpha\log2n=192$, and $\Lambda=8\alpha=512$,
about $2.7\times$ that, holds every leak below $8\times10^{-71}$. That row alone varies $\Lambda$,
which is why the figure below draws only the four that share $\Lambda=8$, since panels on a shared axis
have to be comparable.

$\ell_0$ is fixed, and it sits in $g_L$ (Eq. 9) rather than in $g_S$: it
reaches the situation and utility levels only through the utterance-independent combination
$\ell_0-\varphi_L$ of Eqs. (15)–(16). $\varphi_S^\ast$ is affine in it with a coefficient that does
not depend on the utterance, so the $\varphi_S$ contrasts are exactly $\ell_0$-invariant, verified to $2.5\times10^{-14}$ across all five priors evaluated at a common $\Lambda$. The sweep is
accordingly a semantic test rather than a numerical one: Eq. (21) does not involve $\ell_0$ any more
than it involves $\varphi_L$, so every result in Sec. 8 holds unchanged and the closed forms stay
exact under each prior ($\max\lvert\varphi_S-\varphi_S^\ast\rvert=6.0\times10^{-11}$).

**How far the baseline and the control separate.** They differ by temperature alone, and the sweep
shows how wide that is where the prior is diffuse: under *some* the all-region mass of
$q_{\mathrm{lit}}$ runs $0.0016$, $0.0504$, $0.0001$, $0.1368$ and $0.9568$ down the five rows
below, against $0.0191$, $0.1357$, $0.0066$, $0.2356$ and $0.8982$ for the tempered belief, a
factor of twelve at the Gaussian row. Tempering flattens, and a flatter belief holds more mass at
the ends of the scale, which is where the all-region sits. The two coincide only where the prior
leaves the belief little room to flatten into, which is the delta-like row.

What the sweep establishes is that $\Delta_{\textit{some}}$ is prior-relative:
$+0.0098$ (Gaussian), $+0.0880$ (flat), $+0.0032$ ($\mathrm{Beta}(1,3)$), $+0.1130$
($\mathrm{Beta}(3,1)$), $-0.3488$ (delta-like), which is the first of the two commitments Part C
records. One
might expect the sign to be constant, on the ground that a quantity acting identically on every
utterance can move no contrast between them. The premise is right and the inference does not
follow: $\ell_0$-invariance is a property of the $\varphi_S$ contrasts, and
$\Delta_y$ is not one of them. It is a difference of $q$-masses (Eq. 27) taken after the softmax of
Eq. (12), and the *Reporting statistics* note below is the general form of the point: a statistic nonlinear in $\varphi_S$ need not inherit an invariance the $\varphi_S$ contrasts have. This is the
same failure mode Text cell 5 Part B §4 argues about under $\mu_u$, running in the opposite
direction: there an invariant contrast produces a drifting statistic, here it produces a change of
sign.

**Where each of Part C's two conditions is met.**

| prior | $q_{\mathrm{lit}}$ | $q_H$ | $\Delta_{\textit{some}}$ | first | second |
|---|---|---|---|---|---|
| Gaussian | $0.0016$ | $0.0114$ | $+0.0098$ | not met | met |
| flat | $0.0504$ | $0.1384$ | $+0.0880$ | not met | met |
| $\mathrm{Beta}(1,3)$ | $0.0001$ | $0.0033$ | $+0.0032$ | not met | met |
| $\mathrm{Beta}(3,1)$ | $0.1368$ | $0.2498$ | $+0.1130$ | not met | met |

**The conjunction holds under none of the five.** The first condition is met by the delta-like

### <a id="tc4-reporting-statistics" name="tc4-reporting-statistics"></a>Reporting statistics

Every number quoted above is a functional of Eq. (12).
None is a model quantity: the model's output is the density $q$, and each statistic below is a
summary of $q$, defined here so that what is being reported is unambiguous.

We call

$$\mathbb E[s]
=\int_0^1 s\,q_s(s)\,\mathrm ds
=\int_{\mathbb R}\varsigma(\zeta)\,q(\zeta)\,\mathrm d\zeta
\;\approx\;\sum_k w_k\,\varsigma(\zeta_k)\,q(\zeta_k)
\tag{25}$$

the **expected proportion** under $q$. The two integrals agree because
$q_s(s)\,\mathrm ds=q(\zeta)\,\mathrm d\zeta$. Note where the change of variables lands: for an
*expectation* it acts on the integrand, so Eq. (25) is computed directly on the $\zeta$-grid with
**no Jacobian factor**. This is not true of the *density* $q_s(s)=q(\zeta)/\big(s(1-s)\big)$, which
the figures plot and which does carry one. The two are easily confused, and applying the density's
correction to the expectation, or omitting it from the density, is the natural error.

We call

$$\mathrm{sd}(s)=\Big(\sum_k w_k\,q(\zeta_k)\big(\varsigma(\zeta_k)-\mathbb E[s]\big)^2\Big)^{1/2}
\tag{26}$$

the **dispersion** of $q$, reported on the scale of $s$.

For a measurable region $R$ on the scale, we call

$$P(R)=\sum_k w_k\,q(\zeta_k)\,\mathbb 1[\zeta_k\in R]
\tag{27}$$

the **$q$-mass over $R$**. It is a number in $[0,1]$, since Eq. (12) integrates to $1$ over the whole
grid. It is not the field $\varphi_S$, which is unnormalized log-weights and carries no such
interpretation; the exponential and the normalizer of Eq. (12) separate the two.

The qualifier $q$ is not decorative, because two other masses appear in this notebook, and neither of
them is the $q$-mass. The *quadrature mass* of a region is the measure the grid itself carries there, the weights
$w_k$ of Eq. (2), defined with no density in play, and it is what the endpoints $s=0,1$ lack. The *prior
mass* of a region integrates the base prior $p_0$ over it, and is what the grid renormalization of
$\ell_0$ divides out. Only $q$-mass is a statement about what the model concluded.

Three regions are used. $P(\text{all-region})$ takes $R=\{\zeta>\theta_L\}$, where *all* is true, and
$P(\text{no-region})$ takes $R=\{\zeta\le-\theta_L\}$. We call the $q$-mass over the region the observed
entry excludes, $\langle q,\chi_y\rangle$, the **leak**; it coincides with $P(E_y)$ for a step
$\chi_y$, is its soft-weighted analogue for a smooth one, and vanishes as $\Lambda\to\infty$ by
Eq. (15).

**These statistics are nonlinear in $\varphi_S$, so differences between them are not differences
in $\varphi_S$.** Eq. (25) composes a softmax with $\varsigma$; adding one and the same field to
two different beliefs therefore does not move their $\mathbb E[s]$ equally. Any claim about a
*contrast* between utterances must be checked against $\varphi_S$ directly. Where a reported
contrast in $\mathbb E[s]$ drifts while the underlying $\varphi_S$ contrast is provably invariant,
the nonlinearity is sufficient to account for the drift on its own. That is the situation under $\mu_u$, and the reason Text cell 5 Part B §4 argues against reading its contrast columns as an
effect of that parameter.

### <a id="tc4-integration-cost" name="tc4-integration-cost"></a>Integration cost and conditioning

§8.3 gives $\mathrm{d}t\le\tau_\varepsilon$ for the Euler scheme. That is necessary but not
sufficient. The binding constraint is usually the stability bound below, which the implementation
enforces and which is worth recording here because it governs what the evaluation can afford to run.

Explicit Euler on a linear system is stable only while $\mathrm{d}t\,\lambda<2$ for every eigenvalue.
The $\varphi_S$ mode relaxes at $1/\sigma_L+1/\sigma_S$, a constant. The $\varphi_u$ mode relaxes at

$$\frac{1}{\sigma_u}+\frac{\theta_u^{2}\,\lambda_{\max}(B^{\mathsf T}WB)}{\sigma_S},
\tag{28}$$

which is **quadratic in $\theta_u$**: $r_S$ sees $\varphi_u$ only through $\theta_u B$, so curvature
in that direction scales as $\theta_u^2$, and the table below is accordingly written on
$\lvert\theta_u\rvert$; $B$ is orthonormal, giving
$\lambda_{\max}(B^{\mathsf T}WB)=1$ exactly. Learning $\theta_u$ therefore stiffens the fast
subsystem as it proceeds:

| $\lvert\theta_u\rvert$ | 1 | 10 | 28.44 | 50 |
|---|---|---|---|---|
| $\varphi_u$ rate | 2.0 | 101 | **809.7** | 2501 |
| stiffness ratio | 1.0 | 50.5 | **404.8** | 1250.5 |
| max $\mathrm{d}t$ at $\tau_\varphi=1$ | 1.0 | 0.0198 | **0.0025** | 0.0008 |

At the stationary point $\theta_u^\ast=-28.4375$ the system is $\sim\!405\times$ stiff and each
inference costs some $405\times$ what it did at $\theta_u=1$. This is why `learn_theta_u` chooses
$\mathrm{d}t$ per update from the bound rather than taking a fixed one, why the slow flow is
expensive, and why Part A tests the stationary point in closed form (exact by §8.6) instead of
integrating to it.

The point generalizes past the numerics. §8.1 establishes that $\nabla^2\mathcal F$ is negative
definite for **every** $\theta_u$. Definiteness says nothing about *conditioning*: it secures
convergence, while the condition number, of order
$(1+\theta_u^2)/2$ here and $404.8$ at $\theta_u^\ast$, governs whether convergence is reachable in
practice. (At $\sigma_L=\sigma_S=\sigma_u=1$ the $\varphi_S$ rate is $1/\sigma_L+1/\sigma_S=2$, so
that expression and the stiffness ratio above coincide; they are different quantities and would
separate at unequal $\sigma$.) Slow-parameter learning and fast-subsystem conditioning are coupled,
and §8 does not say so.

````python
# === Code Cell 2: EVALUATION, covering specification checks, demonstration,
#     and the semantic question ===
#
# Part A tests every claim the specification makes that is testable.
# Part B shows what the network does.
# Part C asks whether it produces a scalar implicature. Under the criterion Text
# cell 4 Part C adopts, it does; that cell states what the reading commits us to.

import math
import torch


# ============================================================================
# A. SPECIFICATION CHECKS
# ============================================================================

def check_specification(net, *, dt=0.02, seed=0, verbose=True):
    """Test each testable claim of the specification. Returns {claim: (ok, detail)}."""
    torch.manual_seed(seed)
    results = {}

    # Sec. 6 / Sec. 8.6: `infer` must reproduce the closed forms exactly, since the
    # model is linear-Gaussian while phi_L is clamped.
    worst_state = worst_utility = 0.0
    worst_identity_1 = worst_identity_2 = 0.0
    worst_monotone = math.inf
    worst_q_mass = 0.0
    for utterance in UTTERANCES:
        result = net.infer(utterance, dt=dt)
        if not result["converged"]:
            results["convergence"] = (False, f"{utterance} did not converge")
        phi_S_star, phi_u_star = net.closed_form_fixed_point(utterance)
        worst_state = max(worst_state, float((result["phi_S"] - phi_S_star).abs().max()))
        worst_utility = max(
            worst_utility, float((result["phi_u"] - phi_u_star).abs().max())
        )

        # Eq. (17): eps_S = -eps_L and eps_u = theta_u B^T W eps_S
        units = result["error_units"]
        worst_identity_1 = max(
            worst_identity_1, float((units["state"] + units["lexical"]).abs().max())
        )
        worst_identity_2 = max(
            worst_identity_2,
            float(
                (units["utility"] - net.theta_u * net.project(units["state"]))
                .abs()
                .max()
            ),
        )

        # Sec. 8.2: F is non-decreasing along the flow
        trace = [entry["free_energy"] for entry in result["history"]]
        worst_monotone = min(
            worst_monotone, min(trace[i + 1] - trace[i] for i in range(len(trace) - 1))
        )

        # Eq. (12): the read-out integrates to 1
        worst_q_mass = max(
            worst_q_mass, abs(float((net.weights * result["q"]).sum()) - 1.0)
        )

    results.setdefault("convergence", (True, "all three utterances settled"))
    results["fixed point matches Eqs. (15)-(16)"] = (
        max(worst_state, worst_utility) < 1e-8,
        f"max |phi - phi*| = {max(worst_state, worst_utility):.2e}",
    )
    results["Eq. (17) eps_S = -eps_L"] = (
        worst_identity_1 < 1e-8, f"max residual = {worst_identity_1:.2e}"
    )
    results["Eq. (17) eps_u = theta_u B^T W eps_S"] = (
        worst_identity_2 < 1e-8, f"max residual = {worst_identity_2:.2e}"
    )
    results["Sec. 8.2  F non-decreasing"] = (
        worst_monotone > -1e-9,
        f"min step = {worst_monotone:+.2e} (roundoff at the fixed point)",
    )
    results["Eq. (12) read-out normalized"] = (
        worst_q_mass < 1e-10, f"max |1 - q-mass| = {worst_q_mass:.2e}"
    )

    # Sec. 8.1/8.4: the Hessian is negative definite, and Sec. 8.1(ii) says so
    # for every theta_u and Lambda. Run at m = num_basis, so the form tested is
    # Eq. (22) of Sec. 8.4, which reduces to Eq. (21) at m = 1.
    worst_form = -math.inf
    for theta in [-50.0, -1.0, 0.0, 1.0, 50.0]:
        for _ in range(400):
            v = torch.randn(net.num_nodes, dtype=net.dtype)
            t = torch.randn(net.num_basis, dtype=net.dtype)
            scale = math.sqrt(float(v.pow(2).sum() + t.pow(2).sum()))
            worst_form = max(
                worst_form,
                float(net.hessian_quadratic_form(v / scale, t / scale, theta_u=theta)),
            )
    results["Eq. (22) Hessian negative definite"] = (
        worst_form < 0.0, f"max over unit directions = {worst_form:+.3e}"
    )

    # Sec. 8.1(iii): phi_L enters r_L additively and so shifts the maximizer without
    # touching the Hessian. The testable consequence is about the RELAXATION RATE, not
    # the step count: phi_L moves the fixed point, so different utterances start at
    # different distances from it and reach an absolute tolerance at different times.
    # What must agree is the asymptotic contraction factor, which is set by the
    # eigenvalues of the Hessian alone.
    contraction = {}
    for utterance in UTTERANCES:
        trace = [entry["max_derivative"] for entry in net.infer(utterance, dt=dt)["history"]]
        tail = trace[len(trace) // 2 : len(trace) - 5]          # asymptotic regime
        span = len(tail) - 1
        contraction[utterance] = (tail[-1] / tail[0]) ** (1.0 / span)
    rates = list(contraction.values())
    spread = max(rates) - min(rates)
    # tolerance reflects the precision of a contraction rate estimated by finite
    # differences off a discrete trace, not the sharpness of the claim: the rates
    # below agree to eight significant figures.
    results["Sec. 8.1(iii) relaxation rate independent of phi_L"] = (
        spread < 1e-6,
        "contraction/step " + ", ".join(f"{y}={r:.9f}" for y, r in contraction.items()),
    )

    # Sec. 7: the analytic messages are the gradients of Eq. (13). Checked against
    # central differences at an off-equilibrium point, which is where a sign error
    # would show.
    utterance = "some"
    phi_L = net.lexical_field(utterance)
    settled = net.infer(utterance, dt=dt)
    phi_S = settled["phi_S"] + 0.3 * torch.randn_like(settled["phi_S"])
    phi_u = settled["phi_u"] + 0.3 * torch.randn_like(settled["phi_u"])
    r_lexical, r_state, r_utility = net.residuals(phi_S, phi_u, phi_L)
    step = 1e-6

    analytic_s = -(r_state / net.sigma_state) - (r_lexical / net.sigma_lexical)
    worst_gradient = 0.0
    for k in range(0, net.num_nodes, 17):
        up, down = phi_S.clone(), phi_S.clone()
        up[k] += step
        down[k] -= step
        numeric = (
            float(net.free_energy(up, phi_u, phi_L))
            - float(net.free_energy(down, phi_u, phi_L))
        ) / (2 * step * float(net.weights[k]))
        worst_gradient = max(
            worst_gradient,
            abs(numeric - float(analytic_s[k])) / (abs(float(analytic_s[k])) + 1e-9),
        )

    analytic_u = -(r_utility / net.sigma_utility) + net.theta_u * net.project(
        r_state / net.sigma_state
    )
    for k in range(net.num_basis):
        up, down = phi_u.clone(), phi_u.clone()
        up[k] += step
        down[k] -= step
        numeric = (
            float(net.free_energy(phi_S, up, phi_L))
            - float(net.free_energy(phi_S, down, phi_L))
        ) / (2 * step)
        worst_gradient = max(
            worst_gradient,
            abs(numeric - float(analytic_u[k])) / (abs(float(analytic_u[k])) + 1e-9),
        )

    analytic_theta = float(net.dot(phi_u, net.project(r_state / net.sigma_state)))
    numeric_theta = (
        float(net.free_energy(phi_S, phi_u, phi_L, theta_u=net.theta_u + step))
        - float(net.free_energy(phi_S, phi_u, phi_L, theta_u=net.theta_u - step))
    ) / (2 * step)
    worst_gradient = max(
        worst_gradient,
        abs(numeric_theta - analytic_theta) / (abs(analytic_theta) + 1e-9),
    )
    results["Sec. 7 messages are gradients of Eq. (13)"] = (
        worst_gradient < 1e-4, f"max relative error = {worst_gradient:.2e}"
    )

    # Sec. 1: the grid discretizes the integral, not the state, so the reported
    # belief must be stable under refinement. What that claims is CONVERGENCE in K,
    # which is a statement about the tail of the sequence, not about how close two
    # coarse grids happen to sit. Testing it as a 51-vs-401 gap against a fixed
    # tolerance tests the convergence RATE instead, and the rate here is set by
    # something the claim does not mention: chi_y is a step, and trapezoidal
    # quadrature on a discontinuous integrand converges at O(h) with the step's
    # position jittering against the nodes as K changes. The smooth mask of Sec. 9.2
    # is the control, being the same model with a continuous integrand, and is
    # reported beside it.
    def expectations(num_nodes, sharpness):
        probe = net.respawn(num_nodes=num_nodes, exclusion_sharpness=sharpness)
        return [
            float((probe.weights
                   * probe.read_out(probe.closed_form_fixed_point(y)[0])
                   * logistic(probe.zeta)).sum())
            for y in UTTERANCES
        ]

    ladder = [51, 101, 201, 401, 801]
    hard = {k: expectations(k, None) for k in ladder}
    smooth = {k: expectations(k, 0.1) for k in ladder}
    spread = max(abs(hard[401][i] - hard[51][i]) for i in range(len(UTTERANCES)))
    tail = max(abs(hard[801][i] - hard[401][i]) for i in range(len(UTTERANCES)))
    smooth_tail = max(abs(smooth[801][i] - smooth[401][i]) for i in range(len(UTTERANCES)))
    results["Sec. 1  K is an accuracy parameter"] = (
        tail < 5e-3,
        f"E[s] converging: |401-801| = {tail:.1e} (hard chi_y), {smooth_tail:.1e} "
        f"(smooth chi_y); coarse 51-401 spread {spread:.1e} is the O(h) step artefact",
    )

    # Sec. 8.5: F~(theta_u) = max_x F(x, theta_u) must have an interior stationary
    # point, which is exactly what mu_u = 0 destroys: there F~ climbs monotonically to
    # an asymptote and theta_u diverges. Located in closed form (Sec. 8.6), so this is
    # cheap; the dynamics are checked separately below.
    #
    # THE SEARCH MUST COVER BOTH SIGNS. theta_u is an unconstrained real parameter:
    # Bogacz's Eqs. (25) and (56) give a gradient and no domain, and his own solution
    # theta* = u / v_p takes whichever sign the data give. Eliminating phi_S and phi_u
    # analytically (Sec. 8.6) reduces the objective to
    #
    #     F~(theta) = const - ||theta mu_u - c||^2 / (2 (sigma_u theta^2 + S)),
    #         S = sigma_L + sigma_S,        c = B^T W (ell_0 - phi_L),
    #
    # whose stationarity condition is the quadratic
    #
    #     sigma_u <mu_u,c> theta^2 + (S ||mu_u||^2 - sigma_u ||c||^2) theta
    #                                                        - S <mu_u,c> = 0.
    #
    # Its roots have product -S / sigma_u < 0, so whenever <mu_u,c> != 0 there is
    # exactly ONE root of each sign; the maximizer is the one with sign(theta*) =
    # sign(<mu_u,c>) and the other root is a minimum. So the half-line to search is
    # read off dF~/dtheta at zero, which equals <mu_u,c> / S, a positive multiple of
    # the same inner product, rather than assumed. Bracketing on theta_u > 0 alone
    # is a restriction Bogacz does not impose, and under g_S = theta_u B phi_u (with
    # ell_0 in g_L, Eq. 9) the maximizer is negative, so such a bracket reports a
    # divergence that is an artefact of the bracket and not a property of the model.
    slow = net.respawn()
    lexical = {u: slow.lexical_field(u) for u in UTTERANCES}

    def reduced_energy(theta):
        return sum(
            float(slow.free_energy(
                *slow.closed_form_fixed_point(u, theta_u=theta), lexical[u], theta_u=theta))
            for u in UTTERANCES
        ) / len(UTTERANCES)

    def reduced_gradient(theta):
        total = 0.0
        for u in UTTERANCES:
            phi_S, phi_u = slow.closed_form_fixed_point(u, theta_u=theta)
            _, r_state, _ = slow.residuals(phi_S, phi_u, lexical[u], theta_u=theta)
            total += float(slow.dot(phi_u, slow.project(r_state / slow.sigma_state)))
        return total / len(UTTERANCES)

    ascent_at_zero = reduced_gradient(0.0)
    if ascent_at_zero == 0.0:
        raise RuntimeError(
            "dF~/dtheta_u = 0 at theta_u = 0, i.e. <mu_u, c> = 0: the quadratic above "
            "degenerates and no interior maximizer exists on either half-line"
        )
    sign = 1.0 if ascent_at_zero > 0 else -1.0

    low, high = 1e-3, 1e6                      # MAGNITUDES; the sign is fixed above
    for _ in range(120):                       # geometric bisection on the gradient
        mid = math.sqrt(low * high)
        if sign * reduced_gradient(sign * mid) > 0:
            low = mid
        else:
            high = mid
    theta_star = sign * math.sqrt(low * high)
    margin = reduced_energy(theta_star) - reduced_energy(sign * 1e8)
    interior = (
        reduced_energy(theta_star) > reduced_energy(0.5 * theta_star)
        and reduced_energy(theta_star) > reduced_energy(2.0 * theta_star)
        and margin > 0
    )
    results["Sec. 8.5  F~ has an interior maximum in theta_u"] = (
        interior and abs(reduced_gradient(theta_star)) < 1e-6,
        f"theta_u* = {theta_star:.4f}, |dF~/dtheta_u| = {abs(reduced_gradient(theta_star)):.1e}, "
        f"margin over the asymptote = {margin:+.4f}",
    )

    # and the flow of Eq. (20) must move toward it, with F monotone. It is started on
    # the half-line located above: gradient ascent cannot cross the minimum sitting at
    # the opposite-signed root, so a start of the wrong sign runs to the asymptote
    # correctly, and would be testing the initialization rather than the model.
    dynamic = net.respawn(theta_u=math.copysign(net.theta_u, sign))
    trace = dynamic.learn_theta_u(UTTERANCES, tau_theta=20.0, num_updates=25)
    energies = [entry["free_energy"] for entry in trace]
    worst_step = min(energies[i + 1] - energies[i] for i in range(len(energies) - 1))
    approaching = (
        abs(dynamic.theta_u - theta_star) < abs(trace[0]["theta_u"] - theta_star)
        and abs(trace[-1]["gradient"]) < abs(trace[0]["gradient"])
    )
    results["Sec. 8.5  Eq. (20) ascends toward it"] = (
        approaching and worst_step > -1e-9,
        f"theta_u {trace[0]['theta_u']:.3f} -> {dynamic.theta_u:.3f} (target {theta_star:.3f}), "
        f"|grad| {abs(trace[0]['gradient']):.3g} -> {abs(trace[-1]['gradient']):.3g}, "
        f"min dF step {worst_step:+.1e}",
    )

    # Sec. 9.2: a dense-scale (smooth) lexical field changes nothing structural.
    worst_dense = 0.0
    for sharpness in [0.5, 1.0, 2.0]:
        probe = net.respawn(exclusion_sharpness=sharpness)
        for y in UTTERANCES:
            result = probe.infer(y, dt=dt)
            phi_S_star, _ = probe.closed_form_fixed_point(y)
            worst_dense = max(
                worst_dense, float((result["phi_S"] - phi_S_star).abs().max())
            )
    results["Sec. 9.2 smooth lexical field is compatible"] = (
        worst_dense < 1e-8, f"max |phi_S - phi_S*| = {worst_dense:.2e}"
    )

    if verbose:
        width = max(len(k) for k in results)
        print("SPECIFICATION CHECKS")
        print("=" * (width + 46))
        for claim, (ok, detail) in results.items():
            print(f"  {'PASS' if ok else 'FAIL'}  {claim:<{width}}  {detail}")
        print("=" * (width + 46))
        failures = sum(1 for ok, _ in results.values() if not ok)
        print(f"  {len(results) - failures}/{len(results)} passed")
        print()

    return results


# ============================================================================
# B. DEMONSTRATION
# ============================================================================

def demonstrate(net, *, dt=0.02, verbose=True):
    """Run the network on each utterance and report the settled belief."""
    proportion = logistic(net.zeta)
    upper_region = (net.zeta > net.theta_L).to(net.dtype)
    lower_region = (net.zeta <= -net.theta_L).to(net.dtype)

    table = {}
    for utterance in UTTERANCES:
        result = net.infer(utterance, dt=dt)
        q = result["q"]
        excluded = exclusion_indicator(
            utterance, net.zeta, net.theta_L, sharpness=net.exclusion_sharpness
        )
        table[utterance] = {
            "q": q,
            "expected_s": float((net.weights * q * proportion).sum()),
            "leak": float((net.weights * q * excluded).sum()),
            "upper_q_mass": float((net.weights * q * upper_region).sum()),
            "lower_q_mass": float((net.weights * q * lower_region).sum()),
            "phi_u": result["phi_u"].tolist(),
            "steps": result["steps"],
            "free_energy": float(result["free_energy"]),
        }

    if verbose:
        print("DEMONSTRATION")
        print(f"  Lambda = {net.lexical_strength}, theta_u = {net.theta_u}, "
              f"mu_u = {net.mu_u.tolist()}, boundary = {net.theta_L}")
        print()
        print(f"  {'utterance':<10}{'E[s]':>9}{'P(s~0)':>10}{'P(s~1)':>10}"
              f"{'leak':>10}{'steps':>8}   phi_u (tilt, width)")
        for utterance, row in table.items():
            print(f"  {utterance:<10}{row['expected_s']:>9.4f}{row['lower_q_mass']:>10.4f}"
                  f"{row['upper_q_mass']:>10.4f}{row['leak']:>10.4f}{row['steps']:>8}"
                  f"   [{row['phi_u'][0]:+.3f}, {row['phi_u'][1]:+.3f}]")
        print()
        print("  'leak' is the q-mass over the region the entry excludes;")
        print("  it goes to zero as Lambda grows, which is the Eq. (15) limit of Sec. 6.")
        print()

    return table


def lexical_strength_sweep(net, strengths=(1, 2, 4, 8, 16, 32, 64), dt=0.02):
    """Sec. 6: a hard support constraint is the Lambda -> infinity
    limit of Eq. (15), not a separate assumption. Leak should fall monotonically to 0."""
    rows = []
    for strength in strengths:
        probe = net.respawn(lexical_strength=float(strength))
        leaks = []
        for utterance in UTTERANCES:
            q = probe.infer(utterance, dt=dt)["q"]
            excluded = exclusion_indicator(
                utterance, probe.zeta, probe.theta_L,
                sharpness=probe.exclusion_sharpness,
            )
            leaks.append(float((probe.weights * q * excluded).sum()))
        rows.append((float(strength), max(leaks)))

    print("LAMBDA -> INFINITY RECOVERS THE HARD MASK (Sec. 6)")
    print(f"  {'Lambda':>8}{'max leak':>12}")
    for strength, leak in rows:
        print(f"  {strength:>8.0f}{leak:>12.6f}")
    monotone = all(rows[i + 1][1] <= rows[i][1] + 1e-12 for i in range(len(rows) - 1))
    print(f"  monotone decreasing: {monotone};  leak at Lambda={rows[-1][0]:.0f} "
          f"is {rows[-1][1]:.2e}")
    print()
    return rows


# ============================================================================
# C. THE SEMANTIC QUESTION: IS THERE A SCALAR IMPLICATURE?
# ============================================================================

def scalar_implicature_probe(net, *, dt=0.02, verbose=True):
    """Compare the settled belief against the literal listener.

    The baseline is q_lit, the untempered literal posterior of
    `literal_fixed_point`: phi_S = ell_0 - phi_L, the world prior restricted by the
    lexical entry and nothing else. The shift is measured against it.

    The theta_u = 0 belief is reported beside it as a secondary control. It is the
    same literal posterior tempered by one half (see `literal_fixed_point`), so it
    holds the temperature of the full network fixed while severing the utility level;
    q_lit holds the definition fixed instead. The two differ, and Text cell 4 Part D
    states by how much.

    A scalar implicature for "some" would move q-mass AWAY from the region where
    "all" is true. The measured quantity is therefore the change in P(s in all-region)
    between the control and the full network, and an implicature requires it to be
    NEGATIVE.
    """
    upper_region = (net.zeta > net.theta_L).to(net.dtype)
    proportion = logistic(net.zeta)

    rows = {}
    for utterance in UTTERANCES:
        baseline = net.read_out(net.literal_fixed_point(utterance))
        tempered = net.infer(utterance, dt=dt, theta_u=0.0)["q"]
        full = net.infer(utterance, dt=dt)["q"]
        rows[utterance] = {
            "tempered_upper": float((net.weights * tempered * upper_region).sum()),
            "baseline_upper": float((net.weights * baseline * upper_region).sum()),
            "full_upper": float((net.weights * full * upper_region).sum()),
            "baseline_expected_s": float((net.weights * baseline * proportion).sum()),
            "full_expected_s": float((net.weights * full * proportion).sum()),
            "baseline_q": baseline,
            "full_q": full,
        }
        rows[utterance]["shift"] = (
            rows[utterance]["full_upper"] - rows[utterance]["baseline_upper"]
        )

    if verbose:
        print("SCALAR IMPLICATURE PROBE")
        print("  baseline: q_lit, the untempered literal posterior (phi_S = ell_0 - phi_L)")
        print("  also shown: the theta_u = 0 control, that posterior tempered by 1/2")
        print()
        print(f"  {'utterance':<10}{'E[s] base':>11}{'E[s] full':>11}"
              f"{'P(all) base':>13}{'P(all) full':>13}{'shift':>10}"
              f"{'P(all) th_u=0':>15}")
        for utterance, row in rows.items():
            print(f"  {utterance:<10}{row['baseline_expected_s']:>11.4f}"
                  f"{row['full_expected_s']:>11.4f}{row['baseline_upper']:>13.4f}"
                  f"{row['full_upper']:>13.4f}{row['shift']:>+10.4f}"
                  f"{row['tempered_upper']:>15.4f}")
        print()
        shift = rows["some"]["shift"]
        verdict = "met" if shift < 0 else "NOT met"
        print(f"  Under the criterion of Text cell 4 Part C, the criterion is {verdict}:")
        print(f"  for \"some\" the utility level moves q-mass "
              f"{'away from' if shift < 0 else 'TOWARD'} the "
              f"all-region by {abs(shift):.4f}.")
        print("  Whether the criterion is met is a measurement; what meeting it commits")
        print("  the model to is argued in Text cell 4 Part C, not here. Note that the")
        print("  same criterion is met by other utterances too (see the shift column")
        print("  above), and that its sign is prior-dependent (Part D).")
        print()

    return rows


def theta_u_learning_probe(net, *, tau_theta=20.0, num_updates=60):
    """Sec. 8.5: integrate Eq. (20), tau_theta * d(theta_u)/dt = dF/d(theta_u).

    There is no learning rate: tau_theta is a time constant of the same kind as
    tau_error and tau_state, and the model's one timescale ordering runs
    tau_error << tau_state << tau_theta. It sets the rate, not the fixed point.

    Requires <mu_u, sum_y c_y> != 0 (Appendix B); mu_u != 0 (Sec. 3) is necessary but
    not sufficient. Otherwise F~ has no stationary point and theta_u diverges, which
    is Bogacz's Eq. (25) at v_p = 0, not a departure from it.
    """
    net = net.respawn()
    # Start on the half-line that contains the maximizer. dF~/dtheta_u at theta_u = 0
    # is <mu_u, B^T W (ell_0 - phi_L)> / (sigma_L + sigma_S), whose sign is the sign of
    # theta_u* (Part A); ascent cannot cross the opposite-signed root, which is a
    # minimum. The magnitude of the start is left as configured.
    lexical = {u: net.lexical_field(u) for u in UTTERANCES}
    ascent_at_zero = 0.0
    for u in UTTERANCES:
        phi_S, phi_u = net.closed_form_fixed_point(u, theta_u=0.0)
        _, r_state, _ = net.residuals(phi_S, phi_u, lexical[u], theta_u=0.0)
        ascent_at_zero += float(net.dot(phi_u, net.project(r_state / net.sigma_state)))
    net.theta_u = math.copysign(net.theta_u, ascent_at_zero)

    initial = net.theta_u
    trace = net.learn_theta_u(UTTERANCES, tau_theta=tau_theta, num_updates=num_updates)
    energies = [entry["free_energy"] for entry in trace]
    monotone = all(
        energies[i + 1] >= energies[i] - 1e-9 for i in range(len(energies) - 1)
    )
    print("SLOW PARAMETER (Eq. 20, Sec. 8.5)")
    print(f"  mu_u = {net.mu_u.tolist()}  (must be nonzero)")
    print(f"  theta_u: {initial:.4f} -> {net.theta_u:.6f} over {num_updates} steps "
          f"at tau_theta = {tau_theta:g}")
    print(f"  F increased monotonically: {monotone}")
    print(f"  final dF/dtheta_u = {trace[-1]['gradient']:+.3e}"
          f"  ({'settled' if abs(trace[-1]['gradient']) < 1e-3 else 'still ascending'})")
    print("  The flow is slow near the optimum because F~ is shallow there; the target")
    print("  is the stationary point Part A locates in closed form, not this endpoint.")
    probe = scalar_implicature_probe(net, verbose=False)
    print(f"  implicature shift for \"some\" at the settled theta_u: "
          f"{probe['some']['shift']:+.4f}")
    print("  Ascending F moves theta_u TOWARD the criterion of Text cell 4 Part C, not")
    print("  away from it, and deepens the shift as it goes. Eq. (13) contains no term")
    print("  rewarding informativity, so this is not F pursuing the criterion: raising")
    print("  |theta_u| amplifies the span(B) component of (ell_0 - phi_L), and the sign")
    print("  of the shift is set by where those directions put mass relative to the")
    print("  all-region. The effect saturates in |theta_u| and is the same for either")
    print("  sign; it enters through theta_u^2. See `theta_u_dependence` below.")
    print()
    return net, trace



def theta_u_stationary_point(net):
    """theta_u* for this configuration, in closed form (Eq. B2 via Sec. 8.6)."""
    lexical = {u: net.lexical_field(u) for u in UTTERANCES}

    def gradient(theta):
        total = 0.0
        for u in UTTERANCES:
            phi_S, phi_u = net.closed_form_fixed_point(u, theta_u=theta)
            _, r_state, _ = net.residuals(phi_S, phi_u, lexical[u], theta_u=theta)
            total += float(net.dot(phi_u, net.project(r_state / net.sigma_state)))
        return total / len(UTTERANCES)

    at_zero = gradient(0.0)
    if at_zero == 0.0:
        raise RuntimeError("<mu_u, sum_y c_y> = 0: no interior maximizer (Appendix B)")
    sign = 1.0 if at_zero > 0 else -1.0
    low, high = 1e-3, 1e6
    for _ in range(120):
        middle = math.sqrt(low * high)
        if sign * gradient(sign * middle) > 0:
            low = middle
        else:
            high = middle
    return sign * math.sqrt(low * high)


def theta_u_dependence(net, *, verbose=True):
    """Every headline number in this cell, recomputed across theta_u.

    The rest of the evaluation is reported at theta_u = 1, which is neither the
    learned optimum nor even its sign: theta_u* is NEGATIVE here (Appendix B). This
    probe exists so that no conclusion rests on the pinned value without that being
    visible. Four columns: the control (theta_u = 0), the reported setting, its
    negation, and the stationary point the flow of Eq. (20) actually ascends to.

    The shift row is Eq. (37), against q_lit, exactly as everywhere else in the
    notebook; the control column therefore shows what the theta_u = 0 tempering does
    on its own rather than a column of zeros.

    Read it for two things. Which quantities depend on theta_u at all, and which of
    those depend on its SIGN: the utility level reaches phi_S through theta_u * B and
    the fixed point through theta_u^2 * G, so quantities dominated by the quadratic
    term are near-identical at +theta and -theta while the mu_u-carried linear term
    is what breaks the symmetry.
    """
    star = theta_u_stationary_point(net)
    settings = [("control (0)", 0.0), ("reported (1)", 1.0),
                ("negated (-1)", -1.0), ("stationary", star)]
    proportion = logistic(net.zeta)
    upper = (net.zeta > net.theta_L).to(net.dtype)

    table = {}
    for label, theta in settings:
        column = {}
        for y in UTTERANCES:
            field = net.closed_form_fixed_point(y, theta_u=theta)[0]
            belief = net.read_out(field)
            base = net.read_out(net.literal_fixed_point(y))
            excluded = exclusion_indicator(
                y, net.zeta, net.theta_L, sharpness=net.exclusion_sharpness
            )
            column[y] = {
                "expected_s": float((net.weights * belief * proportion).sum()),
                "upper": float((net.weights * belief * upper).sum()),
                "leak": float((net.weights * belief * excluded).sum()),
                "shift": float((net.weights * belief * upper).sum())
                         - float((net.weights * base * upper).sum()),
            }
        contrast = (net.closed_form_fixed_point("all", theta_u=theta)[0]
                    - net.closed_form_fixed_point("some", theta_u=theta)[0])
        column["contrast"] = float(net.squared_norm(contrast)) ** 0.5
        table[label] = column

    if verbose:
        print("THETA_U DEPENDENCE OF THE REPORTED QUANTITIES")
        print(f"  theta_u* = {star:.4f} (closed form, Appendix B); the rest of this cell")
        print(f"  reports theta_u = {net.theta_u:g}.")
        print()
        for name, key, fmt in [("E[s]", "expected_s", "{:>15.4f}"),
                               ("P(all-region)", "upper", "{:>15.4f}"),
                               ("shift vs q_lit (37)", "shift", "{:>+15.4f}"),
                               ("leak", "leak", "{:>15.2e}")]:
            print(f"  {name}")
            print(f"    {'utterance':<12}" + "".join(f"{lab:>15}" for lab, _ in settings))
            for y in UTTERANCES:
                print(f"    {y:<12}"
                      + "".join(fmt.format(table[lab][y][key]) for lab, _ in settings))
            print()
        print(f"    {'||contrast||':<12}"
              + "".join(f"{table[lab]['contrast']:>15.4f}" for lab, _ in settings))
        print("    (all - some, in the quadrature norm; it GROWS with |theta_u|, so the")
        print("     utility level does move contrasts, unlike mu_u and ell_0.)")
        print()
        symmetric = max(
            abs(table["reported (1)"][y][key] - table["negated (-1)"][y][key])
            for y in UTTERANCES for key in ("expected_s", "upper")
        )
        print(f"  theta_u = +1 against theta_u = -1: largest difference {symmetric:.4f}.")
        print("  That is not small, and it is worth knowing WHERE it is large. phi_u*")
        print("  carries a term linear in theta_u (through mu_u) and one quadratic in it")
        print("  (through G); only the linear one changes under theta_u -> -theta_u, and")
        print("  it is relatively largest when |theta_u| is of order 1:")
        print()
        print(f"    {'|theta_u|':>12}{'max dE[s]':>12}{'max dP(all)':>14}")
        for magnitude in (0.5, 1.0, 2.0, 5.0, 10.0, abs(star), 100.0):
            gaps = [[], []]
            for y in UTTERANCES:
                plus = net.read_out(net.closed_form_fixed_point(y, theta_u=magnitude)[0])
                minus = net.read_out(net.closed_form_fixed_point(y, theta_u=-magnitude)[0])
                gaps[0].append(abs(float((net.weights * plus * proportion).sum())
                                   - float((net.weights * minus * proportion).sum())))
                gaps[1].append(abs(float((net.weights * plus * upper).sum())
                                   - float((net.weights * minus * upper).sum())))
            print(f"    {magnitude:>12.4f}{max(gaps[0]):>12.4f}{max(gaps[1]):>14.4f}")
        print()
        print("  The sign therefore matters MOST at the value the rest of this cell is")
        print("  pinned to, and progressively less toward theta_u*, where the quadratic")
        print("  term dominates and the two signs converge. Numbers reported at")
        print("  theta_u = 1 are being read at close to the worst case for this.")
        print()

    return table, star



# ============================================================================
# D. BASE WORLD PRIOR
# ============================================================================

# The delta-like row. Part D's other four priors are all diffuse over the scale, so
# none of them puts the criterion under any pressure: hearing "some" under a prior
# that is already spread out, there is little all-region mass to move away from. The
# sharp case is the one worth testing, and it is the case Code Cell 3 is built around.
#
# A literal delta at s = 1 is not representable, since s lies in the OPEN interval, so
# zeta = logit(1) is off the grid and ell_0 = log p_0 would be -infinity almost
# everywhere. Beta(alpha, 1) is the limit family that approaches it (Text cell 6).
#
#   alpha = 64 puts the mode in zeta at log(alpha) = 4.16, well inside the all-region
#   (zeta > theta_L = 2.9444) and well clear of the grid ceiling, which Text cell 6
#   warns sets in above alpha ~ 300; it carries 95.7% of its prior mass in the
#   all-region and only 0.95% on the top node.
#
#   Lambda must then be raised to match, or the prior simply overrides the lexicon and
#   the sweep measures the override rather than the criterion. Eq. (41) puts the
#   override threshold at Lambda_crit ~ alpha * log(2n) = 64 * 2.9957 = 192. Taking
#   Lambda = 8 alpha = 512, about 2.7x that, drives every leak below 1e-70: the entry
#   holds against the prior, which is the precondition for the probe to mean anything.
DELTA_ALL_ALPHA = 64.0
DELTA_ALL_LAMBDA = 8.0 * DELTA_ALL_ALPHA          # = 512, about 2.7 x Lambda_crit


def base_prior_sweep(net, priors=None, *, dt=0.02, verbose=True):
    """Repeat the demonstration and the implicature probe under each base world prior.

    ell_0 is a fixed quantity (Sec. 3), so changing it moves the maximizer of Eq. (13)
    but touches nothing in Sec. 8: by Eq. (22) the Hessian does not involve ell_0 any
    more than it involves phi_L, so every convergence result holds unchanged and the
    closed forms stay exact. What the sweep tests is semantic, not numerical: whether
    the conclusion of Part C is an artefact of one prior.

    The flat condition is a literal Uniform(0, 1) over the proportion, not a
    low-precision Gaussian; see `uniform_world_prior` for why the two differ.

    A prior may be given either as a log-density or as a (log_density, overrides) pair,
    the overrides passed on to `respawn`. That is how the delta-like row carries its own
    Lambda: its whole point is a prior sharp enough to need one.
    """
    priors = priors or {
        **BASE_WORLD_PRIORS,
        "delta (all)": (
            beta_world_prior(DELTA_ALL_ALPHA, 1.0),
            {"lexical_strength": DELTA_ALL_LAMBDA},
        ),
    }
    proportion = logistic(net.zeta)
    upper_region = (net.zeta > net.theta_L).to(net.dtype)

    rows = {}
    for name, specification in priors.items():
        prior, overrides = (
            specification if isinstance(specification, tuple) else (specification, {})
        )
        probe = net.respawn(base_prior=prior, **overrides)
        prior_density = torch.exp(probe.base_log_prior)
        entry = {
            "description": probe.base_prior_description,
            "lexical_strength": probe.lexical_strength,
            "prior_expected_s": float((probe.weights * prior_density * proportion).sum()),
            "prior_upper_mass": float((probe.weights * prior_density * upper_region).sum()),
            "utterances": {},
            "network": probe,
        }
        for y in UTTERANCES:
            settled = probe.infer(y, dt=dt)
            baseline = probe.read_out(probe.literal_fixed_point(y))
            tempered = probe.infer(y, dt=dt, theta_u=0.0)["q"]
            full = settled["q"]
            excluded = exclusion_indicator(
                y, probe.zeta, probe.theta_L, sharpness=probe.exclusion_sharpness
            )
            entry["utterances"][y] = {
                "expected_s": float((probe.weights * full * proportion).sum()),
                "baseline_upper": float((probe.weights * baseline * upper_region).sum()),
                "tempered_upper": float((probe.weights * tempered * upper_region).sum()),
                "full_upper": float((probe.weights * full * upper_region).sum()),
                "leak": float((probe.weights * full * excluded).sum()),
                "converged": settled["converged"],
                "closed_form_error": float(
                    (settled["phi_S"] - probe.closed_form_fixed_point(y)[0]).abs().max()
                ),
                "baseline_q": baseline,
                "full_q": full,
            }
            entry["utterances"][y]["shift"] = (
                entry["utterances"][y]["full_upper"]
                - entry["utterances"][y]["baseline_upper"]
            )
        rows[name] = entry

    if verbose:
        print("BASE WORLD PRIOR SWEEP")
        print()
        print(f"  {'prior':<13}{'description':<43}{'Lambda':>8}{'E[s] prior':>11}"
              f"{'P0(all)':>9}{'  E[s] | no    some    all':<28}")
        for name, entry in rows.items():
            means = "  ".join(
                f"{entry['utterances'][y]['expected_s']:.3f}" for y in UTTERANCES
            )
            print(f"  {name:<13}{entry['description']:<43}{entry['lexical_strength']:>8.0f}"
                  f"{entry['prior_expected_s']:>11.4f}{entry['prior_upper_mass']:>9.4f}"
                  f"         {means}")
        print()
        print("  implicature shift, P(all-region) full minus q_lit, the literal listener")
        print("  (a scalar implicature for \"some\" requires a NEGATIVE shift)")
        print()
        print(f"  {'prior':<13}" + "".join(f"{y:>12}" for y in UTTERANCES)
              + f"{'max leak':>12}{'max |phi_S - phi_S*|':>24}")
        for name, entry in rows.items():
            shifts = "".join(
                f"{entry['utterances'][y]['shift']:>+12.4f}" for y in UTTERANCES
            )
            worst = max(entry["utterances"][y]["closed_form_error"] for y in UTTERANCES)
            leak = max(entry["utterances"][y]["leak"] for y in UTTERANCES)
            print(f"  {name:<13}{shifts}{leak:>12.1e}{worst:>24.1e}")
        print()
        signs = {name: entry["utterances"]["some"]["shift"] for name, entry in rows.items()}
        met = [name for name, value in signs.items() if value < 0]
        print(f"  Criterion met under {len(met)} of {len(signs)} priors: "
              + ", ".join(met) + ".")
        print("  The sign is therefore NOT a property of the model alone; it depends on")
        print("  the prior the model is run against, which is a fact about the criterion")
        print("  as much as about the model. Text cell 4 Part C reads this.")
        print()

        # === the delta-like row, and the headroom that makes it readable ===
        if "delta (all)" in rows:
            entry = rows["delta (all)"]
            print("  THE DELTA-LIKE ROW. This is the case the criterion is under the most")
            print("  pressure in: the prior already puts %.1f%% of its mass in the all-region,"
                  % (100 * entry["prior_upper_mass"]))
            print("  so the theta_u control, given \"some\", an entry that does NOT")
            print("  exclude the all-region, stays there. Leak stays below")
            print("  %.0e, so the lexicon is holding and the probe is measuring the utility"
                  % max(entry["utterances"][y]["leak"] for y in UTTERANCES))
            print("  level rather than the prior overriding the entry.")
            print()
            print(f"    {'utterance':<10}{'P(all) base':>13}{'P(all) full':>13}{'shift':>10}"
                  f"{'room to fall':>14}{'room to rise':>14}")
            for y in UTTERANCES:
                row = entry["utterances"][y]
                print(f"    {y:<10}{row['baseline_upper']:>13.6f}{row['full_upper']:>13.6f}"
                      f"{row['shift']:>+10.4f}{row['baseline_upper']:>14.2e}"
                      f"{1.0 - row['baseline_upper']:>14.2e}")
            print()
            print("  READ THE HEADROOM BEFORE THE SHIFTS. \"no\" has essentially no")
            print("  all-region mass to lose and \"all\" essentially none to gain, so their")
            print("  zeros are saturation, not restraint; only \"some\" is free to move in")
            print("  both directions, and it is the one that moves. What is NOT saturation")
            print("  is that \"all\" keeps its all-region mass while \"some\" sheds a third of")
            print("  its own: both had room to fall, and only one fell. That difference is")
            print("  the even-parity (width) coordinate of their lexical projections, which")
            print("  is the one place the two entries differ.")
            print()

    return rows


# ============================================================================
# RUN
# ============================================================================

evaluation_network = LexicalPredictiveCodingNetwork()

check_specification(evaluation_network)
demonstration = demonstrate(evaluation_network)
lexical_strength_sweep(evaluation_network)
implicature = scalar_implicature_probe(evaluation_network)
learned_network, theta_trace = theta_u_learning_probe(evaluation_network)
theta_table, theta_star_value = theta_u_dependence(evaluation_network)
prior_rows = base_prior_sweep(evaluation_network)


# ---- figure ----

try:
    import matplotlib.pyplot as plt

    proportion = logistic(evaluation_network.zeta).numpy()

    # Curves are plotted as densities in the PROPORTION s, so the belief, which is
    # normalized as a density in zeta, must be pushed through the change of variables
    #     p_s(s) = p_zeta(zeta) * |dzeta/ds| = q(zeta) / (s (1 - s)),
    # dividing by ds/dzeta rather than multiplying by it. The transform is checkable:
    # sum_k w_k * to_s_density(q)_k * (ds/dzeta)_k = 1 exactly, whereas multiplying by
    # the Jacobian gives ~0.035.
    ds_dzeta = proportion * (1.0 - proportion)

    def to_s_density(field):
        return field.numpy() / ds_dzeta

    all_region_start = float(logistic(torch.tensor(evaluation_network.theta_L)))

    figure, axes = plt.subplots(1, 3, figsize=(13, 3.6), sharey=True)
    for axis, utterance in zip(axes, UTTERANCES):
        row = implicature[utterance]
        axis.plot(proportion, to_s_density(torch.exp(evaluation_network.base_log_prior)),
                  color="0.6", linewidth=1.0, label="base world prior")
        axis.plot(proportion, to_s_density(row["baseline_q"]),
                  linestyle="--", label="literal listener $q_{\\mathrm{lit}}$")
        axis.plot(proportion, to_s_density(row["full_q"]), label="full network")
        axis.axvspan(all_region_start, 1.0, alpha=0.12, color="tab:red")
        axis.set_title(f'"{utterance}"')
        axis.set_xlabel("proportion $s$")
    axes[0].set_ylabel("read-out density $q(s)$")
    axes[0].legend(fontsize=8, loc="upper center")
    figure.suptitle(
        "Settled belief against the literal listener "
        "(shaded: the region where \"all\" is true)", fontsize=10)
    figure.tight_layout()
    plt.show()

    # second figure: "some" under each base world prior (Part D). Each panel carries its
    # own prior, so the gray line differs from panel to panel.
    #
    # Only rows sharing this network's Lambda are drawn. Panels on a shared y-axis have
    # to be comparable, and the delta-like row deliberately carries its own Lambda; it
    # is reported in the table and the headroom block above instead.
    comparable = {
        name: entry for name, entry in prior_rows.items()
        if entry["lexical_strength"] == evaluation_network.lexical_strength
    }
    prior_figure, prior_axes = plt.subplots(
        1, len(comparable), figsize=(3.4 * len(comparable), 3.4), sharey=True
    )
    for axis, (name, entry) in zip(prior_axes, comparable.items()):
        row = entry["utterances"]["some"]
        axis.plot(proportion, to_s_density(torch.exp(entry["network"].base_log_prior)),
                  color="0.6", linewidth=1.0, label="base world prior")
        axis.plot(proportion, to_s_density(row["baseline_q"]),
                  linestyle="--", label="literal listener $q_{\\mathrm{lit}}$")
        axis.plot(proportion, to_s_density(row["full_q"]), label="full network")
        axis.axvspan(all_region_start, 1.0, alpha=0.12, color="tab:red")
        axis.set_title(f"{name}\n(shift {row['shift']:+.4f})", fontsize=9)
        axis.set_xlabel("proportion $s$")
    prior_axes[0].set_ylabel("read-out density $q(s)$")
    prior_axes[0].legend(fontsize=8)
    prior_figure.suptitle(
        '"some" under the four base world priors sharing Lambda = %g: the sign of the '
        'shift against the $\\theta_u$ control is not constant across them'
        % evaluation_network.lexical_strength, fontsize=10)
    prior_figure.tight_layout()
    plt.show()

    # third figure: the delta-like row of Part D, which the second figure cannot show.
    # It carries its own Lambda (DELTA_ALL_LAMBDA, not the network's), so it cannot
    # share the panels above. This is the sharp prior the criterion is under most
    # pressure from, and the case Code Cell 3 is built around.
    #
    # The vertical axis is LOGARITHMIC, which a near-delta prior forces: the settled
    # densities span four orders of magnitude, peaking at 6.1e3 under "all" against
    # 2.4e1 under "some", so on a linear axis every panel but "all" is a flat line
    # under one spike. On a shared log axis all three curves are legible in all three
    # panels, and the comparison to read is the full network against its own theta_u
    # control WITHIN each panel.
    delta_entry = prior_rows["delta (all)"]
    delta_prior = torch.exp(delta_entry["network"].base_log_prior)

    delta_figure, delta_axes = plt.subplots(1, 3, figsize=(13, 3.6), sharey=True)
    for axis, utterance in zip(delta_axes, UTTERANCES):
        row = delta_entry["utterances"][utterance]
        axis.plot(proportion, to_s_density(delta_prior),
                  color="0.6", linewidth=1.0, label="base world prior")
        axis.plot(proportion, to_s_density(row["baseline_q"]),
                  linestyle="--", label="literal listener $q_{\\mathrm{lit}}$")
        axis.plot(proportion, to_s_density(row["full_q"]), label="full network")
        axis.axvspan(all_region_start, 1.0, alpha=0.12, color="tab:red")
        axis.set_yscale("log")
        axis.set_ylim(1e-4, 3e4)
        axis.set_title(f'"{utterance}"  (shift {row["shift"]:+.4f})', fontsize=9)
        axis.set_xlabel("proportion $s$")
    delta_axes[0].set_ylabel("read-out density $q(s)$, log scale")
    delta_axes[0].legend(fontsize=8, loc="upper left")
    delta_figure.suptitle(
        "The delta-like base world prior of Part D, $\\mathrm{Beta}(%g,1)$ at "
        "$\\Lambda=%g$ (shaded: the region where \"all\" is true). Under "
        "\"some\" the control peaks inside that region and the full network is "
        "pulled out of it" % (DELTA_ALL_ALPHA, DELTA_ALL_LAMBDA), fontsize=10)
    delta_figure.tight_layout()
    plt.show()

except ImportError:
    print("matplotlib unavailable; skipping the figure")
````

## <a id="tc5" name="tc5"></a>Text cell 5: notes after the evaluation

Two matters that belong after the evaluation rather than inside the specification: why the model's
expressiveness depends on the utility basis dimension in a way its convergence proofs do not,
and a report on the settings of $\mu_u$ probed.

---

### <a id="tc5-parta" name="tc5-parta"></a>Part A. the utility basis dimension $m$

Text cell 3 refers to this part. Every result in Sec. 8 is stated for arbitrary fixed $B$ and arbitrary
$m$, and no proof there constrains the choice; what follows is about what the model can *express*.

*The model's expressive power is sensitive to the column size of $B$.* The $u$ level's only access
to the lexical field is through the $m$ **lexical projections**

$$\kappa_y=B^{\mathsf T}W\varphi_{L,y}\in\mathbb R^m,
\tag{29}$$

so two utterances whose projections coincide are indistinguishable to that level, however different
their exclusion sets. $m$ is therefore a substantive commitment.

**The letter is $\kappa$ and not $c$.** Appendix B writes
$c_y=B^{\mathsf T}W(\ell_0-\varphi_{L,y})$ for the vector that enters the closed forms of
Eqs. (15)–(16) and the stationarity condition Eq. (B2). Both are vectors in $\mathbb R^m$ and both
would naturally be called "the projections", so they are given different letters here. They differ by
an utterance-independent offset,

$$c_y=B^{\mathsf T}W\ell_0-\kappa_y,\qquad B^{\mathsf T}W\ell_0=[\,0,\;-18.5996\,]
\tag{30}$$

at the settings used throughout. The parity argument below turns on $\kappa_y$, because $\ell_0$
contributes the same amount to every entry and therefore cannot separate them; everything in
Appendix B turns on $c_y$.

#### <a id="tc5-m1-degenerate" name="tc5-m1-degenerate"></a>$m=1$ is degenerate

The three exclusion sets of Eq. (5) are built from a single pair of thresholds
$\pm\theta_L$. (The endpoints $s=0,1$ lie outside the open interval and carry no quadrature mass, so they
are represented as the neighbourhoods $|\zeta|\ge\theta_L$; $\theta_L$ is semantic, and
Appendix A gives its denotation.) Their pairwise differences are consequently very symmetric:

$$\chi_{\textit{all}}-\chi_{\textit{some}}=\mathbb 1_{(-\theta_L,\,\theta_L)},
\qquad
\chi_{\textit{no}}-\chi_{\textit{all}}=\mathbb 1_{(\theta_L,\,\infty)}-\mathbb 1_{(-\infty,\,-\theta_L)}.
\tag{31}$$

The first is an interval symmetric about the origin; the second is a mirror-image pair carrying
opposite signs. On a grid symmetric about $\zeta=0$, therefore, **any single basis function of definite
parity annihilates one of them**. An odd $b$ gives $\langle\chi_{\textit{all}}-\chi_{\textit{some}},
b\rangle=0$ and cannot separate *some* from *all*; an even $b$ gives
$\langle\chi_{\textit{no}}-\chi_{\textit{all}},b\rangle=0$ and cannot separate *no* from *all*. The
identities are exact, and the argument is one about parity rather than about any particular profile:
$\zeta$, $\zeta^3$ and $\sin\zeta$ all collapse $\{\textit{some},\textit{all}\}$, while $\zeta^2$,
$|\zeta|$ and $\cos\zeta$ all collapse $\{\textit{no},\textit{all}\}$.

The consequence reaches past the $u$ level. That level is the only site in the chain at which an
implicative reading could be computed, since $\varphi_L$ carries the lexical entry and no other quantity does.
With $m=1$ and a basis of definite parity, two of the three utterances receive *identical* utility
states, at every $\theta_u$, $\Lambda$ and $\mu_u$. Their beliefs still differ, since Eq. (15) carries
$\ell_0-\varphi_L$ into $\varphi_S^\ast$ without passing through the $u$ level; what is lost is the
utility level's purchase on the pair, so the contrast between them no longer depends on $\theta_u$.
Appendix C §6 measures that independence and records it as Eq. (C3).

A profile of no definite parity escapes the argument and does separate all three, but it confounds
into one coordinate the two things a basis can do to a log-density, which Part B §2 separates as
tilt and width. The implementation therefore takes $m=2$, the least the parity argument permits.
Its two columns are a linear and a quadratic profile in $\zeta$: the linear one separates *no* from
$\{\textit{some},\textit{all}\}$ and the quadratic separates *some* from
$\{\textit{no},\textit{all}\}$, so all three utterances receive distinct projections. Dropping the
constant column is deliberate, since a constant direction shifts $\varphi_S$ uniformly and is invisible to
the read-out (Eq. 12), so it would consume a degree of freedom without affecting the belief.

Since $m=2$, the terminating prior is an ordered pair $\mu_u=(\mu_{u,1},\mu_{u,2})\in\mathbb R^2$, one
weight per basis direction. Part B below reads what each component does.

---

### <a id="tc5-partb" name="tc5-partb"></a>Part B. probe report on the terminating prior $\mu_u$

Six settings of $\mu_u$ run against Code Cell 1. The configuration is held fixed
throughout: Gaussian base prior, $\Lambda=8$, $\theta_u=1$, $\theta_L=2.9444$ ($n=10$), $m=2$,
every $\sigma=1$, $K=101$. Every quantity below is evaluated in closed form (Eqs. 15-16, exact by
§8.6), so no integration step enters. Reported statistics are as defined at Eqs. (25)-(27). The
base prior is grid-renormalized, as Code Cell 1 renormalizes it.

#### <a id="tc5-1" name="tc5-1"></a>1. The structural result

$\mu_u$ is a **pure common-mode shift**. It translates all three beliefs together and can
move no contrast between them.

From Eq. (16), and writing $S=\sigma_L+\sigma_S$, $G=B^{\mathsf T}WB$:

$$\frac{\partial\varphi_u^\ast}{\partial\mu_u}
=\Big(I+\tfrac{\sigma_u\theta_u^2}{S}G\Big)^{-1},
\qquad
\frac{\partial\varphi_S^\ast}{\partial\mu_u}
=\frac{\sigma_L\theta_u}{S}\,B\Big(I+\tfrac{\sigma_u\theta_u^2}{S}G\Big)^{-1}.
\tag{32}$$

Neither expression contains $\varphi_L$, and $\varphi_L$ is the only utterance-dependent
quantity in the model. At the settings above ($\sigma=1$, $\theta_u=1$, $G=I$, $S=2$) this
collapses to

$$\frac{\partial\varphi_S^\ast}{\partial\mu_u}=\tfrac13 B,
\tag{33}$$

so raising $\mu_{u,j}$ by one unit adds exactly $\tfrac13 b_j$ to the settled log-density, *the same field for every utterance*. Verified numerically to $2.6\times10^{-15}$ (Code Cell 3).

The consequence is exact and worth stating as a test rather than a remark: the differences
$\varphi_S^\ast(u_1)-\varphi_S^\ast(u_2)$ are **bit-identical** across all six settings,
agreeing to $3.6\times10^{-15}$, which is machine epsilon and not a tolerance.

**At fixed $\theta_u$, $\mu_u$ therefore moves no contrast between utterances**, and this is a
measurement, not a reading. The qualifier is necessary: $\theta_u^\ast$ depends on $\mu_u$
(Appendix B), so once $\theta_u$ is learned, $\mu_u$ reaches contrasts through it. Sweeping
$\mu_u=v\,[1,1]$ from $v=0.2$ to $v=3.0$ carries $\theta_u^\ast$ from $-143.2577$ to $-8.8983$, and
with it $\max\lvert\varphi_S^\ast(\textit{all})-\varphi_S^\ast(\textit{some})\rvert$ from $5.8608$
to $5.8152$, a movement of $0.8\%$ that is nevertheless fifteen orders of magnitude above the
$3.6\times10^{-15}$ measured at fixed $\theta_u$. Everything in this report was computed at $\theta_u=1$ and so falls under the qualifier.

**What the structural result does not settle.** The criterion adopted in Text cell 4 Part C is not a contrast between
utterances: $\Delta_y$ compares one utterance's belief against its own literal listener
$q_{\mathrm{lit},y}$ (Eq. 37), and it is a $q$-mass (Eq. 27), nonlinear in $\varphi_S$. The
invariance above therefore does not transfer to it, and measurement confirms that it does not:
across the same six settings at $\theta_u=1$, $\Delta_{\textit{some}}$ runs $+0.0080$ to $+0.0109$
and $\Delta_{\textit{all}}$ runs $-0.3598$ to $-0.2914$, each varying by more than a fifth of its
own size, while the $\varphi_S$ contrasts stay bit-identical throughout. So §1 does
not settle in closed form whether $\mu_u$ is worth sweeping for an implicature under that criterion.
What it does settle is narrower and still worth having: $\mu_u$ moves nothing the model represents *about the relation between* utterances.
§4 makes that argument in full.

#### <a id="tc5-2" name="tc5-2"></a>2. What each position of the ordered pair does

The two basis columns have **definite and opposite parity**, which is what makes $m=2$ able
to separate all three utterances at all (Part A):

| | parity | $\langle\mathbf 1,b_j\rangle$ | $b_j(-6)$ | $b_j(0)$ | $b_j(+6)$ | shape |
|---|---|---|---|---|---|---|
| $b_1$ | **odd** | $-1.4\times10^{-17}$ | $-0.500$ | $0.000$ | $+0.500$ | monotone ramp |
| $b_2$ | **even** | $-3.6\times10^{-16}$ | $+0.645$ | $-0.323$ | $+0.645$ | symmetric "U" |

**Position 1: $\mu_{u,1}$ sets TILT.** Adding $\tfrac13b_1$ is adding a monotone ramp in
$\zeta$, which slides log-density from one end of the scale to the other. It moves every
utterance in the *same* direction and leaves spread essentially untouched.

**Position 2: $\mu_{u,2}$ sets WIDTH.** Adding $\tfrac13b_2$ raises both tails and lowers
the centre, pushing $q$-mass outward. It moves each utterance *away from the centre*, so its
effect on $\mathbb E[s]$ has opposite sign for beliefs sitting below and above $s=0.5$.

**The two entries share one scale.** Each is a coefficient on an orthonormal column, so
$\theta_u\mu_{u,j}b_j$ is a log-density field of quadrature norm $\theta_u\lvert\mu_{u,j}\rvert$:
a unit of either adds the same amount of log-density, distributed differently. One unit of
$\mu_{u,1}$ buys $1.00$ nat of end-to-end tilt; one unit of $\mu_{u,2}$ buys $0.97$ nats of
tail-minus-centre. The entries are therefore commensurable, which is what licenses the isotropic
default $[1,1]$ despite the two meaning different things.

Against the fields it competes with, $\mu_u$ is small:

| $\lVert\ell_0\rVert$ | $\lVert\varphi_L\rVert$ (*no*, *all*) | $\lVert\varphi_L\rVert$ (*some*) | $\lVert\mu_u\rVert$ |
|---|---|---|---|
| $30.34$ | $23.92$ | $13.99$ | $1.41$ |

At $\theta_u=1$ the terminating prior sits some $21\times$ below the prior it anchors. $\varphi_u$ is
accordingly set by $\varepsilon_S$ from below rather than by $\mu_u$ from above. The same weakness is
why the degeneracy at $\mu_u=0$ (§5, Appendix B) needed a closed-form analysis to detect: the quantity
whose vanishing destroys identifiability is an order of magnitude below everything around it.

**The midpoint of both scales is $0$, which Text cell 3 §3 forbids.** Each effect inverts under
$\mu_{u,j}\to-\mu_{u,j}$, and since $\varphi_S^\ast$ is affine in $\mu_u$ (§1) a signed pair
straddles the zero value exactly:

$$\varphi_S^\ast(+v\,e_j)+\varphi_S^\ast(-v\,e_j)=2\,\varphi_S^\ast(0),
\tag{34}$$

verified to $7.1\times10^{-15}$ for $j=1,2$ and $v\in\{0.25,0.5,1,2\}$ (Code Cell 3). Reading
$\mathbb E[s]$ across the same sweep separates the two shapes cleanly:

| entry | $v$ | *no* $(-v)$ | *no* $(+v)$ | *some* $(-v)$ | *some* $(+v)$ | *all* $(-v)$ | *all* $(+v)$ |
|---|---|---|---|---|---|---|---|
| $\mu_{u,1}$ | $0.5$ | 0.2573 | 0.2713 | 0.5328 | 0.5401 | 0.7287 | 0.7427 |
| $\mu_{u,1}$ | $2.0$ | 0.2371 | 0.2932 | 0.5217 | 0.5511 | 0.7068 | 0.7629 |
| $\mu_{u,2}$ | $0.5$ | 0.2693 | 0.2592 | 0.5360 | 0.5369 | 0.7307 | 0.7408 |
| $\mu_{u,2}$ | $2.0$ | 0.2843 | 0.2438 | 0.5348 | 0.5383 | 0.7157 | 0.7562 |

$\mu_{u,1}$ moves all three utterances the same way, up for $+v$ and down for $-v$. $\mu_{u,2}$
moves *no* and *all* in opposite directions, outward from the centre for $+v$ and inward for $-v$.

Zero being both the natural centre and the forbidden value is a property of the architecture rather
than an inconvenience: a tonic node carrying weight zero terminates nothing, so the chain has to end
in a commitment. Bogacz's $v_p$ carries the same requirement, never tested at zero. A sweep of
$\mu_u$ should therefore be symmetric about zero with zero excluded, and §6 does it on that shape, which turns out to remove rather more than a point.

#### <a id="tc5-3" name="tc5-3"></a>3. The six settings

| $\mu_u$ | $\mathbb E[s]$ *no* | *some* | *all* | $\mathrm{sd}(s)$ \| *some* | *all*−*some* | *some*−*no* |
|---|---|---|---|---|---|---|
| $[0,0]$ *(degenerate; see §5)* | 0.2643 | 0.5364 | 0.7357 | 0.2309 | $+0.1993$ | $+0.2722$ |
| $[0,1]$ | 0.2541 | 0.5373 | 0.7459 | 0.2327 | $+0.2086$ | $+0.2833$ |
| $[1,0]$ | 0.2785 | 0.5438 | 0.7496 | 0.2306 | $+0.2058$ | $+0.2653$ |
| $[1,1]$ **(default)** | 0.2683 | 0.5448 | 0.7597 | 0.2324 | $+0.2149$ | $+0.2765$ |
| $[1,2]$ | 0.2580 | 0.5459 | 0.7699 | 0.2342 | $+0.2239$ | $+0.2879$ |

behaviour the even parity of $b_2$ predicts.

#### <a id="tc5-4" name="tc5-4"></a>4. Why we do not read the contrast drift as a pragmatic effect

The last two columns drift: the *all*$-$*some* figure ranges $0.1993$–$0.2239$, a spread of
$0.0246$ on a mean of $0.2122$, or about 12% of its own size. That is large enough that it cannot be
waved past, which is why the argument below is made in full. Whether a movement of that kind
counts as a pragmatic effect is not something this project has settled, so what follows is an
argument rather than a classification.

The premises are three, and the first two are measurements rather than judgements.

1. Across the six rows, the only quantity that varies is $\mu_u$, with $\theta_u$ held at $1$.

2. The differences $\varphi_S^\ast(u_1)-\varphi_S^\ast(u_2)$ are invariant to $\mu_u$ to
   $3.6\times10^{-15}$ (§1), which is machine epsilon. Whatever the model represents about the relation
   *between* utterances is therefore bit-for-bit unchanged across the table.

3. $\mathbb E[s]$ (Eq. 25) is a nonlinear functional of $\varphi_S$: a softmax, then $\varsigma$.
   Adding one and the same field to three different read-outs consequently moves their means by
   three different amounts.

Premises 1 and 2 leave the nonlinearity of premise 3 as the only available source of the drift, and
since premise 2 is exact to machine epsilon it is not merely the *likeliest* source but the only
one available at any size, this one included. We therefore do not think the drift should be read as
a pragmatic effect: it is a property of the statistic we chose to report, and the model computed no
corresponding change in how it relates the three utterances.

**What the argument does not establish.** This claim presupposes that a pragmatic effect ought to show up as a
contrast in $\varphi_S$, and not in a summary of $\varphi_S$ alone. Someone who located pragmatic content directly in $\mathbb E[s]$, or in
some other functional of the read-out, could read the same six rows differently, and nothing measured
here would contradict them. The argument also does not make a claim about whether the model produces a
pragmatic effect anywhere *else*.

The practical consequence is the same either way, and is the reason the section exists: the table
should not be read without §1 beside it, because the columns move for a reason that has nothing to do
with $\mu_u$ acting on the utterances differently.

#### <a id="tc5-5" name="tc5-5"></a>5. Disposition

$\mu_u$ is fixed at $[1,1]$. It **cannot** be $0$: the terminating contribution $\theta_u B\mu_u$
would vanish for every $\theta_u$, leaving $\theta_u$ unidentifiable (Text cell 3 §3, Appendix B).
At $\mu_u=0$ the terminating level stops regularizing $\theta_u$ altogether.

Nothing is stipulated about $\varepsilon_u$ by this choice: Eq. (11) gives $r_u=\varphi_u-\mu_u$, so
the terminating error unit reports the offset of the utility state from $\mu_u$, which Eq. (19) then
opposes. This is Bogacz's $v_p$ prior, and $\mu_u\neq0$ is his premise as much as ours.

Two things this probe does **not** settle. It says what the two components of $\varphi_u$ do
to the belief, but not whether $m=2$ is *sufficient* as well as necessary, since Part A establishes necessity by a parity argument and leaves sufficiency open. (The *shape* of the sweep is settled in
§6, which replaces the six lattice points with the plane.) And it generalizes no
further than $\mu_u$. What $\mu_u$ shares with $\ell_0$ is a position in the closed forms rather than a term in $g_S$, which
carries neither of them, since $\ell_0$ sits in $g_L$ (Eq. 9): $\mu_u$ enters Eq. (16) as the terminating prior and $\ell_0$ enters
Eqs. (15)–(16) only inside the combination $\ell_0-\varphi_L$, and $\varphi_S^\ast$ is affine in
each with an utterance-independent coefficient. That is why neither can move a $\varphi_S$ contrast. The
property is theirs alone: every other quantity touching an utterance-dependent one can move a
contrast, and measurably does: $\Lambda$ and $\theta_L$ through $\chi_y$, the variances through the weighting of $\varphi_L$
against $\ell_0$ in Eq. (15), $B$ through the lexical projections $\kappa_y$ of Eq. (29), and the slow parameter
$\theta_u$ through $B\varphi_u$. Note that $\theta_u$ is *not* a fixed quantity: it is
variational on the slow timescale (Eq. 20), and belongs in this list for its effect, not its
status.

#### <a id="tc5-6" name="tc5-6"></a>6. The sweep on the shape the constraints license

The six settings above are lattice points in one quadrant, which is the wrong shape for this
parameter on three counts. $\mu_u$ lives in $\mathbb R^2$, so a sweep of it is a sweep of a *plane*.
The excluded set is not a point: Text cell 3 §3 requires $\mu_u\neq0$, and Appendix B sharpens that to
$\langle\mu_u,\sum_y c_y\rangle\neq0$, which is a condition on **direction** and removes two whole
rays. And the pair $(\theta_u,\mu_u)$ carries an exact symmetry, so half the plane is redundant.
Code Cell 3 sweeps it in polar coordinates, at five radii and twelve directions.

**The symmetry, first, because it halves the work.** $\varphi_u^\ast$ solves
$(I+\sigma_u\theta_u^2G/S)\,\varphi_u^\ast=\mu_u+\sigma_u\theta_u c/S$. The matrix sees only
$\theta_u^2$, so $(\theta_u,\mu_u)\mapsto(-\theta_u,-\mu_u)$ negates the right-hand side, sends
$\varphi_u^\ast\mapsto-\varphi_u^\ast$, and leaves $g_S=\theta_u B\varphi_u^\ast$, and therefore $\varphi_S^\ast$, untouched:

$$\varphi_S^\ast(\theta_u,\mu_u)=\varphi_S^\ast(-\theta_u,-\mu_u),
\tag{35}$$

verified to $3.6\times10^{-15}$ over the whole sweep. So the plane need only be covered at one sign
of $\theta_u$; the other sign is a reflection, not a second experiment. It shows up directly in the
sweep as $\theta_u^\ast(\text{angle}+180^\circ)=-\theta_u^\ast(\text{angle})$, exact down the table.

**The excluded rays.** At the settings here $\sum_y c_y=[+9.1191,\,-50.0310]$, so
$\langle\mu_u,\sum_y c_y\rangle=0$ on the two rays at $10.33^\circ$ and $190.33^\circ$. Approaching
them, $\theta_u^\ast$ diverges: at $r=1$ the sweep reads $127.94$ at $0^\circ$ against $23.40$ at
$90^\circ$, and at $r=0.25$ it reads $514.17$ at $0^\circ$. That is Appendix B's condition made visible: what
$\mu_u\neq0$ excludes is a point, what identifiability actually excludes is a direction, and the
default $\mu_u=[1,1]$ sits $34.67^\circ$ from the nearer ray. That margin is a property of this
$\ell_0$ and this $\Lambda$, since $\sum_y c_y$ moves with both.

**What the plane confirms and what it adds.** At fixed $\theta_u=1$ the $\varphi_S$ contrast is
invariant over all sixty settings to $3.6\times10^{-15}$, so the
six-row result of §1 was a sample of a plane-wide fact. What varies is $\theta_u^\ast$, and it varies by a factor of $90$ across the
plane, from $5.729$ at $r=4,\,90^\circ$ to $514.17$ at $r=0.25,\,0^\circ$; the *all*$-$*some*
contrast it carries ranges $13.6510$ to $13.9242$. Set against $3.6\times10^{-15}$ at fixed
$\theta_u$, that is the whole of $\mu_u$'s reach: it touches the utterances through the slow
parameter and by no other route.

````python
# === Code Cell 3: the probes reported in Text cell 5. EVALUATION ===
#
# Text cell 5 states two results in prose and tables. This cell computes them, so that
# the numbers there are reproducible rather than quoted.
#
#   Part A, the utility basis dimension: any single basis function of definite
#             parity collapses a pair of utterances, so m = 1 cannot separate all three.
#   Part B, the terminating prior mu_u: at fixed theta_u it moves no contrast, and
#             the drift in the reported E[s] contrasts is read-out curvature.
#
# Everything is evaluated in closed form (Eqs. 15-16, exact by Sec. 8.6).

import math
import torch


# ============================================================================
# PART A. THE BASIS DIMENSION
# ============================================================================

def basis_parity_probe(net, verbose=True):
    """Text cell 5 Part A: which utterance pairs a single basis function collapses.

    The u level sees the lexical field only through the lexical projections
    kappa_y = B^T W phi_L (Eq. 29, not Appendix B's c_y = B^T W (ell_0 - phi_L)), so
    two utterances with equal projections are indistinguishable to it.
    The three exclusion sets are built from one pair of thresholds +-theta_L, so their
    pairwise differences are an interval symmetric about the origin and a mirror-image
    pair; an ODD basis function annihilates the first and an EVEN one the second.
    """
    zeta, weights = net.zeta, net.weights
    chi = {y: exclusion_indicator(y, zeta, net.theta_L) for y in UTTERANCES}
    candidates = [
        ("zeta", zeta, "odd"), ("zeta^3", zeta ** 3, "odd"), ("sin(zeta)", torch.sin(zeta), "odd"),
        ("zeta^2", zeta ** 2, "even"), ("|zeta|", zeta.abs(), "even"), ("cos(zeta)", torch.cos(zeta), "even"),
        ("exp(0.3 zeta)", torch.exp(0.3 * zeta), "none"), ("sigmoid(zeta)", torch.sigmoid(zeta), "none"),
    ]
    rows = []
    for name, profile, parity in candidates:
        projections = {y: float((weights * chi[y] * profile).sum()) for y in UTTERANCES}
        collapsed = [
            (a, b) for a in UTTERANCES for b in UTTERANCES
            if a < b and abs(projections[a] - projections[b]) < 1e-9
        ]
        rows.append((name, parity, projections, collapsed))

    if verbose:
        print("PART A. BASIS PARITY AND UTTERANCE SEPARATION")
        print(f"  theta_L = {net.theta_L:.4f} (n = {net.num_atoms})")
        print()
        print(f"  {'basis function':<16}{'parity':<8}"
              + "".join(f"{y:>13}" for y in UTTERANCES) + "   collapsed pair")
        for name, parity, projections, collapsed in rows:
            pair = f"{collapsed[0][0]} = {collapsed[0][1]}" if collapsed else "none"
            print(f"  {name:<16}{parity:<8}"
                  + "".join(f"{projections[y]:>13.5f}" for y in UTTERANCES)
                  + f"   {pair}")
        print()
        print("  Every ODD profile collapses {some, all}; every EVEN profile collapses")
        print("  {no, all}. m = 1 with a parity-definite basis therefore cannot separate")
        print("  all three, whatever the profile. The implementation takes m = 2, one of")
        print("  each parity, verified below.")
        print()
        for j in range(net.num_basis):
            column = net.basis[:, j]
            reversed_column = torch.flip(column, [0])
            parity = ("odd" if float((column + reversed_column).abs().max()) < 1e-9
                      else "even" if float((column - reversed_column).abs().max()) < 1e-9
                      else "none")
            print(f"    B column {j + 1}: parity {parity:<5} "
                  f"<1,b> = {float((net.weights * column).sum()):+.2e}  "
                  f"b(-Z) = {float(column[0]):+.4f}  b(0) = {float(column[len(column) // 2]):+.4f}  "
                  f"b(+Z) = {float(column[-1]):+.4f}")
        separations = {
            y: tuple(round(v, 9) for v in net.project(net.lexical_field(y)).tolist())
            for y in UTTERANCES
        }
        distinct = len(set(separations.values())) == len(UTTERANCES)
        print(f"    projections distinct across all three utterances: {distinct}")
        for y, value in separations.items():
            print(f"      {y:<6} kappa = {value}")
        print()
    return rows


# ============================================================================
# PART B. THE TERMINATING PRIOR mu_u
# ============================================================================

def mu_u_probe(net, settings=None, theta_u=1.0, verbose=True):
    """Text cell 5 Part B: what each component of mu_u does, and what it cannot do.

    Everything here holds theta_u fixed. That qualifier is essential: theta_u* depends
    on mu_u (Sec. 8.5), so through a learned theta_u the invariance below does not hold.
    """
    settings = settings or [(0., 0.), (0., 1.), (1., 0.), (1., 1.), (1., 2.), (2., 1.)]
    proportion = logistic(net.zeta)

    def spawn(pair):
        return net.respawn(mu_u=torch.tensor(pair, dtype=net.dtype))

    # === sensitivity: d phi_S* / d mu_u, against the analytic prediction ===
    gram = net.basis.T @ (net.weights[:, None] * net.basis)
    total_sigma = net.sigma_lexical + net.sigma_state
    predicted = (net.sigma_lexical * theta_u / total_sigma) * (
        net.basis @ torch.linalg.inv(
            torch.eye(net.num_basis, dtype=net.dtype)
            + net.sigma_utility * theta_u ** 2 * gram / total_sigma
        )
    )
    zero = spawn((0., 0.))
    worst_sensitivity = 0.0
    for j in range(net.num_basis):
        bump = [0.0, 0.0]; bump[j] = 1.0
        moved = spawn(tuple(bump)).closed_form_fixed_point("some", theta_u=theta_u)[0]
        base = zero.closed_form_fixed_point("some", theta_u=theta_u)[0]
        worst_sensitivity = max(
            worst_sensitivity, float((moved - base - predicted[:, j]).abs().max())
        )

    # === the six settings ===
    rows = []
    for pair in settings:
        probe = spawn(pair)
        means, spreads = {}, {}
        for y in UTTERANCES:
            phi_S, _ = probe.closed_form_fixed_point(y, theta_u=theta_u)
            q = probe.read_out(phi_S)
            mean = float((probe.weights * q * proportion).sum())
            means[y] = mean
            spreads[y] = float(((probe.weights * q * (proportion - mean) ** 2).sum()) ** 0.5)
        rows.append((pair, means, spreads))

    # === the invariance: phi_S contrasts must not move with mu_u ===
    def contrast(pair):
        probe = spawn(pair)
        return (probe.closed_form_fixed_point("all", theta_u=theta_u)[0]
                - probe.closed_form_fixed_point("some", theta_u=theta_u)[0])
    reference = contrast(settings[0])
    worst_invariance = max(
        float((contrast(pair) - reference).abs().max()) for pair in settings[1:]
    )

    if verbose:
        print("PART B. THE TERMINATING PRIOR mu_u   (theta_u fixed at %g)" % theta_u)
        print()
        print(f"  sensitivity d(phi_S*)/d(mu_u) matches the analytic "
              f"(sigma_L theta_u / S) B (I + sigma_u theta_u^2 G / S)^-1")
        print(f"    max discrepancy = {worst_sensitivity:.2e}")
        print()
        print(f"  {'mu_u':<12}" + "".join(f"{y + ' E[s]':>13}" for y in UTTERANCES)
              + f"{'sd|some':>10}{'all-some':>11}{'some-no':>10}")
        for pair, means, spreads in rows:
            print(f"  {str(list(pair)):<12}"
                  + "".join(f"{means[y]:>13.4f}" for y in UTTERANCES)
                  + f"{spreads['some']:>10.4f}"
                  + f"{means['all'] - means['some']:>+11.4f}"
                  + f"{means['some'] - means['no']:>+10.4f}")
        print()
        print(f"  phi_S contrasts across all settings: max deviation = {worst_invariance:.2e}")
        print("  (machine epsilon: mu_u translates the common baseline and moves no")
        print("  contrast, AT FIXED theta_u.)")
        print()
        contrasts = [m['all'] - m['some'] for _, m, _ in rows]
        print(f"  reported E[s] contrast all-some ranges {min(contrasts):.4f} to "
              f"{max(contrasts):.4f}, a spread of {max(contrasts) - min(contrasts):.4f}.")
        print("  Since the phi_S contrasts are invariant to machine epsilon, that spread")
        print("  is produced entirely by the nonlinearity of E[s] (Eq. 25: a softmax,")
        print("  then varsigma) acting on a common additive shift, which is the argument")
        print("  of Text cell 5 Part B Sec. 4, not an effect of mu_u on the utterances.")
        print()
    return rows, worst_sensitivity, worst_invariance


# ============================================================================
# RUN
# ============================================================================

basis_rows = basis_parity_probe(evaluation_network)
mu_rows, mu_sensitivity, mu_invariance = mu_u_probe(evaluation_network)

# ============================================================================
# PART B2. THE SCALES OF mu_u, AND THE SIGN OF EACH ENTRY
# ============================================================================

def mu_u_scale_probe(net, theta_u=1.0, verbose=True):
    """What units the two entries of mu_u sit on, and where the middle of each is.

    Both are coefficients on ORTHONORMAL columns of B, so theta_u * mu_u_j * b_j is a
    log-density field of quadrature-norm theta_u * |mu_u_j|: the two entries are
    commensurable and differ in shape, not in size. The shapes are reported as the
    characteristic swing each unit buys.
    """
    columns = [net.basis[:, j] for j in range(net.num_basis)]
    middle = len(net.zeta) // 2
    rows = []
    for j, column in enumerate(columns):
        rows.append({
            "norm": float(net.squared_norm(column)) ** 0.5,
            "at_low": float(column[0]),
            "at_mid": float(column[middle]),
            "at_high": float(column[-1]),
            "end_to_end": float(column[-1] - column[0]),
            "tail_minus_centre": float(column[0] - column[middle]),
        })
    references = {
        "||ell_0||": float(net.squared_norm(net.base_log_prior)) ** 0.5,
        "||mu_u||": float(net.mu_u.norm()),
    }
    for y in UTTERANCES:
        references[f"||phi_L|| ({y})"] = float(net.squared_norm(net.lexical_field(y))) ** 0.5

    if verbose:
        print("PART B2. SCALES OF THE TWO ENTRIES OF mu_u")
        print()
        print("  mu_u_j is a coefficient on an orthonormal column of B, so")
        print("  theta_u * mu_u_j * b_j is a log-density field of norm theta_u * |mu_u_j|.")
        print()
        print(f"  {'column':<10}{'||b_j||':>10}{'b(-Z)':>10}{'b(0)':>10}{'b(+Z)':>10}"
              f"{'end-to-end':>13}{'tail-centre':>14}")
        for j, r in enumerate(rows):
            print(f"  b_{j + 1}{'':<8}{r['norm']:>10.6f}{r['at_low']:>10.4f}{r['at_mid']:>10.4f}"
                  f"{r['at_high']:>10.4f}{r['end_to_end']:>13.4f}{r['tail_minus_centre']:>14.4f}")
        print()
        print("  one unit of mu_u_1 buys %.2f nats of end-to-end tilt;" % rows[0]["end_to_end"])
        print("  one unit of mu_u_2 buys %.2f nats of tail-minus-centre." % rows[1]["tail_minus_centre"])
        print()
        print("  reference magnitudes in the same norm:")
        for name, value in references.items():
            print(f"    {name:<20} {value:9.4f}")
        print(f"    -> at theta_u = {theta_u:g} the terminating prior is "
              f"{references['||ell_0||'] / references['||mu_u||']:.0f}x below the prior it anchors.")
        print()
    return rows, references


def mu_u_sign_probe(net, magnitudes=(0.25, 0.5, 1.0, 2.0), theta_u=1.0, verbose=True):
    """Signed sweep of each entry about zero, testing that zero is the midpoint.

    phi_S* is affine in mu_u (Sec. 1), so a signed pair must straddle the mu_u_j = 0
    value exactly:  phi_S*(+v e_j) + phi_S*(-v e_j) = 2 phi_S*(0).  Verifying that
    identity is what establishes 0 as the midpoint of each scale, the value Sec. 3
    forbids, since the terminating contribution theta_u B mu_u vanishes there.

    The other entry is held at 0 while one is swept, which is legitimate here because
    only the whole vector is required to be nonzero.
    """
    proportion = logistic(net.zeta)
    zero_field = net.respawn(
        mu_u=torch.zeros(net.num_basis, dtype=net.dtype)
    ).closed_form_fixed_point("some", theta_u=theta_u)[0]

    rows, worst_midpoint = [], 0.0
    for j in range(net.num_basis):
        for magnitude in magnitudes:
            entry = {"component": j + 1, "magnitude": magnitude, "means": {}}
            fields = {}
            for sign in (+1, -1):
                vector = torch.zeros(net.num_basis, dtype=net.dtype)
                vector[j] = sign * magnitude
                probe = net.respawn(mu_u=vector)
                fields[sign] = probe.closed_form_fixed_point("some", theta_u=theta_u)[0]
                entry["means"][sign] = {
                    y: float((probe.weights
                              * probe.read_out(probe.closed_form_fixed_point(y, theta_u=theta_u)[0])
                              * proportion).sum())
                    for y in UTTERANCES
                }
            entry["midpoint_error"] = float(
                (fields[+1] + fields[-1] - 2.0 * zero_field).abs().max()
            )
            worst_midpoint = max(worst_midpoint, entry["midpoint_error"])
            rows.append(entry)

    if verbose:
        print("  SIGNED SWEEP ABOUT ZERO")
        print()
        print(f"  {'entry':<7}{'|v|':>6}" + "".join(f"{y + ' -v':>12}{y + ' +v':>12}" for y in UTTERANCES)
              + f"{'midpoint err':>15}")
        for entry in rows:
            cells = "".join(
                f"{entry['means'][-1][y]:>12.4f}{entry['means'][+1][y]:>12.4f}" for y in UTTERANCES
            )
            print(f"  mu_u_{entry['component']:<2}{entry['magnitude']:>6.2f}{cells}"
                  f"{entry['midpoint_error']:>15.1e}")
        print()
        print(f"  phi_S*(+v) + phi_S*(-v) = 2 phi_S*(0) to {worst_midpoint:.1e} in every case.")
        print("  Zero is therefore the exact midpoint of both scales. Sec. 3 forbids it:")
        print("  the terminating contribution theta_u B mu_u vanishes there and theta_u")
        print("  loses identifiability (Sec. 8.5). A terminating prior sits off-centre.")
        print()
    return rows, worst_midpoint


scale_rows, scale_references = mu_u_scale_probe(evaluation_network)
sign_rows, sign_midpoint_error = mu_u_sign_probe(evaluation_network)


# ============================================================================
# PART B3. THE mu_u PLANE, SWEPT PROPERLY
# ============================================================================

def mu_u_plane_probe(net, *, radii=(0.25, 0.5, 1.0, 2.0, 4.0), sectors=12, verbose=True):
    """A 2-D sweep of mu_u, on the shape Sec. 3 and Appendix B actually license.

    Three things make the six hand-picked settings of Text cell 5 Part B Sec. 3 the
    wrong shape, and all three are structural rather than matters of taste.

    1. mu_u lives in R^2, so a sweep of it is a sweep of a PLANE. The six settings
       are lattice points in one quadrant.
    2. The origin is excluded, not merely awkward: Sec. 3 requires mu_u != 0, and
       Appendix B sharpens that to <mu_u, sum_y c_y> != 0, which is a condition on
       DIRECTION and removes two whole rays rather than one point. Polar coordinates
       are the ones the constraint is stated in, so the sweep uses them.
    3. The pair (theta_u, mu_u) carries an exact symmetry. phi_u* solves
       (I + sigma_u theta^2 G / S) phi_u* = mu_u + sigma_u theta c / S; the matrix
       sees only theta^2, so (theta, mu) -> (-theta, -mu) sends phi_u* -> -phi_u* and
       leaves g_S = theta B phi_u* and hence phi_S* alone. Half the plane is therefore
       redundant given both signs of theta_u, and the sweep is taken modulo it: the
       circle is swept at one sign of theta_u, and the identity is verified rather
       than assumed.
    """
    proportion = logistic(net.zeta)
    coupling = sum(
        (net.project(net.base_log_prior - net.lexical_field(y)) for y in UTTERANCES),
        torch.zeros(net.num_basis, dtype=net.dtype),
    )
    degenerate = math.degrees(math.atan2(-float(coupling[0]), float(coupling[1]))) % 180.0

    def spawn(radius, degrees):
        angle = math.radians(degrees)
        return net.respawn(mu_u=radius * torch.tensor(
            [math.cos(angle), math.sin(angle)], dtype=net.dtype))

    angles = [360.0 * k / sectors for k in range(sectors)]

    # === (3) the exact symmetry, verified over the plane ===
    worst_symmetry = 0.0
    for radius in radii:
        for degrees in angles:
            plus, minus = spawn(radius, degrees), spawn(radius, degrees + 180.0)
            for theta in (0.7, 1.0, 4.0):
                for y in UTTERANCES:
                    worst_symmetry = max(worst_symmetry, float((
                        plus.closed_form_fixed_point(y, theta_u=theta)[0]
                        - minus.closed_form_fixed_point(y, theta_u=-theta)[0]
                    ).abs().max()))

    # === (1) at fixed theta_u the whole plane is common-mode ===
    reference = None
    worst_invariance = 0.0
    for radius in radii:
        for degrees in angles:
            probe = spawn(radius, degrees)
            contrast = (probe.closed_form_fixed_point("all", theta_u=1.0)[0]
                        - probe.closed_form_fixed_point("some", theta_u=1.0)[0])
            if reference is None:
                reference = contrast
            worst_invariance = max(
                worst_invariance, float((contrast - reference).abs().max()))

    # === (2) what does vary: theta_u*, and the contrast it carries ===
    rows = []
    for radius in radii:
        for degrees in angles:
            probe = spawn(radius, degrees)
            inner = float(probe.mu_u @ coupling)
            entry = {"radius": radius, "degrees": degrees, "inner": inner}
            if abs(inner) < 1e-9:
                entry["theta_star"] = float("nan")
                entry["contrast"] = float("nan")
            else:
                star = theta_u_stationary_point(probe)
                entry["theta_star"] = star
                fields = {y: probe.closed_form_fixed_point(y, theta_u=star)[0]
                          for y in UTTERANCES}
                entry["contrast"] = float(
                    probe.squared_norm(fields["all"] - fields["some"])) ** 0.5
                entry["expected_s"] = {
                    y: float((probe.weights * probe.read_out(fields[y]) * proportion).sum())
                    for y in UTTERANCES
                }
            rows.append(entry)

    if verbose:
        print("PART B3. THE mu_u PLANE")
        print()
        print(f"  sum_y c_y = [{coupling[0]:+.4f}, {coupling[1]:+.4f}], so Appendix B's")
        print(f"  condition <mu_u, sum_y c_y> != 0 fails on the two rays at")
        print(f"  {degenerate:.2f} deg and {degenerate + 180:.2f} deg. Those are excluded, and so is")
        print("  the origin. What is removed is a DIRECTION, not a point.")
        print()
        print(f"  exact symmetry  phi_S*(theta, mu) = phi_S*(-theta, -mu):")
        print(f"    max deviation over {len(radii)}x{sectors} settings, 3 |theta_u|, 3 utterances "
              f"= {worst_symmetry:.1e}")
        print("    (bit-identical, so the sweep below covers the plane at one sign of")
        print("     theta_u and the other sign is obtained by reflection, not re-run.)")
        print()
        print(f"  common-mode at fixed theta_u = 1:")
        print(f"    max deviation of the all-some phi_S contrast over the plane "
              f"= {worst_invariance:.1e}")
        print("    (the six settings of Sec. 3 were a sample of this; the plane confirms it.)")
        print()
        print("  theta_u* over the plane (the quantity that DOES move), by direction:")
        print(f"    {'angle':>8}" + "".join(f"{f'r={r:g}':>12}" for r in radii))
        for degrees in angles:
            cells = []
            for radius in radii:
                entry = next(e for e in rows
                             if e["radius"] == radius and e["degrees"] == degrees)
                value = entry["theta_star"]
                cells.append("  degenerate" if value != value else f"{value:>12.3f}")
            near = "  <-- near a degenerate ray" if min(
                abs(degrees - degenerate), abs(degrees - degenerate - 180)) < 20 else ""
            print(f"    {degrees:>8.0f}" + "".join(cells) + near)
        print()
        finite = [e for e in rows if e["theta_star"] == e["theta_star"]]
        widest = max(finite, key=lambda e: abs(e["theta_star"]))
        narrowest = min(finite, key=lambda e: abs(e["theta_star"]))
        print(f"    |theta_u*| ranges {abs(narrowest['theta_star']):.3f} "
              f"(r={narrowest['radius']:g}, {narrowest['degrees']:.0f} deg) to "
              f"{abs(widest['theta_star']):.3f} "
              f"(r={widest['radius']:g}, {widest['degrees']:.0f} deg), a factor of "
              f"{abs(widest['theta_star']) / abs(narrowest['theta_star']):.0f}.")
        contrasts = [e["contrast"] for e in finite]
        print(f"    the all-some contrast it carries ranges {min(contrasts):.4f} to "
              f"{max(contrasts):.4f}, a spread of {max(contrasts) - min(contrasts):.4f}")
        print(f"    against {worst_invariance:.1e} at fixed theta_u. mu_u reaches the")
        print("    utterances only through the slow parameter, and that is the whole of it.")
        print()
        default = math.degrees(math.atan2(1.0, 1.0))
        gap = min(abs(default - degenerate), abs(default - degenerate - 180.0),
                  abs(default - degenerate + 180.0))
        print(f"  Note theta_u*(angle + 180) = -theta_u*(angle) exactly down the table:")
        print("  that is the symmetry above, read off the sweep rather than assumed.")
        print()
        print(f"  The default mu_u = [1, 1] sits at r = {math.sqrt(2):.3f}, {default:.0f} deg,")
        print(f"  {gap:.2f} deg from the nearer degenerate ray at {degenerate:.2f} deg, comfortably")
        print("  clear of it, but the margin is a property of this ell_0 and this Lambda,")
        print("  since sum_y c_y moves with both.")
        print()

    return rows, worst_symmetry, worst_invariance


plane_rows, plane_symmetry, plane_invariance = mu_u_plane_probe(evaluation_network)
````

## <a id="tc6" name="tc6"></a>Text cell 6: the $\Lambda\times\alpha$ sweep

Text cell 5 Part B showed that $\mu_u$ and $\ell_0$ move no $\varphi_S$ contrast **at fixed
$\theta_u$**. Neither does so by sitting in $g_S$, which carries neither of them
since $\ell_0$ is in $g_L$ (Eq. 9), but by entering the closed forms only through
utterance-independent terms: $\mu_u$ as the terminating prior in Eq. (16), $\ell_0$ only inside the
combination $\ell_0-\varphi_L$ of Eqs. (15)–(16), and $\varphi_S^\ast$ is affine in each with a
coefficient that does not depend on the utterance. (Through a learned $\theta_u$ they *do* reach
contrasts, since $\theta_u^\ast$ depends on both; everything below holds $\theta_u=1$, and Code
Cell 2's `theta_u_dependence` reports what that pinning costs.) That property is theirs alone:
$\Lambda$, $\theta_L$, $B$, the variances, and the slow parameter $\theta_u$ all touch
utterance-dependent quantities and can move contrasts. This cell takes $\Lambda$, the one that mediates directly between the lexicon and
the world belief, and maps it against the concentration of $\ell_0$, the quantity it competes with in
Eq. (15).

The motivating case is a world prior certain of the top state, $P_0(s{=}1)=1$. It is
not representable here: $s$ lies in the **open** interval $(0,1)$, $\zeta=\operatorname{logit}(1)$
is off the grid, and $\ell_0=\log p_0$ would be $-\infty$ almost everywhere, which makes the
Gaussian residual $r_S$ infinite. The limit family $\mathrm{Beta}(\alpha,1)$ concentrates at
$s=1$ as $\alpha$ grows and is swept in its place.

---

### <a id="tc6-quantities-swept" name="tc6-quantities-swept"></a>Quantities swept

**Prior concentration $\alpha$.** This is the sweep's row coordinate. For $\alpha>0$ the base prior is
$p_0=\mathrm{Beta}(\alpha,1)$ over the proportion, pushed forward to $\zeta$ by the change of
variables of Text cell 4 Part D. Writing $\beta$ for the second shape parameter of the Beta and
$B(\cdot,\cdot)$ for its normalizing constant, the push-forward of $\mathrm{Beta}(\alpha,\beta)$ to
$\zeta$ is $\alpha\log s+\beta\log(1-s)-\log B(\alpha,\beta)$, which at $\beta=1$, where
$B(\alpha,1)=1/\alpha$, collapses to

$$\ell_0(\zeta)=\alpha\log\varsigma(\zeta)+\log\!\big(1-\varsigma(\zeta)\big)+\log\alpha .
\tag{36}$$

$\alpha=1$ is $\mathrm{Uniform}(0,1)$, the flat condition of Part D. As $\alpha$ grows the family
concentrates at $s=1$: its mean is $\alpha/(\alpha+1)$, and its mode *in $\zeta$* sits at the same
place, since $\partial_\zeta\ell_0=0$ gives $\alpha(1-s)=s$. That mode is not the mode in $s$, which
$\mathrm{Beta}(\alpha,1)$ places at the boundary; the Jacobian moves it. $\alpha\to\infty$ is the
limit standing in for the unrepresentable $P_0(s{=}1)=1$.

*Grid caveat.* The concentration a truncated grid can represent is bounded, and the bound bites
gradually rather than at a threshold. On the default grid ($|\zeta|\le6$) the prior mass carried by
the top node runs $0.0095$ at $\alpha=64$, $0.0445$ at $\alpha=300$, $0.1512$ at $\alpha=1024$ and
$0.5477$ at $\alpha=4096$; what fails first is not the representation of the prior but the model's
ability to tell the utterances apart, since the spread $D$ of Eq. (38) collapses $0.0061\to0.0009\to
0.0002$ over $\alpha=64,256,1024$. $\mathbb E[s]$ tends to about $0.984$ for every utterance rather than to
$\varsigma(6)=0.9975$, since the belief spreads over the top several nodes rather than concentrating
on the last one. The top rows of each map therefore measure the grid as much as the model; widening
the grid moves the ceiling without changing the picture.

**Lexical strength $\Lambda$.** This is the sweep's column coordinate, entering
$\varphi_L=\Lambda\chi_y$ exactly as in Eq. (6). No other quantity is varied: $\theta_u=1$,
$\mu_u=[1,1]$, $n=10$ so $\theta_L=\log19=2.9444$ (Eq. A5), $m=2$, every $\sigma=1$. $\mu_u\neq0$
is required by Text cell 3 §3, and at fixed $\theta_u$ it is common-mode across utterances. That
does not make it inert here. Over a hundredfold range of $\lvert\mu_u\rvert$, both signs included,
the leak map barely moves (spread under $0.02$), but at $\alpha=1$, $\Lambda=8$ the same range
moves $\mathbb E[s]$ by $0.14$ and $\Delta_{\textit{some}}$ from $+0.0422$ to $+0.2457$, and at
$\alpha=16$, $\Lambda=128$ it changes the sign of $\Delta_{\textit{some}}$. That is the failure
mode of Text cell 5 Part B §4: a common-mode field moves no $\varphi_S$ contrast but does move a
statistic taken after the softmax. The maps are drawn at $\mu_u=[1,1]$ throughout.

### <a id="tc6-statistics" name="tc6-statistics"></a>Statistics

Recalled from Text cell 4: the **leak** is the $q$-mass over the region the observed entry excludes;
$\mathbb E[s]$ is Eq. (25); and $P(R)$, the $q$-mass over a region, is Eq. (27), written $P(R;q)$ below where the read-out it is taken under must be named. The *all-region* is $R=\{\zeta>\theta_L\}$.

Four further statistics appear here. The first is Part C's shift, restated for the maps; the other
three are used only in this cell. Write $q_y$ for the settled read-out under utterance $y$ and
$q_{\mathrm{lit},y}$ for the literal listener of Part C, the untempered literal posterior
$\varphi_S=\ell_0-\varphi_L$.

We call

$$\Delta_y=P(\text{all-region};q_y)-P(\text{all-region};q_{\mathrm{lit},y})
\tag{37}$$

the **shift** for $y$: how much $q$-mass separates the settled belief from the literal listener
over the region where *all* is true. Two things in the hierarchy produce it, and they come apart
at $\theta_u=0$, where the field is $(\ell_0-\varphi_L)/2$ and $\Delta_y$ is already non-zero by
the tempering alone; the remainder is the utility level. The tempering is the larger of the two in
$74$ of the $121$ cells of this plane, and they carry opposite signs often enough that
$\Delta_{\textit{some}}$ and the utility level's own contribution disagree in sign in $43$ of the
$121$ cells. So $\Delta_y$ is the pragmatic-literal gap and not a measure of the utility level;
Eq. (39) below is the statistic that isolates the latter. Part C's first condition is
$\Delta_{\textit{some}}<0$. The baseline is the one Part D tabulates, so a cell of this plane and
a row of that table are the same measurement; the delta-like row, $\alpha=64$ and $\Lambda=512$,
is such a cell and reads $-0.3488$ in both places. Part C's second condition,
$P(\text{all-region};q_y)<\tfrac12$, is not a difference and so is not mapped; it is reported at
the end of this cell.

We call

$$D=\max_y\mathbb E[s\mid y]-\min_y\mathbb E[s\mid y]
\tag{38}$$

the **spread**: how far apart the three utterances' expected proportions lie. $D\to0$ says the model
has stopped distinguishing them, whatever it believes.

We call

$$\Delta_{\textit{some}}-\Delta_{\textit{no}}
\tag{39}$$

the **differential**. The tempering enters $\Delta_y$ at nearly the same size for every utterance,
so subtracting the *no* shift removes most of it and leaves the utterance-specific part: between
*some* and *no* the tempering differs by a median $0.0007$ and at most $0.0988$, against a median
$0.0153$ and a maximum $0.7523$ for the utility level. The differential's sign accordingly agrees
with the utility level's own in $109$ of the $121$ cells, where $\Delta_{\textit{some}}$ agrees in
$78$, and it is the differential that reports what $\theta_u$ does. It remains a **control** on
$\Delta_{\textit{some}}$ and not a second criterion, since the criterion is stipulated on the raw
shift. On this plane it is not a discriminating control: $\Delta_{\textit{no}}$ never exceeds
$0.1539$ in magnitude anywhere while $\Delta_{\textit{some}}$ reaches $-0.8109$, so wherever the
effect is large the differential simply *is* the raw shift. Both are mapped, and both verdicts are
reported below.

We call

$$\Lambda_{\mathrm{crit}}(\alpha,\theta_L)=\inf\{\Lambda>0:\ \text{leak}(\textit{no};\Lambda,\alpha,\theta_L)<\tfrac12\}
\tag{40}$$

the **override threshold**, the least lexical strength at which the entry for *no* still holds
against the prior. It is located by geometric bisection, and it is the diagonal the maps show. On the
truth region the settled log-density is $\ell_0/2$ and on the excluded region $(\ell_0-\Lambda)/2$
(Eq. 15), so the entry survives only while $\Lambda$ exceeds the log-prior gap between the two. For
Eq. (36) that gap is dominated by $-\alpha\log s$ at the truth region's best point
$s=\varsigma(-\theta_L)$, predicting

$$\Lambda_{\mathrm{crit}}\;\approx\;\lvert\log\varsigma(-\theta_L)\rvert\,\alpha ,
\tag{41}$$

This is linear in $\alpha$, with a slope fixed by the endpoint threshold alone. Since $\theta_L=\log(2n-1)$ is
itself derived from the predicate's granularity (Eq. A5), $\lvert\log\varsigma(-\theta_L)\rvert=\log 2n$
and the law reads $\Lambda_{\mathrm{crit}}\approx\alpha\log 2n$: the lexicon must be stronger, by a
logarithm of the set size, to hold against a confident prior over a finer predicate.

Eq. (41) is derived with the utility level severed, and there it holds: at $\alpha=1024$ the measured
slope is $1.3675$, $2.1572$, $3.0435$ against predictions of $1.3133$, $2.1269$, $3.0486$ for $\theta_L=1,2,3$, giving ratios $1.041$, $1.014$ and $0.998$. So the diagonal is the prior gap and nothing
else.

**Restoring the utility level raises the threshold.** The measured slopes at $\theta_u=1$ are
$1.6122$, $2.4839$, $3.5480$: the lexicon needs $15$–$18\%$ *more* strength to hold, and at
$\theta_u^\ast=-28.4375$ it needs $36$–$45\%$ more. The utility level reinforces the prior against
the entry.

Which mechanism does that matters, because the same one runs through Text cell 4 Part C. It is not
$\varphi_u$ regressing toward $\mu_u$: holding $\theta_u=1$ and moving
$\mu_u$ over a hundredfold range in magnitude and both signs changes the slope by under $0.1\%$, and
the effect is identical at $\theta_u$ and $-\theta_u$ (e.g. $1.6122$ against $1.6114$) and saturates
at $+45.4\%$ as $\lvert\theta_u\rvert$ grows. Those are the signatures of the $\theta_u^{2}$ term.
What is happening is the amplification Part C measures: raising
$\lvert\theta_u\rvert$ doubles the $\operatorname{span}B$ component of $\ell_0-\varphi_L$, and under
$\mathrm{Beta}(\alpha,1)$ that component is dominated by the prior's tilt, so amplifying it pushes
the prior harder against the entry. Both claims are checked in the cell rather than asserted.

---

**What the sweep is for.** $\Lambda$ and $\ell_0$ are both additive in the same log-density
(Eq. 15), so they compete directly, and a sufficiently confident prior overrides the lexical
entry outright, so that the model, told *no*, comes to believe $s\approx1$ with certainty. This is the
cost of the soft mask: a hard mask makes semantics inviolable by construction, and
finite $\Lambda$ buys differentiability and the $\Lambda\to\infty$ limit in exchange for making
the lexicon **defeasible**.

**Method.** With $\varphi_L$ clamped the model is linear-Gaussian and Eqs. (15)–(16) are exact
(Sec. 8.6), so the plane is evaluated in closed form, one small linear solve per cell rather
than thousands of Euler steps. The shortcut is verified rather than assumed: a random sample of
cells is re-run through the full `infer` dynamics and the worst disagreement reported.

**Reading the maps.** Rows are $\alpha$ (upward), columns are $\Lambda$ (rightward), both on
powers of two. Leak and $\mathbb E[s]$ are magnitudes on a fixed $0$–$1$ scale and take single-hue
sequential ramps; the shift is signed with a meaningful zero and takes a diverging pair with
limits symmetric about it, so that the neutral midpoint reads as "no difference from the literal
listener".

The third figure's last panel maps the differential (39). Under the first of Part C's two conditions
an implicature appears as $q$-mass moved *away* from the *all* region **for *some* specifically**, and
the differential is the control on that "specifically". It is a control and not a proof: the common
component need not be identical across utterances, and the criterion it serves is a stipulation of
Part C rather than a result.

**Across the plane the differential is negative almost everywhere.** $95$ of the $121$ cells are
negative; $65$ fall below $-10^{-3}$ and
$50$ below $-10^{-2}$; the most negative is $-0.8109$ (at $\alpha=64$, $\Lambda=2048$) against a
maximum of $+0.1279$ (at $\alpha=1$, $\Lambda=128$). The raw *some* shift is negative in $82$ cells
and reaches the same $-0.8109$. The differential is not flat against zero, and Part C's first
condition is met over most of the plane.

The geography is orderly, and it is what makes the numbers readable.

- **The $26$ positive cells all sit at $\alpha\le8$**, under diffuse priors at or near uniform.
  This is the region Part D samples with its *flat* and $\mathrm{Beta}(3,1)$ rows.
  $\Delta_{\textit{some}}$ is positive over a wider region still, $39$ cells reaching $\alpha=32$,
  and there it is the tempering that carries it: little prior mass sits on the *all* region for
  the utility level to move either way, so the flattening dominates the difference.

- **The $30$ cells with $\lvert\text{differential}\rvert<10^{-3}$ are mostly override cells**: $28$
  of them have $\Lambda\le\alpha$, where Eq. (41) says the prior has already beaten the entry and
  every utterance settles on the same belief. Flatness there is saturation, not the absence of an
  effect, which is the same point Part D makes for the delta-like row: the headroom has to be read
  before the number.

- **The effect is largest in the band $16\le\alpha\le128$ with $\Lambda$ several times $\alpha$**:
  the prior sharp enough to concentrate on the *all* region, the lexicon strong enough to hold
  against it. Part D's delta-like configuration, $\alpha=64$ and $\Lambda=512$, is a cell of this
  plane and sits in that band, reading $\Delta_{\textit{some}}=-0.3488$ with
  $\Delta_{\textit{no}}=0.0000$, so there the differential and the raw shift coincide to four
  decimals and the two statistics agree. That $-0.3488$ is also the number Part D's table carries,
  which is the check that the two cells measure one quantity.

So Part C's first condition is met over most of this plane on either statistic, and the choice
between the two is not load-bearing.

---

### <a id="tc6-conjunction" name="tc6-conjunction"></a>Where both of Part C's conditions hold

The second condition, $q_H(\text{all}\mid\textit{some})<\tfrac12$, is a statement about the
settled belief rather than about a difference, so it does not appear in the maps. Measured across
the same plane it holds in $55$ of the $121$ cells, and **both conditions hold together in $22$**.

The two run against each other in $\alpha$, which is what makes the region a band. Each condition
holds above a floor in $\Lambda$ and keeps holding above it. The first condition's floor **falls**
as the prior sharpens, from $\Lambda\ge2048$ at $\alpha=1$ to $\Lambda\ge128$ at $\alpha=8$ and
$\Lambda\ge2$ by $\alpha=64$: the more prior mass sits on the *all* region, the less lexical
strength the utility level needs in order to take some of it away. The second condition's floor
**rises**, from $\Lambda\ge2$ up to $\alpha=8$ to $\Lambda\ge128$ at $\alpha=16$,
$\Lambda\ge1024$ at $\alpha=64$ and $\Lambda\ge2048$ at $\alpha=128$, and past $\alpha=256$ it is
not reachable on this grid at all: the sharper the prior, the more strength the entry needs before
what is left below the *all* boundary is a majority. Both hold where the higher floor is still on
the plane, a band running from $(\alpha,\Lambda)=(8,128)$ to $(128,2048)$.

**Part D's rows sit outside that band, and the delta-like row sits just beside it.** Part D holds
$\Lambda=8$ for four of its priors, all diffuse enough that the first condition's floor sits far
above it; that floor drops below $\Lambda=8$ only at $\alpha\ge32$. The delta-like row raises
$\Lambda$ to $512$ at $\alpha=64$, which clears the first condition's floor of $2$ but not the
second's of $1024$, and it is the only one of the five to clear the first. Holding that prior and
raising $\Lambda$ alone carries $q_H$ from $0.6080$ at $\Lambda=512$ to $0.3780$ at $1024$ and
$0.1459$ at $2048$, the shift deepening from $-0.3488$ to $-0.5789$ and $-0.8109$. What the row
lacks is lexical strength, not prior concentration.

So the conjunction is within this architecture's reach; the five priors of Part D do not reach it,
and the delta-like row misses it by a factor of two in $\Lambda$. A configuration that meets both
was verified against the full `infer` dynamics rather than the closed form alone: at $\alpha=64$,
$\Lambda=2048$ the settled belief agrees with Eq. (15) to four decimals at
$P(\text{all-region})=0.1459$.

````python
# === Code Cell 4: the Lambda x alpha sweep. EVALUATION ===
#
# Two fixed quantities compete additively inside the same log-density (Eq. 15): the
# lexical field enters as -Lambda*chi_y, and the base world prior as ell_0. A prior
# concentrated near s = 1 therefore does not merely bias the belief; past some
# concentration it OVERRIDES the lexical entry outright. This cell maps that
# competition over the (alpha, Lambda) plane.
#
# The prior family is Beta(alpha, 1), which concentrates at s = 1 as alpha grows. A
# literal delta at s = 1 is not representable: s lies in the OPEN interval (0, 1),
# zeta = logit(1) is off the grid, and ell_0 = log p_0 would be -infinity almost
# everywhere, which makes the Gaussian residual r_S infinite. Beta(alpha, 1) is the
# limit family that approaches it.
#
# METHOD. With phi_L clamped the model is linear-Gaussian and Eqs. (15)-(16) are
# exact (Sec. 8.6), so the sweep is evaluated in closed form, a small linear solve
# per cell instead of some thousands of Euler steps. That shortcut is verified, not
# assumed: `verify_samples` cells are re-run through the full `infer` dynamics and
# the worst disagreement is reported.

import math
import numpy as np
import torch


def lambda_alpha_sweep(
    net,
    *,
    alphas=None,
    lambdas=None,
    verify_samples=6,
    dt=0.02,
    verbose=True,
):
    """Sweep prior concentration against lexical strength.

    Returns a dict of (len(alphas), len(lambdas)) arrays, rows indexed by alpha and
    columns by Lambda, matching the heat-map layout.

    The shift is Eq. (37): P(all; full) - P(all; q_lit), the same baseline Code Cell 2
    uses, so the values here and the ones Text cell 4 Part D tabulates are the same
    quantity. `literal_fixed_point` returns ell_0 - phi_L, the prior restricted by the
    entry and nothing else; the theta_u = 0 control is a DIFFERENT field, tempered by
    one half (Text cell 4 Part C), and is not what the shift is taken against.
    """
    # Equal counts make each map square, and the offset ranges put Lambda = 2*alpha on
    # the main diagonal. That diagonal is NOT the override threshold: Eq. (41) puts
    # Lambda_crit at |log sigma(-theta_L)| * alpha = log(2n) * alpha = 2.996 * alpha at
    # n = 10, measured at 3.04-3.18 * alpha with the utility level severed and
    # 3.50-3.55 * alpha at theta_u = 1. So the main diagonal runs BELOW the threshold
    # and every cell on it has the entry overridden (leak("no") = 1.00 at alpha >= 16).
    # The threshold is the parallel diagonal about 3.5x higher; `override_threshold`
    # locates it.
    alphas = [float(a) for a in (alphas if alphas is not None else [2.0 ** k for k in range(11)])]
    lambdas = [float(l) for l in (lambdas if lambdas is not None else [2.0 ** k for k in range(1, 12)])]

    shape = (len(alphas), len(lambdas))
    fields = ["leak", "expected_s", "shift"]
    grids = {f: {y: np.full(shape, np.nan) for y in UTTERANCES} for f in fields}
    grids["discriminability"] = np.full(shape, np.nan)

    proportion = logistic(net.zeta)

    for i, alpha in enumerate(alphas):
        prior = beta_world_prior(alpha, 1.0)
        for j, strength in enumerate(lambdas):
            probe = net.respawn(base_prior=prior, lexical_strength=strength)
            upper = (probe.zeta > probe.theta_L).to(probe.dtype)
            means = []
            for utterance in UTTERANCES:
                excluded = exclusion_indicator(
                    utterance, probe.zeta, probe.theta_L,
                    sharpness=probe.exclusion_sharpness,
                )
                full = probe.read_out(probe.closed_form_fixed_point(utterance)[0])
                base = probe.read_out(probe.literal_fixed_point(utterance))
                mean = float((probe.weights * full * proportion).sum())
                means.append(mean)
                grids["expected_s"][utterance][i, j] = mean
                grids["leak"][utterance][i, j] = float(
                    (probe.weights * full * excluded).sum()
                )
                grids["shift"][utterance][i, j] = float(
                    (probe.weights * full * upper).sum()
                    - (probe.weights * base * upper).sum()
                )
            grids["discriminability"][i, j] = max(means) - min(means)

    grids["alphas"] = alphas
    grids["lambdas"] = lambdas

    # === verify the closed-form shortcut against the full dynamics ===
    rng = np.random.default_rng(0)
    worst = 0.0
    checked = []
    for _ in range(verify_samples):
        i = int(rng.integers(len(alphas)))
        j = int(rng.integers(len(lambdas)))
        probe = net.respawn(
            base_prior=beta_world_prior(alphas[i], 1.0), lexical_strength=lambdas[j]
        )
        utterance = UTTERANCES[int(rng.integers(len(UTTERANCES)))]
        settled = probe.infer(utterance, dt=dt, max_steps=200000)
        target = probe.closed_form_fixed_point(utterance)[0]
        worst = max(worst, float((settled["phi_S"] - target).abs().max()))
        checked.append((alphas[i], lambdas[j], utterance, settled["converged"]))
    grids["verification"] = {"worst_error": worst, "checked": checked}

    if verbose:
        print("LAMBDA x ALPHA SWEEP")
        print(f"  alpha  in [{alphas[0]:.0f}, {alphas[-1]:.0f}]  ({len(alphas)} values)")
        print(f"  Lambda in [{lambdas[0]:.0f}, {lambdas[-1]:.0f}]  ({len(lambdas)} values)")
        print(f"  evaluated in closed form (Eqs. 15-16, exact by Sec. 8.6)")
        print(f"  verification: {verify_samples} cells re-run through full dynamics, "
              f"worst |phi_S - phi_S*| = {worst:.2e}, "
              f"all converged = {all(c[3] for c in checked)}")
        print()

        # where does the lexicon lose? first Lambda at which "no" still leaks > 1/2
        print("  smallest Lambda holding the entry for \"no\" (leak < 0.5):")
        for i, alpha in enumerate(alphas):
            row = grids["leak"]["no"][i]
            ok = [lambdas[j] for j in range(len(lambdas)) if row[j] < 0.5]
            print(f"    alpha = {alpha:>6.0f}   " +
                  (f"Lambda >= {min(ok):.0f}" if ok else f"none up to {lambdas[-1]:.0f}"))
        print()

    return grids


# === heat maps =============================================================

def plot_lambda_alpha_sweep(grids):
    """Three figures. Rows are alpha (upward), columns are Lambda (rightward).

    Colour follows the job each statistic does. Leak and E[s] are magnitudes on a
    fixed 0-1 scale, so each takes a single-hue sequential ramp, light to dark.
    The shift is signed and its zero is meaningful, so it takes a diverging pair
    with a neutral midpoint and limits placed symmetrically about zero, since otherwise
    the midpoint colour would land somewhere other than "no effect".

    The shift is Eq. (37), measured against q_lit exactly as in Code Cell 2.
    """
    import matplotlib.pyplot as plt
    from matplotlib.colors import TwoSlopeNorm

    alphas, lambdas = grids["alphas"], grids["lambdas"]

    def draw(axis, values, *, cmap, norm=None, vmin=None, vmax=None, title=""):
        image = axis.imshow(
            values, origin="lower", aspect="equal", cmap=cmap,
            norm=norm, vmin=vmin, vmax=vmax, interpolation="nearest",
        )
        axis.set_xticks(range(len(lambdas)))
        axis.set_xticklabels([f"{l:.0f}" for l in lambdas], fontsize=6.5, rotation=90)
        axis.set_yticks(range(len(alphas)))
        axis.set_yticklabels([f"{a:.0f}" for a in alphas], fontsize=6.5)
        axis.set_title(title, fontsize=9)
        axis.set_xlabel("lexical strength $\\Lambda$", fontsize=8)
        for spine in axis.spines.values():          # recessive frame
            spine.set_color("0.8")
        axis.tick_params(length=2, color="0.8")
        return image

    # 1. leak: how much belief survives where the entry excludes
    figure, axes = plt.subplots(1, 3, figsize=(11.5, 4.9), sharey=True, layout="constrained")
    for axis, utterance in zip(axes, UTTERANCES):
        image = draw(axis, grids["leak"][utterance], cmap="Reds", vmin=0.0, vmax=1.0,
                     title=f'leak  |  "{utterance}"')
    axes[0].set_ylabel("prior concentration $\\alpha$", fontsize=8)
    figure.colorbar(image, ax=axes, location="bottom", fraction=0.055, pad=0.02,
                    label="q-mass over the excluded region")
    figure.suptitle(
        "Where the lexical entry holds. Dark = the entry is overridden by the prior.",
        fontsize=10)
    plt.show()

    # 2. E[s] and discriminability
    figure, axes = plt.subplots(1, 4, figsize=(14.5, 4.9), sharey=True, layout="constrained")
    for axis, utterance in zip(axes[:3], UTTERANCES):
        image = draw(axis, grids["expected_s"][utterance], cmap="Blues", vmin=0.0, vmax=1.0,
                     title=f'$\\mathbb{{E}}[s]$  |  "{utterance}"')
    figure.colorbar(image, ax=axes[:3], location="bottom", fraction=0.055, pad=0.02,
                    label="$\\mathbb{E}[s]$")
    separation = draw(axes[3], grids["discriminability"], cmap="Greens", vmin=0.0, vmax=1.0,
                      title="$\\max_y\\mathbb{E}[s]-\\min_y\\mathbb{E}[s]$")
    figure.colorbar(separation, ax=axes[3], location="bottom", fraction=0.055, pad=0.02,
                    label="spread")
    axes[0].set_ylabel("prior concentration $\\alpha$", fontsize=8)
    figure.suptitle(
        "What the model believes. Rightmost: how far apart the three utterances are: "
        "pale means the model has stopped distinguishing them.", fontsize=10)
    plt.show()

    # 3. implicature shift: signed, so diverging about zero
    figure, axes = plt.subplots(1, 4, figsize=(14.5, 4.9), sharey=True, layout="constrained")
    stack = np.concatenate([grids["shift"][y].ravel() for y in UTTERANCES])
    limit = float(np.nanmax(np.abs(stack))) or 1.0
    norm = TwoSlopeNorm(vmin=-limit, vcenter=0.0, vmax=limit)
    for axis, utterance in zip(axes[:3], UTTERANCES):
        image = draw(axis, grids["shift"][utterance], cmap="RdBu_r", norm=norm,
                     title=f'shift  |  "{utterance}"')
    figure.colorbar(image, ax=axes[:3], location="bottom", fraction=0.055, pad=0.02,
                    label="shift in $P(\\mathrm{all})$")

    differential = grids["shift"]["some"] - grids["shift"]["no"]
    d_limit = float(np.nanmax(np.abs(differential))) or 1.0
    diff_image = draw(axes[3], differential, cmap="PuOr_r",
                      norm=TwoSlopeNorm(vmin=-d_limit, vcenter=0.0, vmax=d_limit),
                      title="shift(some) $-$ shift(no)")
    figure.colorbar(diff_image, ax=axes[3], location="bottom", fraction=0.055, pad=0.02,
                    label="difference")
    axes[0].set_ylabel("prior concentration $\\alpha$", fontsize=8)
    figure.suptitle(
        "Implicature probe across the plane. Blue = q-mass moved AWAY from the \"all\" "
        "region (the implicature direction). Rightmost is the differential (39), a "
        "control on the \"some\" panel; Text cell 6 reports where the two coincide.",
        fontsize=10)
    plt.show()


# === the override threshold =================================================

def override_threshold(net, *, alpha=1024.0, boundaries=(1.0, 2.0, 3.0), verbose=True):
    """Locate the diagonal the heat maps show, and check what sets it.

    On the truth region the settled log-density is ell_0/2; on the excluded region it
    is (ell_0 - Lambda)/2 (Eq. 15). The entry therefore survives only while Lambda
    exceeds the log-prior gap between the two regions. For Beta(alpha, 1) that gap is
    dominated by -alpha*log s at the truth region's best point s = sigma(-theta_L),
    predicting

        Lambda_crit  ~  |log sigma(-theta_L)| * alpha,                       (41)

    linear in alpha with a slope fixed by the endpoint threshold alone.

    Eq. (41) is derived with the utility level severed, and holds to a few percent
    there. Restoring it RAISES the required Lambda: the lexicon needs MORE strength to
    hold against the prior, not less. The mechanism is the one Text cell 4 Part C
    measures: raising |theta_u| amplifies the span(B) component of (ell_0 - phi_L),
    and under Beta(alpha, 1) that component is dominated by the prior's tilt, so
    amplifying it reinforces the prior against the entry. Two things follow, and both
    are checked below rather than asserted: the effect is quadratic in theta_u, so it
    is the SAME for either sign and saturates as |theta_u| grows; and it is
    independent of mu_u, which rules out reading it as regression toward mu_u.
    """
    def critical(boundary, theta, mu=None):
        def leak(strength):
            overrides = dict(
                base_prior=beta_world_prior(alpha, 1.0),
                lexical_strength=strength,
                theta_L=boundary,
            )
            if mu is not None:
                overrides["mu_u"] = torch.tensor(mu, dtype=net.dtype)
            probe = net.respawn(**overrides)
            belief = probe.read_out(
                probe.closed_form_fixed_point("no", theta_u=theta)[0]
            )
            excluded = exclusion_indicator("no", probe.zeta, boundary)
            return float((probe.weights * belief * excluded).sum())

        low, high = 1e-3, 1e7
        for _ in range(90):                       # geometric bisection
            middle = math.sqrt(low * high)
            if leak(middle) < 0.5:
                high = middle
            else:
                low = middle
        return math.sqrt(low * high)

    star = theta_u_stationary_point(net)
    rows = []
    for boundary in boundaries:
        predicted = -math.log(float(torch.sigmoid(torch.tensor(-boundary))))
        rows.append({
            "boundary": boundary,
            "predicted_slope": predicted,
            "slope_with_u": critical(boundary, 1.0) / alpha,
            "slope_without_u": critical(boundary, 0.0) / alpha,
            "slope_negated": critical(boundary, -1.0) / alpha,
            "slope_at_star": critical(boundary, star) / alpha,
        })

    if verbose:
        print("OVERRIDE THRESHOLD (the diagonal in the leak map)")
        print(f"  Lambda at which leak(\"no\") = 1/2, divided by alpha, at alpha = {alpha:.0f}")
        print()
        print(f"  {'theta_L':>8}{'Eq. (41)':>11}{'theta_u=0':>11}{'(41) ratio':>10}"
              f"{'theta_u=1':>11}{'theta_u=-1':>12}{'theta_u*':>10}{'change at 1':>13}")
        for row in rows:
            print(f"  {row['boundary']:>8.0f}{row['predicted_slope']:>11.4f}"
                  f"{row['slope_without_u']:>11.4f}"
                  f"{row['slope_without_u'] / row['predicted_slope']:>10.3f}"
                  f"{row['slope_with_u']:>11.4f}{row['slope_negated']:>12.4f}"
                  f"{row['slope_at_star']:>10.4f}"
                  f"{row['slope_with_u'] / row['slope_without_u'] - 1.0:>+12.1%}")
        print()
        print("  With the utility level severed, Eq. (41) is accurate to a few percent,")
        print("  so the diagonal is the prior gap and nothing else. Restoring the level")
        print("  RAISES the required Lambda, by 15-18% at theta_u = 1 and by 36-45% at")
        print(f"  theta_u* = {star:.4f}. The utility level reinforces the prior against the")
        print("  lexical entry; it does not oppose it.")
        print()
        print("  Two checks that this is amplification (Text cell 4 Part C) and not")
        print("  regression toward mu_u:")
        print()
        baseline = critical(boundaries[0], 0.0) / alpha
        print(f"    (a) sign and magnitude, at theta_L = {boundaries[0]:.0f}:")
        print(f"        {'theta_u':>10}{'slope':>10}{'vs theta_u=0':>15}")
        for theta in (1.0, -1.0, 3.0, -3.0, 10.0, 100.0, 10000.0):
            value = critical(boundaries[0], theta) / alpha
            print(f"        {theta:>10.0f}{value:>10.4f}{value / baseline - 1.0:>+15.1%}")
        print("        +theta and -theta agree; the effect saturates. Both are")
        print("        signatures of the theta_u^2 term, not of a mu_u-carried one.")
        print()
        print(f"    (b) mu_u-independence, at theta_L = {boundaries[0]:.0f}, theta_u = 1:")
        print(f"        {'mu_u':>16}{'slope':>10}{'vs theta_u=0':>15}")
        for mu in ([1.0, 1.0], [0.05, 0.05], [5.0, 5.0], [-1.0, -1.0], [1.0, -1.0]):
            value = critical(boundaries[0], 1.0, mu) / alpha
            print(f"        {str(mu):>16}{value:>10.4f}{value / baseline - 1.0:>+15.1%}")
        print("        A hundredfold change in |mu_u|, and both signs, move the slope by")
        print("        less than a tenth of a percent. mu_u is not what does this.")
        print()

    return rows


sweep_grids = lambda_alpha_sweep(evaluation_network)
override_threshold(evaluation_network)
plot_lambda_alpha_sweep(sweep_grids)
````

## <a id="appa" name="appa"></a>Appendix A: the utterance level and the threshold $\theta_L$

**Realizing $E_y$ on the open scale: the threshold $\theta_L$.** $\operatorname{logit}$ is asymptotic
at $s=0$ and $s=1$, so it is a bijection only on the **open** interval; $s$ therefore lives on an open
scale, and the two endpoints at which Eq. (5) anchors *no* and *all* have no $\zeta$ coordinate.
Realizing $E_y$ on the grid consequently requires a threshold $\theta_L>0$, under which

$$E_{\textit{no}}\mapsto\{\zeta>-\theta_L\},\qquad
E_{\textit{some}}\mapsto\{\zeta\le-\theta_L\},\qquad
E_{\textit{all}}\mapsto\{\zeta<\theta_L\}.
\tag{A1}$$

**$g_y$, and $\theta_L$ as its generative weight.** $g_y$ maps lexical content to word form. Write
$A\in\mathbb R^{|\mathcal Y|\times K}$ for a fixed profile matrix, one projection per word-form unit,
as $b$ is one projection at the $S$ level (§3), and let $W$ be the quadrature measure of Eq. (3). Then

$$\hat\varphi_y=g_y(\varphi_L)=\theta_L\,A W\varphi_L.
\tag{A2}$$

This is $\Theta h(\varphi)$ with $\Theta=\theta_L A$ and $h=\mathrm{id}$: the form of $g_S$ (Eq. 10)
with the direction reversed, $g_S$ expanding $\mathbb R^m$ into a field where $g_y$ contracts a field
onto the $|\mathcal Y|$ word-form units. $A$ is fixed and $\theta_L$ is the whole of the parameter,
exactly as $B$ is fixed and $\theta_u$ is the whole of the parameter in $g_S$.

The three fields $\Lambda\chi_y(\theta_L)$ of Eq. (A1) are linearly independent under $W$ (their Gram
matrix has rank $3$), so an $A$ exists for which $g_y$ returns the observed word form exactly at each
clamp. Under it Eq. (8) is a *consequence*, as §4 states, rather than a stipulation.

**The free energy this appendix differentiates.** Eq. (13) drops the $y$ level on the grounds that
$r_y\equiv0$ contributes nothing. That is true of the term's *value* and not of its *gradient*: a term
pinned at zero by a clamp still has a derivative in the parameter that would move it off the clamp.
Read literally, Eq. (13) gives $\partial\mathcal F/\partial\theta_L=0$ identically, which is the right number at this phase, for the wrong reason, and with no direction attached to it. What is differentiated
below is Eq. (13) with the $y$ level restored:

$$\mathcal F=-\frac{\lVert r_y\rVert^{2}}{2\sigma_y}
-\frac{\lVert r_L\rVert^{2}}{2\sigma_L}
-\frac{\lVert r_S\rVert^{2}}{2\sigma_S}
-\frac{\lVert r_u\rVert^{2}}{2\sigma_u}
-\tfrac12\big(\lvert\mathcal Y\rvert\log\sigma_y+K\log\sigma_L+K\log\sigma_S+m\log\sigma_u\big)+C,
\qquad r_y=\varphi_y-\theta_L\,AW\varphi_L .
\tag{A3}$$

Here $\lVert r_y\rVert^{2}$ is the ordinary squared norm on $\mathbb R^{\lvert\mathcal Y\rvert}$
(Eq. 4), as $\lVert r_u\rVert^{2}$ is on $\mathbb R^m$; the two field norms remain those of Eq. (2).
The $y$ level contributes one $\log\sigma$ per component on the same rule as the other three, hence
$\lvert\mathcal Y\rvert$ and not $K$.

**Eq. (A3) is not a different model, which is why Text cell 3 may state Eq. (13) as it does.** $r_y$
depends on $\varphi_y$, $\theta_L$ and $\varphi_L$, and on nothing below $\varphi_L$. So the restored
term is constant in $x=(\varphi_S,\varphi_u)$, leaving §8's concavity,
uniqueness and convergence results untouched together with every closed form of §6, and at the
clamp its value is $0$, so
$\mathcal F$ takes the same numbers everywhere the implementation evaluates it. The truncation loses
exactly one thing, and it is the thing this appendix needs: the direction $\theta_L$ would move in.

**Its gradient, and its flow.** Differentiating Eq. (A3) in the gain of Eq. (A2), with
$\partial r_y/\partial\theta_L=-AW\varphi_L$ and $\varepsilon_y=r_y/\sigma_y$ on the rule of
Eq. (14),

$$\tau_{\theta_L}\dot\theta_L=\frac{\partial\mathcal F}{\partial\theta_L}
=\frac{1}{\sigma_y}\,r_y\cdot AW\varphi_L
=\varepsilon_y\cdot AW\varphi_L\,,
\tag{A4}$$

The right-hand side is the error below times the activity above. This is Bogacz Eq. (25), and a product of pre- and
post-synaptic activity, so Eq. (A4) is local on the same terms as Eq. (20); $\theta_L$ takes a time
constant of its own at the slow end of $\tau_\varepsilon\ll\tau_\varphi\ll\tau_\theta$, beside
$\theta_u$. Nothing here requires a normalization across the word-form units, which would not be
local. Note that Eq. (A4) is the partial through the gain alone; $\theta_L$'s second appearance, in
the threshold that builds $\chi_y$, is taken up under *$\theta_L$ enters twice* below.

Eq. (A4) vanishes identically while $\varepsilon_y\equiv0$ (Eq. 8). **$\theta_L$ is therefore fixed
because $\varphi_L$ is clamped, not by stipulation**, and unclamping $\varphi_L$ sets it running by
the same route that lets an auditory model stack beneath (§4). Until then it is carried as a fixed
quantity with a variational quantity's status.

**$\theta_L$ enters twice.** It is the gain of Eq. (A2) and, through Eq. (A1), the threshold building
$\chi_y$ and so the clamp itself, which makes it the only parameter of the model appearing in both the
generative map and the observation model. That is what identifies it: $\theta_L\,AWX(\theta_L)=I$, with $X$ the matrix of clamped fields $\Lambda\chi_y$ of Eq. (4), is a
fixed-point condition in $\theta_L$, and any departure from it leaves an $\varepsilon_y$ for Eq. (A4)
to act on.

**$n$: the denotation of $\theta_L$.** A threshold at $\varsigma(-\theta_L)$ commits the predicate to
a finest distinguishable quantity. Declaring

$$\varsigma(-\theta_L)=\frac{1}{2n}
\qquad\Longleftrightarrow\qquad
\theta_L=\log(2n-1),
\qquad n=\frac{e^{\theta_L}+1}{2},
\tag{A5}$$

gives $\theta_L$ its denotation: **$n$ is the number of atoms the predicate resolves**. It also
settles how a threshold can serve as the gain of Eq. (A2): $e^{\theta_L}=2n-1$, so $\theta_L$ is a
*log*-gain, and since $\zeta=\operatorname{logit}s$ is a log-odds coordinate, translating by
$\theta_L$ in $\zeta$ is multiplying by $e^{\theta_L}$ in odds. $\theta_L$ is multiplicative on the
odds scale and appears as a location in $\zeta$ only because $\zeta$ is logarithmic. The scale then
carries $n+1$ quantities, the counts $k=0,\dots,n$ at proportions $s_k=k/n$, with resolution $1/n$.

The **Voronoi cell** of a count follows: the band of width $1/n$ about it, the proportions closer to
that count than to any other. Eq. (A5) places the threshold at exactly half a cell, so Eq. (A1)
excludes precisely the states lying outside the cell of the count each entry anchors,
$s_0$ for *no* and $s_n$ for *all*, symmetrically at both ends. At the canonical $n=10$, $\theta_L=\log19=2.9444$
and $\varsigma(-\theta_L)=0.05$.

*$\theta_L$ is semantic.* Three consequences follow.

First, this design predicts that **exact zero is unrepresentable** by natural language. There is no resolution limit
without a finite set to resolve, and a predicate over $n$ atoms cannot distinguish "none" from "fewer
than half an atom". *No* means *below the resolution limit*, which is what a finite perceptual system
can mean by it. Eq. (5) states the semantics; Eq. (A1) states what a system of finite granularity does
with it. I invite the reader to have their own judgment on whether this prediction is a desideratum, as one might
intuitively think zero is representable. However, it is useful to distinguish representation from imagination per induction.

Second, $\theta_L$ grows like $\log 2n$, so the grid must leave $E_{\textit{no}}$ some unexcluded
node: the implementation requires $\theta_L<Z$ for grid half-width $Z$, i.e.
$n<(e^{Z}+1)/2=202.21$, capping $n$ at $202$ when $Z=6$ ($\theta_L=5.99894$; $n=203$ gives
$6.00389$ and is rejected). Sharpness is bounded by the semantics rather than free.

*The gradable case: $\theta_L$ from a just-noticeable difference.* A gradable predicate supplies no
$n$, so $\theta_L$ needs a different source. It does not need a different *formula*. Read Eq. (A5)
back to what it asserts: the threshold sits at half the scale's resolution, and for a counted
predicate the resolution happens to be $1/n$ because $n$ equally spaced atoms is what generates it.
Write $\delta$ for the resolution directly, the just-noticeable
difference as a fraction of the scale, and the same declaration reads

$$\varsigma(-\theta_L)=\frac{\delta}{2}
\qquad\Longleftrightarrow\qquad
\theta_L=\log\!\Big(\frac{2}{\delta}-1\Big),
\qquad \delta=\frac{2}{e^{\theta_L}+1}.
\tag{A6}$$

Eq. (A5) is the case $\delta=1/n$, and the two are the same statement: $n=1/\delta$, a count where
one exists and a resolution reciprocal where it does not. Nothing in §§4–8 changes, since $\theta_L$
enters only through $\chi_y$ and Eq. (A2) and neither asks where it came from. The grid cap transfers
with it: $\theta_L<Z$ becomes $\delta>2/(e^{Z}+1)=0.004945$ at $Z=6$, so a JND finer than
about half a percent of the scale is unrepresentable on the default grid, which is the $n\le202$
bound restated.

**What Eq. (A6) does not settle, and this is the open part.** It fixes $\theta_L$ *given* $\delta$
and says nothing about what fixes $\delta$, and the candidates are not equivalent. A $\delta$
constant in $s$, a fixed absolute discrimination width, is what the counting case supplies, since
atoms are equally spaced in the proportion. A $\delta$ constant in $\zeta$ is the Weber-type
alternative, a fixed ratio in odds, and it is the one the coordinate of Eq. (1) is natural for:
$\zeta$ is a log-odds scale and $\theta_L$ is already a log-gain, so a constant-in-$\zeta$ JND would
make $\theta_L$ a JND count rather than a function of one. The two disagree, and they disagree most
near the endpoints, which is exactly where Eq. (A1) places the boundaries. Choosing between them is
an empirical question about the predicate, not something the architecture decides, and it is not
settled here. What Eq. (A6) buys is that the choice is now a choice about one scalar with a stated
meaning, rather than about whether the construction extends at all, which is where §9.2's
dense-scale extension reconnects.

## <a id="appb" name="appb"></a>Appendix B: the stationary point of $\theta_u$, its locality, and alternative spaces

**The premise §8.5 rests on is $\mu_u\neq0$.** The terminating
contribution to $g_S$ is $\theta_u B\mu_u$. At $\mu_u=0$, the contribution vanishes for every $\theta_u$. Thus, varying
$\theta_u$ cannot move the terminating prediction at all, the chain stops rather than terminates, whereby
the only remaining use for $\theta_u$ is to shift magnitude out of the priced factor $\varphi_u$ into
the unpriced $\theta_u$ along the exact degeneracy
$(\theta_u,\varphi_u)\mapsto(c\,\theta_u,\varphi_u/c)$, which leaves $g_S$ invariant.
$\tilde{\mathcal F}$ then has **no stationary point**; instead, it climbs monotonically to a finite asymptote
and $\lvert\theta_u\rvert\to\infty$.

**The reduced objective.** Eliminating $x$ analytically (legitimate by §8.6) collapses
$\tilde{\mathcal F}$ to a ratio of quadratics in the single scalar $\theta_u$:

$$\tilde{\mathcal F}(\theta_u)=\mathrm{const}-\frac{\lVert\theta_u\mu_u-c\rVert^{2}}
{2\big(\sigma_u\theta_u^{2}+S\big)},\qquad S=\sigma_L+\sigma_S,\qquad
c=B^{\mathsf T}W\big(\ell_0-\varphi_L\big),
\tag{B1}$$

$\mathrm{const}$ collecting what $\theta_u$ cannot reach. Stationarity is therefore the quadratic

$$\sigma_u\langle\mu_u,c\rangle\,\theta_u^{2}
+\big(S\lVert\mu_u\rVert^{2}-\sigma_u\lVert c\rVert^{2}\big)\,\theta_u
-S\langle\mu_u,c\rangle=0,
\tag{B2}$$

That is the condition for a single utterance. Here $c_y=B^{\mathsf T}W(\ell_0-\varphi_{L,y})$ is that utterance's target
projected onto the utility basis, one vector in $\mathbb R^m$ per entry, and
$\sum_y$ denotes a
**sum over the three utterances**, not the variance $\sigma_y$ of §3:

| | $c_y$ | $\langle\mu_u,c_y\rangle$ |
|---|---|---|
| *no* | $[-9.11909,\;-12.83178]$ | $-21.9509$ |
| *some* | $[+9.11909,\;-24.36744]$ | $-15.2484$ |
| *all* | $[+9.11909,\;-12.83178]$ | $-3.7127$ |
| **sum** | $[+9.11909,\;-50.03100]$ | $-40.9119$ |

For the flow of Eq. (20), which visits all three, the same quadratic governs with
$\lVert\mu_u\rVert^{2}\mapsto3\lVert\mu_u\rVert^{2}$, $c\mapsto\sum_y c_y$ and
$\lVert c\rVert^{2}\mapsto\sum_y\lVert c_y\rVert^{2}$. The three utterances share the denominator
$\sigma_u\theta_u^{2}+S$, so only their numerators combine:

$$\sum_y\lVert\theta_u\mu_u-c_y\rVert^{2}
=3\theta_u^{2}\lVert\mu_u\rVert^{2}
-2\theta_u\Big\langle\mu_u,\sum_y c_y\Big\rangle
+\sum_y\lVert c_y\rVert^{2}.
\tag{B3}$$

The individual $c_y$ reach $\theta_u$ only through the term linear in it, that is, only through their sum.

**The standing condition, stated exactly.** §3 requires $\mu_u\neq0$. What Eq. (B2) actually needs is
$\langle\mu_u,\sum_y c_y\rangle\neq0$, which is strictly stronger: $\mu_u=0$ violates it
automatically, but so would any nonzero $\mu_u$ orthogonal to $\sum_y c_y$, and there both roots
collapse onto $\theta_u=0$. At $\mu_u=\mathbf 1$ the quantity is $-40.9119$, well clear of the
degenerate case. The weights in that sum are moreover a **prior over utterances**: the flow averages
the three gradients equally, so the operative quantity is $3\,\mathbb E_{p(y)}[c_y]$ at uniform
$p(y)$, and a non-uniform exposure moves $\theta_u^\ast$ accordingly.

The roots of Eq. (B2) have product $-S/\sigma_u<0$, so whenever $\langle\mu_u,c\rangle\neq0$
there is **exactly one root of each sign**: the maximizer is the one with
$\operatorname{sign}\theta_u^\ast=\operatorname{sign}\langle\mu_u,c\rangle$ and the other is a
minimum, which ascent cannot cross. At $\mu_u=[1,1]$ this is a genuine interior maximum at
$\theta_u^\ast=-28.4375$, with $\partial\tilde{\mathcal F}/\partial\theta_u=5.5\times10^{-16}$ there
and a margin of $+0.2398$ over the asymptote; the flow reaches it from any $\tau_\theta$ started on
the same side of zero.

**The sign is not free to assume.** $\theta_u$ is an unconstrained real parameter: Bogacz's Eqs. (25)
and (56) supply a gradient and no domain, and his own $\theta^\ast=u/v_p$ takes whichever sign the
data give. Because $\ell_0$ sits in $g_L$ (Eq. 9), $c_y=B^{\mathsf T}W(\ell_0-\varphi_{L,y})$ carries $\ell_0$
with a positive sign against $\varphi_L$ with a negative one, which puts
$\sum_y\langle\mu_u,c_y\rangle$ at $-40.912$ and the maximizer at $-28.437$. The sign is a
consequence of that placement, not a choice: were the two signs reversed the maximizer would be
$+22.578$. A search bracketed on $\theta_u>0$ reports, in consequence, a
divergence that is an artefact of the bracket and not a property of the model.

This is **not** a departure from Bogacz. His Eq. (25) model diverges identically at $v_p=0$: with his
own Exercise 1 numbers ($u=2$, $\Sigma=1$) the optimum is $\theta^\ast=u/v_p$, so $v_p=3$ gives
$0.667$, $v_p=1$ gives $2.0$, and $v_p=0$ gives none. **$v_p\neq0$ is an unstated premise of his
construction**, whose worked example takes $v_p=3$, and §3's $\mu_u\neq0$ is the same premise stated.

**What pins $\theta_u$ here is not what pins $\theta$ there, and the difference runs in our favour.**
In Bogacz's scalar model $\theta v_p=u$ is solvable for any $u$ once $v_p\neq0$, so at $\theta^\ast$
*both residuals vanish* and $F=0$: it is **realizability** that identifies $\theta$. Our $\theta_u$ is
one scalar facing $c\in\mathbb R^m$, so $\theta_u\mu_u=c$ would need $c\parallel\mu_u$, a coincidence
of measure zero, and here $\cos(\mu_u,c)=-0.986,\,-0.414,\,-0.167$ for the three utterances. The
residuals stay far from zero at the optimum ($\lVert r_S\rVert=22.5,\,15.8,\,22.5$, and
$\tilde{\mathcal F}=-420.209$ rather than $0$). Promoting $\theta_u$ to the matrix of §8.4 would not
repair this: one $\theta_u$ is shared across utterances, so zeroing all three at once would require
$c_{\textit{no}}=c_{\textit{some}}=c_{\textit{all}}$, and those differ by up to $21.58$.

$\theta_u$ is identified regardless, because Eq. (B1) is a ratio of quadratics and not a residual to
be zeroed: it trades fit against the prior cost $\lVert\varphi_u-\mu_u\rVert^{2}$, and that trade has
an interior optimum whenever $\langle\mu_u,c\rangle\neq0$. The argument therefore has a **weaker premise and a wider reach** than Bogacz's, since it never asks the generative model to be able to fit the
situation level at all. What $\mu_u\neq0$ buys is not realizability but a nonzero
$\langle\mu_u,c\rangle$: a prior cost that $\theta_u$ has something to trade against.

We therefore claim **global** convergence for $(\varphi_S,\varphi_u)$ and, for $\theta_u$, convergence
of $\tilde{\mathcal F}$ together with convergence to a stationary point that is only guaranteed
**local**, since $\tilde{\mathcal F}$ is not concave. This is the status of the corresponding claim in
Bogacz, where inference is convex and plasticity is not.

### <a id="appb-locality" name="appb-locality"></a>Locality

Bogacz states two constraints in his Introduction:

> **Local computation.** A neuron performs computations only on the basis of the activity of its input
> neurons and synaptic weights associated with these inputs.
>
> **Local plasticity.** Synaptic plasticity is only based on the activity of pre-synaptic and
> post-synaptic neurons.

**The sum over $y$ is not a locality matter.** No equation the network integrates contains
$\sum_y c_y$. Eq. (20) is evaluated at one presented utterance, after the fast subsystem has
equilibrated, from the two local factors $\varphi_u$ and $\langle\varepsilon_S,b\rangle$; the sum is
an accumulation across successive presentations, and appears only in the closed-form analysis of
where the accumulated flow settles. The implementation batches the three for numerical convenience,
but the batching is not load-bearing: updating once per presentation instead, one local update three
times per cycle, reaches the same fixed point, the two trajectories agreeing to
$4\times10^{-3}$ at $\theta_u=-16.73$ and to $1.6\times10^{-4}$ at $\theta_u=-24.83$ as both approach
$\theta_u^\ast$. What the sum carries is an assumption about **exposure**, not about circuitry: the
ensemble must be visited within one $\tau_\theta$, at the frequencies $p(y)$ that weight it.

**Eq. (20) itself is the real question, and it has nothing to do with alternatives.** The forward
weights from the $u$ level to the $S$ level are $M=\theta_u B\in\mathbb R^{K\times m}$. Were $M$ free, as
Bogacz's $\Theta$ of Eq. (56) is, each entry would update as
$\dot M_{kj}\propto\varepsilon_{S,k}\varphi_{u,j}$: one error neuron times one state neuron, Hebbian,
and local without further comment. We have instead tied all $K\times m$ entries to a single scalar
times a fixed $B$, so moving $\theta_u$ requires pooling the local products across every one of them.
That is **weight sharing**: the update of a single parameter reads products formed at synapses on
several different neurons, which is what the constraint of local plasticity quoted above forbids.
The ranking therefore inverts against expectation: **Bogacz's full matrix is local, and it
is our scalar restriction that needs defending.**

**The commitment: $b$ is a single projection.** We take $b$ to be the fixed spatial profile of one connection, one projection whose terminal field on the $S$ population
has shape $b$ and which pools error back through that same fixed profile, and *not* $K$ independently
modifiable synapses sharing a gain. Under that reading $\langle\varepsilon_S,b\rangle$ is a single signal physically present at a
single synapse, $\varphi_u$ is that synapse's presynaptic activity, and Eq. (20),
$\tau_\theta\dot\theta_u=\varphi_u\langle\varepsilon_S,b\rangle$, is a product of pre- and
post-synaptic activity: Hebbian, and local in Bogacz's sense.

Nothing extra is assumed for plasticity that inference did not already require. Eq. (19) reads
$\tau_\varphi\dot\varphi_u=-\varepsilon_u+\theta_u\langle\varepsilon_S,b\rangle$, so the $\varphi_u$
neuron already receives the pooled $\langle\varepsilon_S,b\rangle$ as dendritic input: **local
computation** already obliges that signal to exist as a physical quantity, and local plasticity then
gets it for free. Bogacz reasons the same way about his own $\theta$, which likewise sits on two
connections at once:

> "the rule also satisfies the constraint of local plasticity … as $\phi$ fully determines $h(\phi)$,
> so the change in weight is fully determined by the activity of pre-synaptic and post-synaptic
> neurons."

Our $b$ extends that argument from two connections to $K$, on the same grounds.

**The price.** This is what forecloses the matrix case of §8.4 as a *physical* generalization: the
scalar is what makes the projection single, and the single projection is what makes Eq. (20) local.

**What fails at $m>1$.** At $m=1$, Eq. (20) reads $\varphi_u\langle\varepsilon_S,b\rangle$: one
presynaptic activity, one pooled error signal, one synapse. At $m>1$ the same gradient is

$$\frac{\partial\mathcal F}{\partial\theta_u}
=\sum_{j=1}^{m}\varphi_{u,j}\big\langle\varepsilon_S,b_j\big\rangle.
\tag{B4}$$

Every summand is itself a legitimate Hebbian product, formed where column $j$ meets the error field.
**The violation is the sum.** $\theta_u$ is one scalar gaining all $m$ columns, so moving it requires
the whole of Eq. (B4), and the $j$-th synapse holds $\varphi_{u,j}$ and
$\langle\varepsilon_S,b_j\rangle$ and nothing else: the remaining $m-1$ terms are formed at synapses
on other neurons. A synapse cannot read them, which is what the constraint of local plasticity
forbids.

The single-projection commitment does not reach this case. It licenses pooling across the $K$
terminals of *one* projection, and it does so because Eq. (19) already obliges
$\langle\varepsilon_S,b\rangle$ to exist as one signal at one site. Nothing obliges the sum over $j$
to exist anywhere, so at $m>1$ the pooling is across $m$ sites and the argument gives out. At $m=1$
the sum has one term and the question does not arise, which is why Eq. (20) is stated there.
Appendix E forms the sum in a neuron rather than in the rule, and reports what that costs.

### <a id="appb-alternative-spaces" name="appb-alternative-spaces"></a>Alternative spaces

§3 fixes $E_y$ by the utterance entry alone, with no reference to an alternatives space. Since
$\theta_u^\ast$ depends on $\sum_y c_y$, the two appear to be in conflict. They are not, but the reason commits the model to a position, so it is stated rather than assumed.

**At inference there is no conflict whatever.** With $\theta_u$ held fixed, Eqs. (15)–(16) at
utterance $y$ involve $c_y$ and nothing else: reconstructing $\varphi_u^\ast$ from $c_y$ alone
reproduces the full fixed point to at worst $2.2\times10^{-16}$. One utterance in, one belief out; no
other entry is consulted, and §3's claim holds exactly.

**At learning, $\theta_u$ does see the utterance set**, through exposure and substantially:

| exposure | $\theta_u^\ast$ |
|---|---|
| *no* | $-11.2844$ |
| *some* | $-44.1766$ |
| *all* | $-65.7004$ |
| *no*, *some* | $-24.7250$ |
| *no*, *all* | $-19.1053$ |
| *some*, *all* | $-48.3901$ |
| all three | $-28.4375$ |

A factor of nearly six separates the single-utterance optima.

**But this is not an alternatives set in the sense the neo-Gricean literature uses.** Grice (1975)
states maxims and says nothing about processing architecture; it is the neo-Gricean formalizations
that make alternatives a *within-trial* comparison: a lexical scale supplies the competitors
(Horn, 1972), and what the speaker declined to say is used, on that occasion, to strengthen what was
said (Gazdar, 1979; Levinson, 2000; Sauerland, 2004). The strengthening step is not uniform across
those accounts, and nothing here needs it to be: Gazdar (1979) and Sauerland (2004) derive in the
first instance the epistemically weak inference that the speaker was not in a position to assert the
stronger alternative, and reach its negation only through a further assumption about what the speaker
knows, where Levinson (2000) takes the negation itself as the default. What they share is the
*occasion*: on all of them a listener hearing *some* consults what *all* would have conveyed
**on this trial**. What
$\theta_u$ holds is a *cross-trial average*, one shared weight, shaped by the frequencies at which
the entries have been met and consulted identically whatever is heard. No two entries are compared
at any moment, and under the single-projection commitment above there is no anatomy in which such a
comparison could take place: one synapse, one scalar, one pooled error signal.

The position this commits us to is that any implicature reaching the belief through $\theta_u^\ast$
is **conventionalized rather than computed**, lexicalized into a weight by exposure and not derived
online from a reasoning step. That is a substantive claim. What the neo-Gricean (Horn, 1972;
Sauerland, 2004; Geurts, 2010), grammatical (Chierchia, 2004; Katzir, 2007; Chierchia et al., 2012)
and probabilistic (Frank & Goodman, 2012; Goodman & Frank, 2016) accounts share, whatever else
divides them, is that the strengthening is *computed on the occasion of use* from a set of
alternatives. What does the computing differs sharply, being pragmatic reasoning about the speaker for the
first, a covert exhaustification operator in the parse of that very utterance for the second, and
recursive Bayesian inference over a literal listener for the third, but all three put the
alternatives inside the trial. Ours is not computed at the occasion at all, but read off a weight.
Levinson's (2000) default generalized implicatures are the nearest neighbour, a default doing less online work than a full derivation, but a default is still keyed to an alternatives set at the
utterance, and Levinson holds a GCI to be a defeasible inference and not a part of conventional
meaning, where $\theta_u^\ast$ is a single scalar keyed to nothing but exposure frequency and, once
learned, not defeasible within the trial at all. The position should be argued rather than
inherited. It also bounds what the present
architecture can deliver: within-trial alternatives would require a mechanism that compares entries
inside a single inference pass, and Eq. (20) is not one.

## <a id="appc" name="appc"></a>Appendix C: how many utility directions an inventory needs

Text cell 5 Part A shows that $m=1$ is
degenerate and takes $m=2$ as the bare minimum. That leaves open what fixes the number in general:
whether $2$ is a fact about this inventory of three, or about the architecture. This appendix
answers that, and the answer is neither: it is a fact about the **thresholds**, and $m=2$ follows
from the scale having two ends.

Throughout, $\Lambda$ is factored out of the projections, since $\varphi_L=\Lambda\chi_y$ makes it a
common scalar. The two indicators $\chi_{\textit{no}}=\mathbb 1[\zeta>-\theta_L]$ and
$\chi_{\textit{all}}=\mathbb 1[\zeta<\theta_L]$ are reflections of each other on a grid symmetric
about $\zeta=0$.

---

### <a id="appc-1" name="appc-1"></a>1. Two readings of the criterion, which come apart

Eq. (29) says the $u$ level reaches the lexicon only through $\kappa_y=B^{\mathsf T}W\varphi_{L,y}$.
Two conditions can be asked of $B$, and they are not the same condition:

- **Separation.** The $\kappa_y$ are pairwise distinct, so no two utterances are confused at the $u$
  level. This is Part A's criterion.

- **Spanning.** $B$ spans the whole space the $\chi_y$ occupy modulo the constant, so nothing the
  lexicon can express is invisible to the $u$ level.

Spanning implies separation and not conversely. Section 7 below exhibits the gap; until then the
two agree.

### <a id="appc-2" name="appc-2"></a>2. Complementation costs nothing

$E_{\textit{some}}$ is the complement of $E_{\textit{no}}$, so $\chi_{\textit{some}}=\mathbf 1-\chi_{\textit{no}}$.
Part A drops the constant from $B$ deliberately, since a constant direction shifts $\varphi_S$
uniformly and is invisible to the read-out (Eq. 12), so $\langle\mathbf 1,b_j\rangle=0$ for every
column, and

$$E_{y'}=[0,1]\setminus E_y\quad\Longrightarrow\quad
\kappa_{y'}=B^{\mathsf T}W(\mathbf 1-\chi_y)=-\kappa_y .
\tag{C1}$$

Complementation is **negation** in projection space, not a new direction. Two consequences run
through everything below: a complementary pair costs no dimension at all, and it is the *easiest*
kind of pair to separate, since $\kappa$ and $-\kappa$ are antipodal and every functional that sees
one sees the difference doubled.

### <a id="appc-3" name="appc-3"></a>3. The count is thresholds, not utterances

With $t$ distinct cut points the scale is divided into $t+1$ intervals; every $\chi_y$ is a union of
them, so the $\chi_y$ span at most $t+1$ dimensions including the constant, hence

$$\dim\operatorname{span}\{\chi_y\}\big/\langle\mathbf 1\rangle\;\le\;t .
\tag{C2}$$

Here $t=2$, at $-\theta_L$ and $+\theta_L$, and the reason is Eq. (A1): the scale is **open**, its two
endpoints have no $\zeta$ coordinate, and *no* is anchored at $s=0$ while *all* is anchored at $s=1$.
There is one threshold per end, and **two ends give two thresholds and $m=2$**, which is the whole
of the coincidence.

$m=2$ is moreover exactly right rather than merely sufficient. Solving for what $B$ cannot see, in
coordinates on the intervals:

| cuts | intervals | dim of the unseen subspace | an unseen direction |
|---|---|---|---|
| $\{-\theta_L,+\theta_L\}$ | 3 | 1 | $[1,\,1,\,1]$ |
| $\{-\theta_L,0,+\theta_L\}$ | 4 | 2 | $[0.512,\,-0.485,\,1.000,\,0.043]$ |

At two thresholds the only thing $m=2$ misses is the constant, which the read-out misses
too, so nothing is lost. Part A's parity argument is the symmetric instance of this count: with the cuts at
$\pm\theta_L$ the two-dimensional space splits into one even and one odd direction, so a single $b$ of
definite parity spans exactly half of it.

### <a id="appc-4" name="appc-4"></a>4. A fourth utterance: the complement of *all*

Consider adding a word with $E=\{1\}$, true everywhere except $s=1$, that is *not all*. By the
pattern of Eq. (A1) it maps to $\chi=\mathbb 1[\zeta\ge\theta_L]=\mathbf 1-\chi_{\textit{all}}$, verified exactly. By
Eq. (C1) its projection is $-\kappa_{\textit{all}}$, and the four utterances land on the corners of a
rectangle about the origin:

| | $\kappa_{y,1}$ (tilt) | $\kappa_{y,2}$ (width) |
|---|---|---|
| *no* | $+1.13989$ | $-0.72098$ |
| *some* | $-1.13989$ | $+0.72098$ |
| *all* | $-1.13989$ | $-0.72098$ |
| *not all* | $+1.13989$ | $+0.72098$ |

All four are distinct, so **$m=2$ still suffices at $\lvert\mathcal Y\rvert=4$**, and the two pairs of contradictories, *no* against *some* and *all* against *not all*, sit at opposite
corners. The rank confirms that nothing was added:
$\operatorname{span}\{\chi_{\textit{no}},\chi_{\textit{some}},\chi_{\textit{all}}\}$ has rank $3$,
adjoining $\chi_{\textit{not all}}$ leaves it at $3$, and adjoining $\mathbf 1$ leaves it at $3$ again, since $\mathbf 1=\chi_{\textit{no}}+\chi_{\textit{some}}$ was already inside.
Modulo the constant the rank is $2$. The fourth word introduces no threshold, so by Eq. (C2) it cannot introduce a
dimension.

*An aside worth recording.* $E=\{1\}$ is the **O corner of the Aristotelian square**, the vertex
Horn (1972) observes is never lexicalized in a natural language. This model's verdict is that the gap
is not representational: the O corner is exactly as expressible as the other three and costs the
architecture nothing. Whatever explains the gap, nothing here supplies a reason for it.

### <a id="appc-5" name="appc-5"></a>5. Inventories of two, and the parity each demands

Dropping one utterance leaves three inventories. Each needs only **one** dimension to separate, since two points always admit a
separating functional, but *which* direction is forced, and the answer differs:

| inventory | thresholds | span mod $\mathbf 1$ | odd $b_1$ | even $b_2$ |
|---|---|---|---|---|
| *no*, *some* | 1 | 1 | $2.27977$ | $1.44190$ |
| *no*, *all* | 2 | 2 | $2.27977$ | **collapses** ($6.9\times10^{-18}$) |
| *some*, *all* | 2 | 2 | **collapses** ($-6.9\times10^{-18}$) | $1.44190$ |

Read in Part B §2's vocabulary, where $b_1$ is **tilt** and $b_2$ is **width**:

- *no* and *all* differ by *which end of the scale*, an odd difference, and want tilt.

- *some* and *all* differ by *middle versus both ends*, an even difference, and want width.

- *no* and *some* are complementary, so by Eq. (C1) either direction serves; this is the only
  inventory with no parity constraint, and also the only one whose span drops to $1$, since dropping
  *all* removes the $+\theta_L$ threshold outright.

Hence the general statement, of which Part A is the instance:

> $m\ge2$ is forced exactly when the inventory contains two pairs whose differences have **opposite
> parity**, and with symmetric thresholds that requires three utterances.

$\chi_{\textit{no}}-\chi_{\textit{all}}$ is odd and $\chi_{\textit{all}}-\chi_{\textit{some}}$ is
even, so no single $b$ of definite parity serves both. Section 4 now follows without recomputation:
*not all* adds neither a threshold nor a third parity class.

### <a id="appc-6" name="appc-6"></a>6. What an odd $m=1$ basis costs

$b=\zeta$, a monotone ramp meaning "more versus less", is the natural first choice for a single direction,
and it is odd. Under it $\kappa_{\textit{some}}=\kappa_{\textit{all}}=-1.13989$ and the two utterances
are one point at the $u$ level. Running the network at $m=1$:

| | $\lvert\Delta\varphi_u^\ast\rvert$ | $\lvert\Delta\varphi_S^\ast\rvert$ | $\lvert\Delta\mathbb E[s]\rvert$ |
|---|---|---|---|
| *some* vs *all* | $5.8\times10^{-15}$ | $4.0000$ | $0.2525$ |
| *no* vs *all* | $6.0790$ | $5.5200$ | $0.6018$ |

$\varphi_S^\ast$ still separates *some* from *all* by $4.0000$, because Eq. (15) carries
$\ell_0-\varphi_L$ directly and not only through the projection. The model is not blind; the utility
level is.

That distinction is exactly what makes the failure serious rather than cosmetic. With
$\varphi_u^\ast(\textit{some})=\varphi_u^\ast(\textit{all})$, the term $\theta_u B\varphi_u^\ast$ is
the *same field* for both, so the utility level acts on the pair purely common-mode and

$$\varphi_S^\ast(\textit{all})-\varphi_S^\ast(\textit{some})\ \text{ is independent of }\theta_u ,
\tag{C3}$$

measured to $8.9\times10^{-16}$ over $\theta_u\in\{0,1,-1,5\}$, against $3.445$ for the same contrast
at $m=2$. Under an odd one-dimensional basis $\theta_u$ becomes to $\{\textit{some},\textit{all}\}$
precisely what $\mu_u$ is to all three utterances at $m=2$ (Part B §1): a pure common-mode shift that
can move no contrast between them. The mechanism Text cell 4 Part C measures, amplification of
the $\operatorname{span}B$ component carried by $\theta_u$, is therefore unavailable for the one pair a
scalar implicature is about, at every $\theta_u$, $\Lambda$ and $\mu_u$.

$\Delta_y$ nevertheless differs between them, $+0.0300$ against $-0.1912$, and differs against the
$\theta_u$ control as well, $+0.0125$ against $+0.1276$. That is not a residual
effect surviving Eq. (C3) but the failure mode of Part B §4 again: $\Delta_y$ is a $q$-mass (Eq. 27)
taken after the softmax, so a field added identically to two differently-placed beliefs moves their
masses by different amounts. Nothing the model represents *about the relation between* the two
utterances has moved.

### <a id="appc-7" name="appc-7"></a>7. Separation is not spanning

Section 3 gives $m=t$ as the design rule, but that rule is about spanning, and separation of a
particular inventory can survive $m<t$. Adding *most*, true where $s>\tfrac12$ and so a third cut at
$\zeta=0$, raises the span from $3$ to $4$, so $m=2$ no longer spans and there is a genuinely unseen
direction beyond the constant. Yet **no exclusion set realizes it**: its interval weights
$[0.512,-0.485,1.000,0.043]$ are not a $\{-1,0,1\}$ combination, and all $14$ nonempty proper unions
of the four intervals still separate under $m=2$, with a minimum $\kappa$ distance of $0.72097$ and
no collisions.

The contrast with Part A's degeneracy is the point. At two thresholds the direction $m=1$ misses *is* realizable, being the middle interval
$\chi_{\textit{all}}-\chi_{\textit{some}}=\mathbb 1_{(-\theta_L,\theta_L)}$, an exclusion-set difference,
which is why $m=1$ actually fails. Whether $m<t$ bites is a question about the integer
lattice of exclusion sets, not about dimension alone. The margin does erode: at asymmetric cuts
$\{-4,-1,2.5\}$ the minimum separation falls to $0.13100$.

### <a id="appc-8" name="appc-8"></a>8. What this appendix does not settle

Everything above concerns separation and spanning. It does not address whether $m=2$ is *sufficient*
in the wider sense Part A leaves open. Part A's own preference for $m=2$ over a single $b$ of no
definite parity is a different argument, that one coordinate confounds tilt and width, and that
argument still favours $m=2$ in the two-utterance cases of §5 where one dimension separates. Nor does
Eq. (C2) say anything about a scale with more than one dimension, which §9.3 takes up separately.

## <a id="appd" name="appd"></a>Appendix D: why emission is exclusion

Eq. (5) fixes each entry by the states it **rules out**. Fixing it by the states it admits is the
more immediate choice, and this appendix says why the model does not take it. The reason cannot be
that the two representations give different beliefs, since Section 1 shows they do not; nor is it a
property of the two families of sets, which Section 2 shows behave symmetrically. It is a property of the
map that carries an entry into a field, and Section 3 states it.

---

### <a id="appd-1" name="appd-1"></a>1. The two representations, and what the network sees of the difference

Write $T_y\subseteq[0,1]$ for the **truth set** of an utterance, the states at which it is true, keeping $E_y$ for the exclusion set of Eq. (5). The two are complements, $E_y=[0,1]\setminus T_y$, so
on the scale of Eq. (5)

$$T_{\textit{no}}=\{0\},\qquad T_{\textit{some}}=(0,1],\qquad T_{\textit{all}}=\{1\}.
\tag{D1}$$

Read a set as the mask that keeps it: the map $[0,1]\to[0,1]$ acting as the identity on that set and
sending everything else to $0$. Its **kernel** is what the mask destroys, which is the set's
complement, so each family is the other's kernel,

$$E_y=\ker T_y,\qquad T_y=\ker E_y .
\tag{D2}$$

**No quantity the model reports separates them.** Writing the
entry as $\varphi_L=\Lambda\chi_{T_y}$ and making $g_L$ order-preserving, $g_L(\varphi_S)=\varphi_S-\ell_0$,
reproduces the model: $\varphi_u^\ast$ agrees to $5.3\times10^{-15}$, $c_y$ to $1.4\times10^{-14}$, and
$\theta_u^\ast$ is $-28.43749$ under both. Eq. (21) is untouched, since $g_L'=+I$ contributes
$\lVert v\rVert^{2}/\sigma_L$ to the Hessian exactly as $-I$ does, so every proof of §8 stands. The
settled field moves by a constant, writing $\vert_E$ and $\vert_T$ for a quantity under the two
conventions,

$$\varphi_S^\ast\big\vert_{T}=\varphi_S^\ast\big\vert_{E}
+\frac{\sigma_S\Lambda}{\sigma_L+\sigma_S}\,\mathbf 1 ,
\tag{D3}$$

$+4.000000$ at the settings of Text cell 3 §3, which Eq. (12) normalizes away: the read-out agrees to
$4.4\times10^{-16}$, and with it every statistic of Eqs. (25)–(27). The settled *states* do move:
at $\theta_u=1$, $\lVert r_S\rVert$ runs $23.19,\,18.08,\,23.10$ under $E$ against
$10.56,\,9.43,\,10.36$ under $T$. Eq. (D3) confines that movement to the constant direction, and
Section 3 turns on exactly this.

The other visible difference is the sign of $g_L'$: the lexical error reaches $\varphi_S$
excitatorily where Eq. (9) has it reach inhibitorily. Eq. (19) reads
$\tau_\varphi\dot\varphi_S=-\varepsilon_S\mp\varepsilon_L$ under the two, so the sign sits in the
return weight rather than in what the unit reads, and both are local on the terms §7 already
establishes. The choice is not settled there either.

### <a id="appd-2" name="appd-2"></a>2. As sets, $E$ and $T$ are symmetric

Complementation is an order-anti-isomorphism of the Boolean algebra of subsets of $[0,1]$
(Davey & Priestley, 2002), so every structural property of one family transfers to the other with
$\cup$ and $\cap$ exchanged and $\subseteq$ reversed. Three instances:

- **Entailment.** *all* entails *some*, which is $T_{\textit{all}}\subseteq T_{\textit{some}}$ and
  equally $E_{\textit{some}}\subseteq E_{\textit{all}}$. The remaining relations of the Aristotelian square read
  the same either way: *no* and *some* are contradictories, being disjoint and exhaustive, and *no*
  and *all* are contraries, being disjoint and not exhaustive (Horn, 1972).

- **Complementation.** Each family holds exactly one complementary pair, *no* with *some*, and one
  member whose complement falls outside it: $E_{\textit{all}}$, whose complement $\{1\}$ is the
  *not all* of Appendix C §4, and $T_{\textit{all}}$, whose complement is $[0,1)$.

- **The kernel.** By Eq. (D2) the two families close under $\ker$ to precisely the same extent:

| | closes inside the family | escapes it |
|---|---|---|
| $E$ | $\ker E_{\textit{no}}=E_{\textit{some}}$, $\ker E_{\textit{some}}=E_{\textit{no}}$ | $\ker E_{\textit{all}}=\{1\}$ |
| $T$ | $\ker T_{\textit{no}}=T_{\textit{some}}$, $\ker T_{\textit{some}}=T_{\textit{no}}$ | $\ker T_{\textit{all}}=[0,1)$ |

Two of three close in each family, and in each it is the *all* entry that escapes.

**No property of the two families as sets can prefer one of them.** That is a result rather than a
gap: it rules out every argument from the set algebra, and says where the justification has to be
found instead.

### <a id="appd-3" name="appd-3"></a>3. The asymmetry lies in the representation map

By Eq. (D3) the two conventions differ by $\Lambda\mathbf 1=\Lambda\chi_{[0,1]}$, which is the field of a **tautology**, the entry that rules out nothing. Three readings follows from this fact.

**The empty entry is the zero vector.**

| entry | | $\lVert\varphi_L\rVert$ under $E$ | under $T$ |
|---|---|---|---|
| tautology | $T=[0,1]$, $E=\varnothing$ | $0$ | $27.7128$ |
| contradiction | $T=\varnothing$, $E=[0,1]$ | $27.7128$ | $0$ |

Under $E$, "this entry carries no lexical information" and "this entry contributes no lexical field"
are one statement. Under $T$ the semantically empty entry carries the largest field the lexicon can
supply, and the entry true of nothing carries none.

**Field magnitude orders the entailment.**

| | $\lVert\varphi_L\rVert$ under $E$ | under $T$ |
|---|---|---|
| *no* | $23.9199$ | $13.9943$ |
| *some* | $13.9943$ | $23.9199$ |
| *all* | $23.9199$ | $13.9943$ |

*all* and *some* are the one entailment-comparable pair, and under $E$ the entailing member carries
the larger field where under $T$ it carries the smaller. $\Lambda$ times the root quadrature measure
of $E_y$ is not a canonical measure of informativity, so the claim is the limited one: it orders that
pair, and it orders it in opposite directions under the two conventions.

**A tautologous entry leaves the belief where it found it, as a state and not only as a belief.**
Against what the network settles on with no lexical field at all,

$$\max_k\big\lvert\varphi^\ast_{S,k}(\text{tautology})-\varphi^\ast_{S,k}(\text{no lexical field})\big\rvert
=\begin{cases}0 & \text{under }E,\\[2pt] 4.0000 & \text{under }T.\end{cases}
\tag{D4}$$

Under $E$ the two coincide by construction, a tautology having $\varphi_L=0$. Under $T$ they agree in
$q$ and differ in $\varphi_S$, and the difference is invisible only because $B\perp\mathbf 1$, Part A having dropped the constant
column, and because Eq. (12) normalizes.

**The third reading is the decisive one.** §9.1 establishes that $\varphi_S$ is not a log-density defined
up to an additive constant: $\nabla^2\mathcal F$ has no flat direction, the absolute scale is pinned,
and this scale functions as the temperature under Eq. (15). A convention whose adequacy rests on an
additive constant being invisible relies on the property §9.1 declines to grant. Under $E$ the
constant is not invisible; it is absent.

§4.2's account of $g_L$ turns on the same absent constant. The complement is $x\mapsto\mathbf 1-x$ on
indicators and $x\mapsto\ell_0-x$ on log-weights, both order-reversing involutions. An involution has
a fixed point, and $g_L$'s is $\varphi_S=\ell_0/2$, which is what Eq. (15) returns at
$\varphi_L=0,\ \theta_u=0$, so the centre of the reflection is a field the model settles on. Under $T$
the corresponding map is order-preserving, hence a translation, which has no fixed point at all, and
$\ell_0$ enters as an offset that no structure of the map locates. Eq. (9) is derived under $E$ from
where its own centre sits; under $T$ its $-\ell_0$ would have to be argued on separate grounds.

### <a id="appd-4" name="appd-4"></a>4. What this settles, and what it leaves

The above settles E-T assymetry over the representation map. Exclusion is the convention under which
$y\mapsto\varphi_L$ sends the entry that says nothing to the field that does nothing, and the
inhibitory $g_L'=-I$ is understood as a consequence of this choice.
